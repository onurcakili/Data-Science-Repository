# Python'da Diziler ve Koleksiyonlar

Python'da birden fazla değeri saklamak için koleksiyon yapıları kullanılır.  
Veri bilimine geçmeden önce `list`, `tuple`, `set` ve `dict` mantığı çok önemlidir.  

## Konu Dokümantasyonu

Python'da birden fazla değeri tek bir yapı altında saklamak için koleksiyonlar kullanılır.    
Günlük konuşmada bunlara çoğu zaman dizi denir; Python'da en sık kullanılan karşılığı `list` yapısıdır.  

Temel koleksiyon türleri:

* `list  []` : Sıralı  - Tekrarlı  - Değiştirilebilir yapı,
* `tuple ()` : Sıralı  - Tekrarlı  - Değiştirilemez yapı,
* `set   {}` : Sırasız - Tekrarsız - Değiştirilebilir yapı,
* `dict  {:}`: Sıralı  - Tekrarsız - Değştirilebilir çiftlerle çalışan yapıdır.

Modern Python sürümlerinde dictionary elemanları eklenme sırasını korur.  
Ancak dictionary yapısında değerlere sıra numarasıyla değil, anahtarlarla erişilir.

## Koleksiyonlar Neden Önemlidir?

Koleksiyonlar, gerçek hayattaki verileri program içerisinde düzenli şekilde temsil etmemizi sağlar.

Örneğin:

- Bir müşterinin yaptığı satışlar list içerisinde tutulabilir.
- Bir koordinat bilgisi tuple ile saklanabilir.
- Tekrarsız şehir veya kategori değerleri set ile bulunabilir.
- Bir müşterinin adı, yaşı ve şehri dict ile temsil edilebilir.
- Birden fazla müşteri kaydı, dictionary'lerden oluşan bir liste içerisinde tutulabilir.

Veri bilimi açısından koleksiyonlar önemlidir çünkü:

- NumPy dizileri Python listelerinin mantığı üzerine kuruludur.
- Pandas DataFrame satırları ve sütunları koleksiyon mantığıyla çalışır.
- JSON verileri çoğunlukla iç içe dict ve list yapılarından oluşur.
- API'lerden gelen veriler genellikle dictionary ve listelerle temsil edilir.
- Filtreleme, gruplama ve toplama işlemleri önce Python koleksiyonlarında öğrenilir.

## Kolay Seviye

Bu bölümde Python'daki temel dizi/koleksiyon tiplerini tek tek ele alacağız.

Python'da "array" kelimesi günlük kullanımda çoğu zaman birden fazla değeri tutan yapılar için kullanılır.  
Python tarafında bu konunun temelini şu yapılar oluşturur:

| Tür | Yazım | Sıralı mı? | Tekrar eder mi? | Değiştirilebilir mi? | Ne zaman kullanılır? |
| --- | --- | --- | --- | --- | --- |
| `list` | `[]` | Evet | Evet | Evet | Değişebilen sıralı veri listeleri |
| `tuple` | `()` | Evet | Evet | Hayır | Sabit kalması gereken çoklu değerler |
| `set` | `{}` | Hayır | Hayır | Evet | Benzersiz değerler ve küme işlemleri |
| `dict` | `{"anahtar": "değer"}` | Evet | Anahtarlar tekrar etmez | Evet | Kayıt, satır veya özellik bilgisi |
| `str` | `"metin"` | Evet | Evet | Hayır | Metin verileri |
| `range` | `range()` | Evet | Hayır | Hayır | Sayı aralıkları ve döngüler |

Veri biliminde en sık `list`, `tuple`, `set` ve `dict` ile karşılaşırız. `str` ve `range` de dizi gibi davranır; çünkü indekslenebilir veya döngü içinde gezilebilir.


````python
liste_ornegi = [70, 85, 90]
tuple_ornegi = ("İstanbul", 41.01, 28.97)
set_ornegi = {"Elektronik", "Mobilya", "Elektronik"}
dict_ornegi = {"urun": "Laptop", "fiyat": 30000}
metin_ornegi = "Python"
range_ornegi = range(1, 6)

print(type(liste_ornegi))
print(type(tuple_ornegi))
print(type(set_ornegi))
print(type(dict_ornegi))
print(type(metin_ornegi))
print(type(range_ornegi))
````

## 1. Liste (`list`)

Liste, Python'da en çok kullanılan koleksiyon tiplerinden biridir.

Liste özellikleri:

* Sıralıdır.
* Aynı değeri birden fazla kez saklayabilir.
* Değiştirilebilir.
* Farklı veri türlerini aynı anda tutabilir.
* İndeksleme ve slicing destekler.

Veri biliminde listeler; notlar, satış tutarları, müşteri yaşları, dosya adları veya küçük veri kümeleri için sık kullanılır.

Liste oluşturmak için köşeli parantez kullanılır:

```python
notlar = [70, 85, 90]
```


### Liste Metotları

Listelerde kullanılan temel metotlar:

| Metot | Açıklama |
| --- | --- |
| `append(x)` | Listenin sonuna tek eleman ekler |
| `extend(iterable)` | Listenin sonuna başka bir koleksiyonun elemanlarını ekler |
| `insert(index, x)` | Belirtilen indekse eleman ekler |
| `remove(x)` | Belirtilen değeri listeden siler |
| `pop(index)` | Belirtilen indeksteki elemanı siler ve döndürür |
| `clear()` | Listenin tüm elemanlarını siler |
| `index(x)` | Değerin ilk bulunduğu indeksi döndürür |
| `count(x)` | Değerin kaç kez geçtiğini sayar |
| `sort()` | Listeyi yerinde sıralar |
| `reverse()` | Listeyi yerinde ters çevirir |
| `copy()` | Listenin yüzeysel kopyasını oluşturur |

Not: `len()`, `sum()`, `min()`, `max()` ve `sorted()` metot değil, Python fonksiyonudur. Ama listelerle çok sık kullanılır.


````python
notlar = [70, 85, 90, 60, 95]

print(notlar[0])          # İlk eleman: indeks 0
print(notlar[-1])         # Son eleman: negatif indeks sondan başlar
print(notlar[1:4])        # 1 dahil, 4 dahil değil
print(notlar[::2])        # Baştan sona ikişer adımla ilerler
````

````python
urunler = ["Laptop", "Telefon", "Tablet"]

urunler.append("Kulaklık")          # Listenin sonuna tek eleman ekler
urunler.extend(["Mouse", "Klavye"]) # Birden fazla elemanı listenin sonuna ekler
urunler.insert(1, "Monitör")        # 1. indekse yeni eleman ekler

print(urunler)
````

````python
urunler = ["Laptop", "Telefon", "Tablet", "Telefon", "Kulaklık"]

urunler.remove("Telefon")       # İlk bulduğu "Telefon" değerini siler
silinen_urun = urunler.pop(-1)   # Son elemanı siler ve değişkene aktarır

print("Güncel liste:", urunler)
print("Silinen ürün:", silinen_urun)
print("Telefon kaç kez var?", urunler.count("Telefon"))
print("Tablet indeksi:", urunler.index("Tablet"))
````

````python
satislar = [1200, 950, 1800, 2100, 1750]

sirali_kopya = sorted(satislar)  # Orijinal listeyi bozmaz, yeni sıralı liste üretir
satislar.sort()                  # Orijinal listeyi yerinde sıralar

print("sorted sonucu:", sirali_kopya)
print("sort sonrası:", satislar)

satislar.reverse()               # Listeyi yerinde ters çevirir
print("reverse sonrası:", satislar)
````

````python
orijinal_liste = [70, 80, 90]
kopya_liste = orijinal_liste.copy()  # Yeni liste oluşturur

kopya_liste.append(100)

print("Orijinal:", orijinal_liste)
print("Kopya:", kopya_liste)
````

## 2. Tuple (`tuple`)

Tuple, sıralı fakat değiştirilemeyen bir koleksiyon tipidir.

Tuple özellikleri:

* Sıralıdır.
* Aynı değeri birden fazla kez saklayabilir.
* Değiştirilemez.
* İndeksleme ve slicing destekler.
* Listeye göre daha sabit ve güvenli veri taşımak için kullanılır.

Tuple genellikle değişmemesi gereken veriler için tercih edilir.  
Örneğin koordinat, renk kodu, tarih parçası, model ayarı veya sabit sütun bilgileri tuple ile saklanabilir.


### Tuple Metotları

Tuple değiştirilemez olduğu için liste kadar çok metodu yoktur.

| Metot | Açıklama |
| --- | --- |
| `count(x)` | Değerin tuple içinde kaç kez geçtiğini sayar |
| `index(x)` | Değerin ilk bulunduğu indeksi döndürür |

Tuple üzerinde `append`, `remove`, `sort` gibi değiştirme metotları yoktur. Çünkü tuple oluşturulduktan sonra içeriği değiştirilemez.


````python
koordinat = (41.01, 28.97)
renk = (255, 120, 80)
model_ayarlari = ("RandomForest", 100, 42)

print(koordinat[0])       # Enlem
print(koordinat[1])       # Boylam
print(renk)
print(model_ayarlari)
````

````python
kategoriler = ("Elektronik", "Mobilya", "Elektronik", "Aksesuar")

print(kategoriler.count("Elektronik"))  # Elektronik kaç kez geçiyor?
print(kategoriler.index("Mobilya"))     # Mobilya ilk olarak hangi indekste?
````

````python
musteri_kaydi = ("Ali", 32, "İstanbul")

ad, yas, sehir = musteri_kaydi          # Tuple unpacking: değerleri değişkenlere ayırabilir

print(ad)
print(yas)
print(sehir)
````

````python
sabit_sutunlar = ("musteri_id", "yas", "gelir", "segment")
listeye_cevrilmis = list(sabit_sutunlar)

listeye_cevrilmis.append("tahmin")

print("Tuple:", sabit_sutunlar)
print("Listeye çevrilmiş:", listeye_cevrilmis)
````

## 3. Set (`set`)

Set, benzersiz değerlerden oluşan koleksiyon tipidir.

Set özellikleri:

* Sırasızdır.
* Tekrar eden değerleri saklamaz.
* Değiştirilebilir.
* İndeksleme desteklemez.
* Küme işlemleri için çok kullanışlıdır.

Veri biliminde set; benzersiz kategori bulma, iki veri kümesini karşılaştırma, ortak müşterileri bulma veya hızlı üyelik kontrolü için kullanılır.


### Set Metotları

Set yapısında sık kullanılan metotlar:

| Metot | Açıklama |
| --- | --- |
| `add(x)` | Set içine tek eleman ekler |
| `update(iterable)` | Birden fazla eleman ekler |
| `remove(x)` | Elemanı siler, eleman yoksa hata verir |
| `discard(x)` | Elemanı siler, eleman yoksa hata vermez |
| `pop()` | Rastgele bir elemanı siler ve döndürür |
| `clear()` | Tüm elemanları siler |
| `union()` | İki setin birleşimini döndürür |
| `intersection()` | Ortak elemanları döndürür |
| `difference()` | İlk sette olup diğerinde olmayanları döndürür |
| `symmetric_difference()` | Sadece birinde olan elemanları döndürür |
| `issubset()` | Alt küme kontrolü yapar |
| `issuperset()` | Üst küme kontrolü yapar |
| `copy()` | Set kopyası oluşturur |


````python
kategoriler = {"Elektronik", "Mobilya", "Elektronik", "Aksesuar"}

print(kategoriler)        # Elektronik tekrarlı yazılsa bile bir kez tutulur

kategoriler.add("Kitap")
kategoriler.update(["Spor", "Kozmetik"])

print(kategoriler)
````

````python
kategoriler = {"Elektronik", "Mobilya", "Aksesuar"}

kategoriler.discard("Oyuncak")      # Eleman yoksa hata vermez
kategoriler.remove("Mobilya")       # Eleman varsa siler

print(kategoriler)
````

````python
kampanya_musterileri = {"Ali", "Ayşe", "Mehmet", "Zeynep"}
satin_alan_musteriler = {"Ayşe", "Mehmet", "Can"}

ortak_musteriler = kampanya_musterileri.intersection(satin_alan_musteriler)
sadece_kampanya_gorenler = kampanya_musterileri.difference(satin_alan_musteriler)
tum_musteriler = kampanya_musterileri.union(satin_alan_musteriler)

print("Ortak:", ortak_musteriler)
print("Kampanya gördü ama satın almadı:", sadece_kampanya_gorenler)
print("Tüm müşteriler:", tum_musteriler)
````

````python
gerekli_sutunlar = {"musteri_id", "yas", "gelir"}
gelen_sutunlar = {"musteri_id", "yas", "gelir", "sehir", "segment"}

print(gerekli_sutunlar.issubset(gelen_sutunlar))    # Gerekli sütunlar gelen veri içinde var mı?
print(gelen_sutunlar.issuperset(gerekli_sutunlar))  # Gelen veri gerekli sütunları kapsıyor mu?
````

## 4. Dictionary (`dict`)

Dictionary, anahtar-değer çiftleriyle çalışan koleksiyon tipidir.

Dict özellikleri:

* Anahtarlar benzersizdir.
* Değerler tekrar edebilir.
* Değiştirilebilir.
* Modern Python sürümlerinde eklenme sırasını korur.
* Değerlere indeksle değil, anahtar ile erişilir.

Veri biliminde dictionary çok önemlidir çünkü JSON verileri, API cevapları ve satır bazlı kayıtlar çoğunlukla dictionary mantığıyla temsil edilir.


### Dictionary Metotları

Dictionary yapısında sık kullanılan metotlar:

| Metot | Açıklama |
| --- | --- |
| `keys()` | Anahtarları döndürür |
| `values()` | Değerleri döndürür |
| `items()` | Anahtar-değer çiftlerini döndürür |
| `get(key, default)` | Anahtarı güvenli şekilde okur |
| `update(dict)` | Yeni anahtar-değer çiftleri ekler veya günceller |
| `pop(key)` | Anahtarı siler ve değerini döndürür |
| `popitem()` | Son eklenen anahtar-değer çiftini siler |
| `setdefault(key, default)` | Anahtar yoksa varsayılan değerle oluşturur |
| `copy()` | Dictionary kopyası oluşturur |
| `clear()` | Tüm elemanları siler |


````python
ogrenci = {
    "ad": "Ayşe",
    "yas": 22,
    "not": 88
}

print(ogrenci["ad"])                # Anahtar varsa doğrudan erişim yapılır
print(ogrenci.get("sehir", "Yok"))  # Anahtar yoksa hata yerine varsayılan değer ("Yok") döner

print(ogrenci.keys())
print(ogrenci.values())
print(ogrenci.items())
````

````python
urun = {
    "ad": "Laptop",
    "fiyat": 30000,
    "stok": 5
}

urun.update({"stok": 8, "kategori": "Elektronik"})  # Var olanı günceller, olmayanı ekler
silinen_fiyat = urun.pop("fiyat")                   # fiyat anahtarını siler ve değerini döndürür

print(urun)
print("Silinen fiyat:", silinen_fiyat)
````

````python
kategori_satislari = {}

kategori_satislari.setdefault("Elektronik", 0)  # Anahtar yoksa 0 ile oluşturur
kategori_satislari["Elektronik"] += 30000
kategori_satislari["Elektronik"] += 20000

kategori_satislari.setdefault("Mobilya", 0)
kategori_satislari["Mobilya"] += 7500

print(kategori_satislari)
````

## 5. Diğer Dizi Benzeri Yapılar

Python'da `str`, `range` ve `frozenset` de dizi/koleksiyon mantığıyla ilişkilidir.

### String (`str`)

Metinler karakter dizisi gibi davranır çünkü metinler karakterlerden (char) oluşmaktadır.  
İndekslenebilir ve döngüyle gezilebilir. Ancak değiştirilemez.

Sık kullanılan string metotları:

| Metot | Açıklama |
| --- | --- |
| `lower()` | Küçük harfe çevirir |
| `upper()` | Büyük harfe çevirir |
| `strip()` | Baştaki ve sondaki boşlukları siler |
| `replace()` | Metin parçasını değiştirir |
| `split()` | Metni parçalara ayırır |
| `join()` | Metin parçalarını birleştirir |
| `count()` | Bir parçanın kaç kez geçtiğini sayar |
| `find()` | Bir parçanın ilk konumunu bulur |

### Range (`range`)

Sayı aralıkları üretir. Özellikle `for` döngülerinde kullanılır. Değiştirilemez ve belleği verimli kullanır.

### Frozenset (`frozenset`)

Set'in değiştirilemeyen hâlidir. Küme işlemlerini destekler ama `add` ve `remove` gibi değiştirme metotları yoktur.


````python
metin = "  Python Veri Bilimi  "

temiz_metin = metin.strip().lower()     # Boşlukları siler, küçük harfe çevirir
kelimeler = temiz_metin.split()         # Boşluklardan parçalar

print(temiz_metin)
print(kelimeler)
print("-".join(kelimeler))              # Liste elemanlarını "-" ile birleştirir
````

````python
for sayi in range(1, 10, 2):            # başlangıç=1, bitiş=10, artış=2
    print(sayi)                         # 1, 3, 5, 7, 9 değerlerini üretir

aralik = range(1, 10, 2)
print(aralik.count(3))                  # 3 değeri aralıkta kaç kez var?
print(aralik.index(5))                  # 5 değeri aralıkta kaçıncı pozisyonda?
````

````python
sabit_kategoriler = frozenset(["Elektronik", "Mobilya", "Aksesuar"])
yeni_kategoriler = frozenset(["Elektronik", "Kitap"])

print(sabit_kategoriler.union(yeni_kategoriler))
print(sabit_kategoriler.intersection(yeni_kategoriler))
````

## Orta Seviye

Hangi koleksiyon tipinin seçileceği problemin yapısına bağlıdır.

| İhtiyaç | Uygun tür |
| --- | --- |
| Sıralı ve değişebilir veri saklamak | `list` |
| Sabit kalması gereken sıralı veri saklamak | `tuple` |
| Benzersiz değer bulmak | `set` |
| Bir kaydı alan adlarıyla temsil etmek | `dict` |
| Tablo benzeri kayıtlar oluşturmak | `list[dict]` |
| Metin temizlemek ve parçalamak | `str` |
| Döngüde sayı aralığı üretmek | `range` |

Veri biliminde en yaygın örüntülerden biri `list[dict]` yapısıdır.  
Bu yapı, her satırı dictionary olan küçük tablo gibi düşünülebilir.


````python
satis_kayitlari = [
    {"urun": "Laptop", "kategori": "Elektronik", "tutar": 30000},        # 0. index
    {"urun": "Telefon", "kategori": "Elektronik", "tutar": 20000},       # 1. index
    {"urun": "Masa", "kategori": "Mobilya", "tutar": 7500},              # 2. index
    {"urun": "Sandalye", "kategori": "Mobilya", "tutar": 2500},          # 3. index
]

# Orn. satis_kayıtlarının 0. indexindeki değerleri getirmek
satis_kayitlari[0]                      
# {'urun': 'Laptop', 'kategori': 'Elektronik', 'tutar': 30000}

# Orn. Tablonun 0. indexindeki urun adını almak için:
satis_kayitlari[0]["urun"]              
# Laptop

# Orn. Tablonun 2. indexindeki kategori adını almak için:
satis_kayitlari[2]["kategori"]          
# Elektronik
````

````python
# Tüm kayıtları getirmek 

# "Kayit" satis_kayitlari listesini tek tek gezerek tüm kayıtları getirmektedir.

for kayit in satis_kayitlari:
    print(kayit)
    
# Kayıt = 0
# Kayıt = 1
# ... olarak devam ederek aslında satis_kayitlari[0] gösterimindeki gibi ifade edilmektedir.
````

````python
# Tüm satırları gezerek "kategori" değerlerini liste içerisinde toplamak

tum_kategoriler = []

for kayit in satis_kayitlari:
    if kayit["kategori"] not in tum_kategoriler:
        tum_kategoriler.append(kayit["kategori"])


print(tum_kategoriler)
````

````python
# Bir list içerisine lambda yazarak doğrudan döngü aracılığıyla liste oluşturulabilmektedir.

elektronik_satislar = [
    kayit for kayit in satis_kayitlari
    if kayit["kategori"] == "Elektronik"
]

elektronik_satislar
````

````python
toplam_elektronik = sum(kayit["tutar"] for kayit in elektronik_satislar)

print(elektronik_satislar)
print("Elektronik toplam:", toplam_elektronik)
````

````python
satis_kayitlari = [
    {"urun": "Laptop", "kategori": "Elektronik", "tutar": 30000},
    {"urun": "Telefon", "kategori": "Elektronik", "tutar": 20000},
    {"urun": "Masa", "kategori": "Mobilya", "tutar": 7500},
    {"urun": "Sandalye", "kategori": "Mobilya", "tutar": 2500},
]

kategori_toplamlari = {}

for kayit in satis_kayitlari:
    kategori = kayit["kategori"]
    tutar = kayit["tutar"]

    kategori_toplamlari.setdefault(kategori, 0)  # Kategori yoksa 0 ile başlat
    kategori_toplamlari[kategori] += tutar       # Kategori toplamına tutarı ekle

print(kategori_toplamlari)
````

## İleri Seviye

Koleksiyonlar birlikte kullanıldığında gerçek veri problemlerini çözebiliriz.

Örneğin:

* `list`: Kayıtları sırayla saklar.
* `dict`: Her kaydın alanlarını temsil eder.
* `set`: Benzersiz kategori veya şehir bulur.
* `tuple`: Değişmemesi gereken özet değerleri taşır.
* `Counter`: Tekrar eden değerleri sayar.
* `defaultdict`: Gruplama işlemlerini kolaylaştırır.

Bu yapılar Pandas'a geçmeden önce veri manipülasyon mantığını öğrenmek için çok değerlidir.


````python
from collections import Counter

sehirler = ["İstanbul", "Ankara", "İstanbul", "İzmir", "Ankara", "İstanbul"]

sehir_sayilari = Counter(sehirler)       # Her şehrin kaç kez geçtiğini sayar

print(sehir_sayilari)
print(sehir_sayilari.most_common(2))     # En sık geçen 2 şehri döndürür
````

````python
from collections import defaultdict

satis_kayitlari = [
    {"sehir": "İstanbul", "kategori": "Elektronik", "tutar": 30000},
    {"sehir": "Ankara", "kategori": "Elektronik", "tutar": 20000},
    {"sehir": "İstanbul", "kategori": "Mobilya", "tutar": 7500},
    {"sehir": "İzmir", "kategori": "Mobilya", "tutar": 9000},
]

sehir_toplamlari = defaultdict(int)      # Yeni şehir geldiğinde varsayılan 0 üretir

for kayit in satis_kayitlari:
    sehir_toplamlari[kayit["sehir"]] += kayit["tutar"]

print(dict(sehir_toplamlari))
````

## Mini Alıştırma ve Özet

Bu bölümde Python'daki temel dizi/koleksiyon tiplerini daha detaylı öğrendik.

Hatırlanması gerekenler:

* `list`: En esnek ve en sık kullanılan sıralı koleksiyon tipidir.
* `tuple`: Değişmemesi gereken sıralı veriler için uygundur.
* `set`: Benzersiz değerler ve küme işlemleri için kullanılır.
* `dict`: Anahtar-değer yapısıyla kayıt temsil eder.
* `str`: Metin verisidir ama dizi gibi indekslenebilir.
* `range`: Döngüler için sayı aralığı üretir.
* `frozenset`: Değiştirilemeyen set yapısıdır.


````python
musteriler = [
    {"ad": "Ali", "yas": 17, "sehir": "İstanbul", "harcama": 1200},
    {"ad": "Ayşe", "yas": 24, "sehir": "Ankara", "harcama": 2500},
    {"ad": "Mehmet", "yas": 31, "sehir": "İstanbul", "harcama": 1800},
    {"ad": "Zeynep", "yas": 16, "sehir": "İzmir", "harcama": 900},
]

yetiskinler = [musteri for musteri in musteriler if musteri["yas"] >= 18]
yetiskin_toplam_harcama= sum(musteri["harcama"] for musteri in yetiskinler)
yetiskin_ortalama_harcama = yetiskin_toplam_harcama / len(yetiskinler)
benzersiz_sehirler = {musteri["sehir"] for musteri in musteriler}


ozet = ("yetiskin_sayisi", len(yetiskinler), "ortalama_harcama",yetiskin_ortalama_harcama )

print("Yetişkin müşteriler:", yetiskinler)
print("Yetişkin Toplam Harcama", yetiskin_toplam_harcama)
print("Yetişkin Ortalama Harcama", yetiskin_ortalama_harcama)
print("Benzersiz şehirler:", benzersiz_sehirler)
print("Özet tuple:", ozet)
````

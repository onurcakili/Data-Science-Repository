# Python Veri Türleri (Data Types)
 
Veri türü, bir değerin hangi işlemlerde kullanılabileceğini belirler.  
Veri bilimi açısından bu konu çok kritiktir; çünkü sayısal görünen bir değer metin olarak saklanıyorsa matematiksel analiz yapılamaz.

## Konu Dokümantasyonu

Veri türü, bir değerin program tarafından nasıl yorumlanacağını belirler.  
Python için `"10"` ile `10` aynı şey değildir. İlki metindir ise ikincisi sayıdır.

Veri türlerini anlamak, hatasız program yazmanın temelidir çünkü her veri türünün desteklediği işlemler farklıdır.  
Sayılarla matematik yapılabilir, metinlerle birleştirme veya temizleme yapılabilir, mantıksal değerlerle karar mekanizmaları kurulabilir.

Temel veri türleri:

* `int`: Tam sayılar
* `str`: Metinler
* `float`: Ondalıklı sayılar
* `bool`: `True` veya `False`
* `NoneType`: Henüz değer olmadığını anlatan `None`

Veri biliminde veri türleri daha da kritik hâle gelir.  
Yaş, fiyat, satış adedi gibi alanlar metin olarak gelirse hesaplama yapılamaz.  
Tarih alanı yanlış türdeyse zaman analizi bozulur. Bu yüzden analizden önce veri türleri kontrol edilir ve gerekirse dönüştürülür.


Temel veri türleri olarak : 

Python'da `type()` fonksiyonu bir değişkenin hangi türde olduğunu gösterir. Bu kontrol özellikle yeni veriyle çalışırken önemlidir.

Örneğin   
`yas = 26`   tam sayı,  
`boy = 1.78` ondalıklı sayı,   
`isim = "Onur"` metin,   
`mezun_mu = True` mantıksal değerdir.   
`None` ise değer yokluğunu temsil eder; bu, `0` veya boş metinle aynı şey değildir.

Temel veri türlerinin yanı sıra diziler (arrays) için de daha detaylı bir bölüm bir sonraki dosyada yer almaktadır.

```python
yas = 26                         # int: tam sayı
boy = 1.78                       # float: ondalıklı sayı
isim = "Onur"                    # str: metin
mezun_mu = True                  # bool: mantıksal değer
ikinci_telefon = None            # None: şu an değer yok

print(type(yas))
print(type(boy))
print(type(isim))
print(type(mezun_mu))
print(type(ikinci_telefon))
```

```python
sayi1 = 10
sayi2 = 3

print(sayi1 + sayi2)             # Toplama
print(sayi1 - sayi2)             # Çıkarma
print(sayi1 / sayi2)             # Normal bölme sonucu float döner
print(sayi1 // sayi2)            # Tam böler, küsuratı atar
print(sayi1 % sayi2)             # Direkt Kalanı verir
print(sayi1 ** sayi2)            # Üs alma: 10 üzeri 3
```

Tür dönüşümü ve formatlama İşlemleri

Dışarıdan gelen veriler çoğu zaman metin olarak gelir.  
Kullanıcıdan alınan input, CSV dosyasından okunan değer veya API'den gelen bazı alanlar sayı gibi görünse bile `str` olabilir.  
Matematiksel işlem yapmadan önce `int()` veya `float()` ile dönüştürme gerekebilir.

Ayrıca sonuçları okunabilir göstermek için f-string formatlama kullanılır.  
Para, oran ve rapor çıktılarında formatlama hem okunabilirliği hem profesyonelliği artırır.


```python
yas_metin = "25"
gelir_metin = "12000"

yas = int(yas_metin)                 # "26" metnini tam sayıya çevirir
gelir = float(gelir_metin)           # "18500" metnini ondalıklı sayıya çevirir

print(yas + 1)
print(gelir * 2)
```

```python
urun_adi = "Laptop"
fiyat = 32_500.5
stok_adedi = 7

bilgi = f"{urun_adi} fiyatı {fiyat:,.2f} TL, stok: {stok_adedi}"  

# :, sayının binlik basamaklarını ayırmak için kullanılır
# :.2f sayının virgülden sonra 2 basamaklı almasını sağlar

print(bilgi)
```

```python
# isdigit() metnin sadece rakamlardan oluşup oluşmadığını kontrol eder

deger = "125"

if deger.isdigit():                 
    sayi = int(deger)
    print(sayi * 2)
else:
    print("Sayısal olmayan değer var.")


# "125" ifadesi metin ya da harf içermediği için doğru çalışmaktadır
```

```python
value = "125"

if type(value) == int:                 # type() metnin tipini kontrol eder
    print("Sayısal değer")
else:
    print("Sayısal olmayan değer var.")
    

# type kontrolü yaparak işlem yapmak daha güvenlidir çünkü isdigit() metodu sadece string tipindeki değişkenler için çalışır.
```

Python'da bazı değerler koşul içinde otomatik olarak `False` kabul edilir.  
Boş liste, boş metin, `0`, `None` gibi değerler buna örnektir.  
Bu davranış pratik olsa da neyi kontrol ettiğimizi bilmek gerekir.

```python
# Boş metin False kabul edilir

kullanici_adi = ""

if kullanici_adi:                  
    print("Kullanıcı adı var.")
else:
    print("Kullanıcı adı boş.")
```

```python
# Boş liste False kabul edildiği için not ile kontrol edilir  

sepet = []

if not sepet:   # True değil mi? gibi düşünbebiliriz. (!=)    
    print("Sepet boş")
else:
    print("Sepet Dolu")
    

# 
```

Finans gibi hassas hesaplarda `float` küçük yuvarlama farkları oluşturabilir çünkü ikili sistemde (binary) tutulmaktadırlar.
Örn. Çıktılar bazen şu şekilde olabilir : 0.30000000000000004  

Bu yüzden finansal hesaplarda `Decimal` kullanmak daha doğru olabilir.

```python
from decimal import Decimal

fiyat = Decimal("19.99")           # Para hesaplarında metinden Decimal oluşturmak daha güvenlidir
adet = Decimal("3")
kdv_orani = Decimal("0.20")

ara_toplam = fiyat * adet
genel_toplam = ara_toplam + (ara_toplam * kdv_orani)

print(genel_toplam)
```

## Mini Alıştırma ve Özet

Hatırlanması gerekenler:

* Sayı gibi görünen her değer sayı değildir.

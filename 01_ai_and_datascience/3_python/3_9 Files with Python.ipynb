# Python'da Dosya İşlemleri

Dosya işlemleri, Python'ın dış dünyadaki verilerle çalışmasını sağlar. Veri bilimi projelerinde ham veriyi okumak, temizlenmiş çıktıyı kaydetmek ve rapor üretmek için dosya işlemleri gerekir.

Öğrenme akışı:

* **Kolay seviye:** Metin dosyası yazma ve okuma
* **Orta seviye:** `with`, `pathlib`, CSV yazma/okuma
* **İleri seviye:** `DictReader`, klasör tarama, güvenli okuma

Not: Örnekler bulunduğun notebook klasöründe küçük çalışma dosyaları oluşturabilir.


## Konu Dokümantasyonu

Dosya işlemleri, programın kalıcı veriyle çalışmasını sağlar. Program kapandığında bellekteki değişkenler kaybolur; fakat dosyaya yazılan veriler daha sonra tekrar okunabilir.

Temel dosya işlemleri:

* Dosya oluşturma
* Dosyaya yazma
* Dosyadan okuma
* Dosyanın sonuna ekleme
* CSV gibi tablo formatlarını işleme
* Dosya bulunamadığında hata yönetimi

Python'da dosya yolları için `pathlib.Path` kullanmak okunabilirliği artırır. `with` bloğu, dosyayı işlem sonunda otomatik kapattığı için güvenli bir kullanımdır.

Veri biliminde dosya işlemleri ham veriyi okumak, temizlenmiş veriyi kaydetmek, rapor oluşturmak ve ara çıktıları saklamak için kullanılır.


## Kolay Seviye

Bu seviyede metin dosyası yazma ve okuma işlemlerini göreceğiz.

`write_text()` dosyaya metin yazar. Dosya yoksa oluşturur, varsa içeriğini yeniden yazar. `read_text()` dosyanın tamamını okur.

Türkçe karakterlerde sorun yaşamamak için `encoding="utf-8"` kullanmak iyi bir alışkanlıktır.


```python
from pathlib import Path

dosya_yolu = Path("ornek_notlar.txt")
dosya_yolu.write_text("Python dosya işlemleri\n", encoding="utf-8")
# write_text dosya yoksa oluşturur, varsa içeriği yeniden yazar

icerik = dosya_yolu.read_text(encoding="utf-8")
print(icerik)
```

```python
from pathlib import Path

dosya_yolu = Path("ornek_notlar.txt")

with dosya_yolu.open("a", encoding="utf-8") as dosya:
    dosya.write("Yeni satır eklendi.\n")      # a modu dosyanın sonuna ekleme yapar

print(dosya_yolu.read_text(encoding="utf-8"))
```

## Orta Seviye

Bu seviyede `with` bloğu ve CSV dosyalarıyla çalışacağız.

`with`, dosyanın işlem bitince otomatik kapanmasını sağlar. CSV dosyaları satır ve sütunlardan oluşur; veri bilimi projelerinde en sık karşılaşılan basit veri formatlarından biridir.

CSV okurken sayısal değerler çoğu zaman metin olarak gelir. Bu yüzden hesaplama yapmadan önce `int()` veya `float()` dönüşümü gerekebilir.


```python
import csv
from pathlib import Path

csv_yolu = Path("satislar.csv")

with csv_yolu.open("w", encoding="utf-8", newline="") as dosya:
    yazici = csv.writer(dosya)
    yazici.writerow(["urun", "kategori", "fiyat"])  # Başlık satırı
    yazici.writerow(["Laptop", "Elektronik", 30000])
    yazici.writerow(["Telefon", "Elektronik", 20000])
    yazici.writerow(["Masa", "Mobilya", 7500])

print(csv_yolu.read_text(encoding="utf-8"))
```

```python
import csv
from pathlib import Path

csv_yolu = Path("satislar.csv")
toplam_fiyat = 0

with csv_yolu.open("r", encoding="utf-8", newline="") as dosya:
    okuyucu = csv.reader(dosya)
    baslik = next(okuyucu)                 # İlk satır başlık olduğu için ayrıca alınır

    for satir in okuyucu:
        fiyat = int(satir[2])              # 3. sütundaki fiyat metinden tam sayıya çevrilir
        toplam_fiyat += fiyat

print("Başlık:", baslik)
print("Toplam fiyat:", toplam_fiyat)
```

## İleri Seviye

Bu seviyede CSV dosyalarını daha okunabilir şekilde işlemeyi göreceğiz.

`csv.DictReader`, satırlara sütun adıyla erişmemizi sağlar. Bu yöntem indeksle erişmeye göre daha güvenlidir. Örneğin `satir["fiyat"]`, `satir[2]` ifadesinden daha anlaşılırdır.

Ayrıca `glob()` ile klasördeki belirli uzantıdaki dosyalar listelenebilir. Bu, çok sayıda dosya üzerinde işlem yapılacak projelerde faydalıdır.


```python
import csv
from pathlib import Path

csv_yolu = Path("satislar.csv")
elektronik_toplam = 0

with csv_yolu.open("r", encoding="utf-8", newline="") as dosya:
    okuyucu = csv.DictReader(dosya)

    for satir in okuyucu:
        if satir["kategori"] == "Elektronik":
            elektronik_toplam += int(satir["fiyat"])

print("Elektronik toplam:", elektronik_toplam)
```

```python
from pathlib import Path

klasor = Path(".")
metin_dosyalari = list(klasor.glob("*.txt"))    # Bulunulan klasördeki .txt dosyalarını listeler

for dosya_yolu in metin_dosyalari:
    print(dosya_yolu.name)
```

```python
from pathlib import Path

dosya_yolu = Path("olmayan_dosya.txt")

try:
    icerik = dosya_yolu.read_text(encoding="utf-8")
except FileNotFoundError:
    print("Dosya bulunamadı:", dosya_yolu)
else:
    print(icerik)
```

## Mini Alıştırma ve Özet

Bu bölümde Python ile dosya okuma ve yazma işlemlerini öğrendik.

Hatırlanması gerekenler:

* Dosyalar veriyi kalıcı hâle getirir.
* `Path` dosya yollarıyla çalışmayı kolaylaştırır.
* `with` dosyayı güvenli şekilde açar ve kapatır.
* CSV tablo formatında veri saklar.
* Okunan sayısal değerler gerekirse dönüştürülmelidir.
* Dosya işlemlerinde hata yönetimi önemlidir.

Şimdi CSV'den kategori bazlı rapor üretip metin dosyasına yazalım.


```python
import csv
from pathlib import Path

rapor_yolu = Path("kategori_raporu.txt")
csv_yolu = Path("satislar.csv")

kategori_toplamlari = {}

with csv_yolu.open("r", encoding="utf-8", newline="") as dosya:
    okuyucu = csv.DictReader(dosya)

    for satir in okuyucu:
        kategori = satir["kategori"]
        fiyat = int(satir["fiyat"])
        kategori_toplamlari[kategori] = kategori_toplamlari.get(kategori, 0) + fiyat

rapor_satirlari = [
    f"{kategori}: {toplam} TL"
    for kategori, toplam in kategori_toplamlari.items()
]

rapor_yolu.write_text("\n".join(rapor_satirlari), encoding="utf-8")
print(rapor_yolu.read_text(encoding="utf-8"))
```

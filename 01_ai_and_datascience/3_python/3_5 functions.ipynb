# Python'da Fonksiyonlar

Fonksiyonlar, tekrar eden işlemleri tek bir isim altında toplar. Veri bilimi çalışmalarında temizleme, dönüştürme, hesaplama ve raporlama adımlarını fonksiyonlaştırmak kodu çok daha anlaşılır yapar.

Öğrenme akışı:

* **Kolay seviye:** Fonksiyon tanımlama, parametre ve `return`
* **Orta seviye:** Varsayılan değer, keyword argument, docstring
* **İleri seviye:** Type hint, `*args`, `**kwargs`, küçük veri işleme fonksiyonları


## Konu Dokümantasyonu

Fonksiyon, belirli bir işi yapan ve gerektiğinde tekrar çağrılabilen kod bloğudur. Fonksiyonlar kod tekrarını azaltır ve programı küçük, anlaşılır parçalara böler.

Bir fonksiyonun temel parçaları:

* Fonksiyon adı
* Parametreler
* Fonksiyon gövdesi
* Opsiyonel olarak `return` değeri

`print()` sonucu ekrana yazar, `return` ise sonucu program içinde kullanılabilir hâle getirir. Bu ayrım çok önemlidir.

Veri biliminde fonksiyonlar veri temizleme, özellik üretme, metrik hesaplama, rapor hazırlama ve model değerlendirme gibi tekrar eden işlemleri düzenlemek için kullanılır.


## Kolay Seviye

Bu seviyede fonksiyon tanımlamayı ve çağırmayı öğreneceğiz.

Fonksiyon `def` anahtar kelimesi ile tanımlanır. Tanımlanan fonksiyon kendiliğinden çalışmaz; fonksiyonun adını parantez ile yazarak çağırmamız gerekir.

Parametre almayan fonksiyonlar sabit bir işi yapar. Parametre alan fonksiyonlar ise dışarıdan gelen değerlere göre farklı sonuç üretebilir.


```python
def selamla():
    print("Merhaba Python")      # Fonksiyon çağrılınca bu satır çalışır


selamla()                        # Fonksiyon burada çağrılır
```

```python
def topla(sayi1, sayi2):
    return sayi1 + sayi2         # return sonucu dışarı verir


sonuc = topla(10, 20)            # Dönen değer sonuc değişkenine aktarılır
print(sonuc)
```

## Orta Seviye

Bu seviyede fonksiyonları daha kullanışlı hâle getiren yapıları göreceğiz.

Varsayılan parametre, fonksiyona değer gönderilmediğinde kullanılacak değeri belirler. Keyword argument ise parametre adını yazarak değer göndermemizi sağlar; bu da fonksiyon çağrısını daha okunabilir yapar.

Docstring, fonksiyonun ne yaptığını açıklayan metindir. Özellikle eğitim materyali ve ekip çalışmalarında fonksiyonların anlaşılmasını kolaylaştırır.


```python
def ortalama_hesapla(degerler):
    """Bir sayı listesinin ortalamasını hesaplar."""
    toplam = sum(degerler)
    adet = len(degerler)
    return toplam / adet


notlar = [70, 85, 90, 60]
print(ortalama_hesapla(notlar))
```

```python
def indirimli_fiyat_hesapla(fiyat, indirim_orani=0.10):
    # indirim_orani gönderilmezse varsayılan olarak %10 kullanılır
    return fiyat - (fiyat * indirim_orani)


print(indirimli_fiyat_hesapla(1000))
print(indirimli_fiyat_hesapla(1000, indirim_orani=0.25))  # keyword argument okunabilirliği artırır
```

```python
def metni_temizle(metin):
    temiz_metin = metin.strip()       # Baştaki ve sondaki boşlukları siler
    temiz_metin = temiz_metin.lower() # Büyük harfleri küçük harfe çevirir
    return temiz_metin


print(metni_temizle("  PYTHON Veri Bilimi  "))
```

## İleri Seviye

Bu seviyede aktif kullanılan daha pratik fonksiyon özelliklerine bakacağız.

Type hint, fonksiyonun hangi türde parametre beklediğini ve hangi türde sonuç döndürdüğünü gösterir. Python bunu zorunlu kılmaz ama kod okunabilirliğini artırır.

`*args`, değişken sayıda konumsal argüman almak için kullanılır. `**kwargs`, değişken sayıda anahtar-değer argümanı almak için kullanılır. Bu yapılar özellikle esnek raporlama ve yardımcı fonksiyonlarda karşımıza çıkar.


```python
def yuzde_degisim(eski_deger: float, yeni_deger: float) -> float:
    if eski_deger == 0:
        return 0                  # Sıfıra bölme hatasını engeller
    return ((yeni_deger - eski_deger) / eski_deger) * 100


print(yuzde_degisim(100, 125))
```

```python
def ortalama(*degerler):
    # *degerler fonksiyona gönderilen tüm sayıları tuple olarak toplar
    return sum(degerler) / len(degerler)


print(ortalama(10, 20, 30))
print(ortalama(5, 15, 25, 35))
```

```python
def rapor_yazdir(**bilgiler):
    # **bilgiler anahtar-değer çiftlerini dictionary olarak toplar
    for anahtar, deger in bilgiler.items():
        print(f"{anahtar}: {deger}")


rapor_yazdir(ad="Aylık Satış", toplam=125000, ortalama=2500)
```

## Mini Alıştırma ve Özet

Bu bölümde fonksiyonlarla kodu tekrar kullanılabilir parçalara ayırmayı öğrendik.

Hatırlanması gerekenler:

* Fonksiyonlar `def` ile tanımlanır.
* Parametreler fonksiyona dışarıdan değer taşır.
* `return` sonucu programın geri kalanına verir.
* Varsayılan parametreler esneklik sağlar.
* Fonksiyonlar veri işleme adımlarını sadeleştirir.

Şimdi satış verisini temizleyen ve özetleyen küçük bir fonksiyon yazalım.


```python
def satis_ozeti_hazirla(satislar):
    temiz_satislar = [satis for satis in satislar if satis >= 0]  # Negatif hatalı değerleri çıkarır

    return {
        "adet": len(temiz_satislar),
        "toplam": sum(temiz_satislar),
        "ortalama": sum(temiz_satislar) / len(temiz_satislar),
        "maksimum": max(temiz_satislar),
    }


ozet = satis_ozeti_hazirla([1200, 950, -1, 1800, 2100])
print(ozet)
```

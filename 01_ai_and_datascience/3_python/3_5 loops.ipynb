# Python'da Döngüler

Döngüler, tekrar eden işlemleri yönetmek için kullanılır. Veri biliminde satır gezmek, liste dönüştürmek, dosya işlemek ve farklı senaryoları denemek için sık kullanılır.

Öğrenme akışı:

* **Kolay seviye:** `for`, `range`, `while`
* **Orta seviye:** `enumerate`, `zip`, `break`, `continue`, list comprehension
* **İleri seviye:** İç içe döngüler, gruplama, `any` ve `all`


## Konu Dokümantasyonu

Döngüler, aynı işlemi birden fazla kez yapmak için kullanılır. Eğer bir liste içindeki her elemanı kontrol etmek, sayıları belirli bir aralıkta gezmek veya satır satır veri işlemek istiyorsak döngü yazarız.

Python'da iki temel döngü vardır:

* `for`: Bir koleksiyon veya aralık üzerinde dolaşır.
* `while`: Bir koşul doğru olduğu sürece çalışır.

`for` döngüsü genellikle veri üzerinde gezinmek için kullanılır. `while` döngüsü ise tekrar sayısı baştan net değilse tercih edilir.

Veri bilimi çalışmalarında döngüler veri temizleme, dosya okuma, küçük veri yapıları üzerinde toplama/filtreleme ve model denemelerini tekrarlama gibi işlerde kullanılır. Büyük tablolarda ise çoğu zaman Pandas işlemleri tercih edilir.


## Kolay Seviye

Bu seviyede `for`, `range()` ve `while` mantığını öğreneceğiz.

`range(başlangıç, bitiş, artış)` belirli bir sayı dizisi üretir. Bitiş değeri dahil değildir. Bu davranış Python'da çok yaygındır ve slicing mantığında da karşımıza çıkar.

`while` döngüsünde koşulun bir noktada yanlış hâle gelmesi gerekir. Aksi hâlde sonsuz döngü oluşabilir.


```python
for x in range(1, 10, 2):       # range(başlangıç=1, bitiş=10, artış=2)
    print(x)                    # 1, 3, 5, 7, 9 yazdırılır; 10 bitiş olduğu için dahil değildir
```

```python
sayac = 1

while sayac <= 5:               # Koşul True olduğu sürece döngü çalışır
    print(sayac)
    sayac += 1                  # Sayaç artmazsa sonsuz döngü oluşabilir
```

## Orta Seviye

Bu seviyede döngüleri koleksiyonlarla birlikte daha pratik kullanacağız.

`enumerate()` hem sıra numarası hem değer almak için kullanılır. `zip()` birden fazla listeyi aynı anda dolaştırır. `continue`, döngünün o anki turunu atlar. List comprehension ise basit döngüleri tek satırda yazmanın okunabilir yoludur.


```python
satislar = [1200, 950, 1800, 2100]

for indeks, satis in enumerate(satislar, start=1):
    # start=1 raporlama için sıra numarasını 1'den başlatır
    print(f"{indeks}. satış: {satis} TL")
```

```python
urunler = ["Laptop", "Telefon", "Tablet"]
fiyatlar = [30000, 20000, 12000]

for urun, fiyat in zip(urunler, fiyatlar):
    # zip aynı sıradaki elemanları eşleştirir
    print(f"{urun}: {fiyat} TL")
```

```python
notlar = [45, 70, 88, 30, 95]
gecen_notlar = []

for not_degeri in notlar:
    if not_degeri < 50:
        continue                 # 50 altındaki notları atla, döngünün sonraki turuna geç
    gecen_notlar.append(not_degeri)

print(gecen_notlar)
```

```python
satislar = [1200, 950, 1800, 2100]

kdv_dahil_satislar = [satis * 1.20 for satis in satislar]
# Liste üreteci: her satis değerini gez, 1.20 ile çarp, yeni listeye koy

print(kdv_dahil_satislar)
```

## İleri Seviye

Bu seviyede döngüleri küçük veri gruplama işlemlerinde kullanacağız.

Dictionary ile toplam tutmak, kategori bazlı hesaplamaların temelidir. Bir kategori ilk kez görülüyorsa başlangıç değeri oluşturulur, sonra aynı kategori geldikçe toplam artırılır.

`any()` ve `all()` fonksiyonları da döngü mantığını daha okunabilir hâle getirir. `all()` tüm koşullar doğru mu diye bakar, `any()` en az bir koşul doğru mu diye bakar.


```python
satis_kayitlari = [
    {"kategori": "Elektronik", "tutar": 30000},
    {"kategori": "Elektronik", "tutar": 20000},
    {"kategori": "Mobilya", "tutar": 7500},
    {"kategori": "Mobilya", "tutar": 2500},
]

kategori_toplamlari = {}

for kayit in satis_kayitlari:
    kategori = kayit["kategori"]
    tutar = kayit["tutar"]

    if kategori not in kategori_toplamlari:
        kategori_toplamlari[kategori] = 0       # Kategori ilk kez görülüyorsa başlangıç değeri oluştur

    kategori_toplamlari[kategori] += tutar      # Mevcut toplamın üzerine yeni tutarı ekle

print(kategori_toplamlari)
```

```python
notlar = [70, 85, 90, 60]

tumu_gecti_mi = all(not_degeri >= 50 for not_degeri in notlar)
# all: tüm koşullar True ise True döner

yuksek_not_var_mi = any(not_degeri >= 90 for not_degeri in notlar)
# any: en az bir koşul True ise True döner

print("Tümü geçti mi?", tumu_gecti_mi)
print("90 ve üzeri not var mı?", yuksek_not_var_mi)
```

## Mini Alıştırma ve Özet

Bu bölümde tekrar eden işlemleri döngülerle yönetmeyi öğrendik.

Hatırlanması gerekenler:

* `for` koleksiyonlar üzerinde dolaşır.
* `range()` sayı aralığı üretir.
* `while` koşul doğru oldukça çalışır.
* `enumerate()` indeks ve değeri birlikte verir.
* `zip()` listeleri eşleştirerek dolaştırır.
* Döngüler veri temizleme ve küçük analizlerde sık kullanılır.

Şimdi hatalı satış değerlerini atlayıp toplam ve ortalama hesaplayalım.


```python
satislar = [1200, -1, 950, 0, 1800, 2100]
temiz_satislar = []

for satis in satislar:
    if satis <= 0:
        continue                 # 0 ve negatif değerleri analiz dışı bırak
    temiz_satislar.append(satis)

toplam = sum(temiz_satislar)
ortalama = toplam / len(temiz_satislar)

print("Temiz satışlar:", temiz_satislar)
print("Toplam:", toplam)
print("Ortalama:", ortalama)
```

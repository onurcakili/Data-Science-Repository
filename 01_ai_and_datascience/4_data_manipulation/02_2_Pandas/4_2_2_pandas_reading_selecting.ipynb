# Pandas ile Veri Okuma ve Seçme

Veri analizinde ilk adımlardan biri veriyi okumak ve doğru satır/sütunları seçmektir. Bu notebook'ta `read_csv`, sütun seçimi, `loc`, `iloc` ve temel filtreleme konularını işleyeceğiz.


## Konu Dokümantasyonu

Pandas birçok kaynaktan veri okuyabilir:

* CSV
* Excel
* JSON
* SQL
* Parquet

Bu notebook'ta CSV mantığını göstermek için `StringIO` kullanacağız. Gerçek projede `pd.read_csv("dosya.csv")` şeklinde dosya yolu verilir.

Satır ve sütun seçimi için iki temel araç vardır:

* `loc`: Etiket bazlı seçim yapar.
* `iloc`: Pozisyon bazlı seçim yapar.


## Kolay Seviye

CSV verisini okuyup DataFrame oluşturalım.


```python
import pandas as pd
from io import StringIO

csv_verisi = StringIO("""urun,kategori,fiyat,stok
Laptop,Elektronik,30000,5
Telefon,Elektronik,20000,8
Tablet,Elektronik,12000,12
Kulaklık,Aksesuar,2500,30
""")

df = pd.read_csv(csv_verisi)

print(df)
```

```python
import pandas as pd
from io import StringIO

csv_verisi = StringIO("""urun,kategori,fiyat,stok
Laptop,Elektronik,30000,5
Telefon,Elektronik,20000,8
Tablet,Elektronik,12000,12
Kulaklık,Aksesuar,2500,30
""")

df = pd.read_csv(csv_verisi)

print(df["urun"])                 # Tek sütun Series döner
print(df[["urun", "fiyat"]])      # Birden fazla sütun DataFrame döner
```

## Orta Seviye

`loc` ve `iloc` seçim mantığını öğrenmek Pandas için çok önemlidir.

* `loc[satır_etiketi, sütun_etiketi]`
* `iloc[satır_pozisyonu, sütun_pozisyonu]`

Pozisyon sayıları `0` ile başlar.


```python
import pandas as pd
from io import StringIO

csv_verisi = StringIO("""urun,kategori,fiyat,stok
Laptop,Elektronik,30000,5
Telefon,Elektronik,20000,8
Tablet,Elektronik,12000,12
Kulaklık,Aksesuar,2500,30
""")

df = pd.read_csv(csv_verisi)

print(df.loc[0, "urun"])          # 0 etiketli satırın urun sütunu
print(df.loc[0:2, ["urun", "fiyat"]])
print(df.iloc[0, 0])              # 1. satır, 1. sütun
print(df.iloc[:2, :3])            # İlk 2 satır, ilk 3 sütun
```

```python
import pandas as pd
from io import StringIO

csv_verisi = StringIO("""urun,kategori,fiyat,stok
Laptop,Elektronik,30000,5
Telefon,Elektronik,20000,8
Tablet,Elektronik,12000,12
Kulaklık,Aksesuar,2500,30
""")

df = pd.read_csv(csv_verisi)

pahali_urunler = df[df["fiyat"] >= 10000]
# Koşulu sağlayan satırlar seçilir

print(pahali_urunler)
```

## İleri Seviye

`query()` yöntemi bazı filtreleri daha okunabilir yazmayı sağlar. Sütun adları uygunsa koşullar metin olarak yazılabilir.

Ayrıca `isin()` belirli değerlerden birine sahip satırları seçmek için kullanılır.


```python
import pandas as pd
from io import StringIO

csv_verisi = StringIO("""urun,kategori,fiyat,stok
Laptop,Elektronik,30000,5
Telefon,Elektronik,20000,8
Tablet,Elektronik,12000,12
Kulaklık,Aksesuar,2500,30
Masa,Mobilya,7500,6
""")

df = pd.read_csv(csv_verisi)

sonuc = df.query("fiyat >= 10000 and stok >= 8")
print(sonuc)
```

```python
import pandas as pd
from io import StringIO

csv_verisi = StringIO("""urun,kategori,fiyat,stok
Laptop,Elektronik,30000,5
Telefon,Elektronik,20000,8
Tablet,Elektronik,12000,12
Kulaklık,Aksesuar,2500,30
Masa,Mobilya,7500,6
""")

df = pd.read_csv(csv_verisi)

secili_kategoriler = ["Elektronik", "Mobilya"]
sonuc = df[df["kategori"].isin(secili_kategoriler)]

print(sonuc)
```

## Mini Veri Bilimi Uygulaması

Veri setinden sadece analizde kullanılacak sütunları seçip pahalı ürünleri filtreleyelim.


```python
import pandas as pd
from io import StringIO

csv_verisi = StringIO("""urun,kategori,fiyat,stok
Laptop,Elektronik,30000,5
Telefon,Elektronik,20000,8
Tablet,Elektronik,12000,12
Kulaklık,Aksesuar,2500,30
Masa,Mobilya,7500,6
""")

df = pd.read_csv(csv_verisi)

analiz_df = df.loc[:, ["urun", "kategori", "fiyat"]]
analiz_df = analiz_df[analiz_df["fiyat"] >= 10000]

print(analiz_df)
```

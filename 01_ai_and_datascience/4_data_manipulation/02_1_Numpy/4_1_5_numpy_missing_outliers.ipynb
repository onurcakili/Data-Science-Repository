# NumPy ile Eksik Değer ve Aykırı Değer İşlemleri

Gerçek veri setleri her zaman temiz gelmez. Eksik değerler, hatalı sayılar ve aykırı değerler veri analizinde sık karşılaşılan durumlardır.

Bu notebook'ta `np.nan`, `np.isnan`, `np.nanmean`, `np.where`, `np.clip` ve koşullu dönüşüm mantığını öğreneceğiz.


## Konu Dokümantasyonu

NumPy'de eksik sayısal değerler genellikle `np.nan` ile temsil edilir. `nan`, "not a number" anlamına gelir.

Normal `mean()` fonksiyonu içinde `nan` varsa sonuç da `nan` olur. Bu yüzden eksik değerleri yok sayan `np.nanmean()`, `np.nanmedian()` gibi fonksiyonlar kullanılır.

Aykırı değer ise veri setinin genel yapısından çok uzak olan değerdir. Aykırı değerler bazen gerçek olabilir, bazen veri giriş hatası olabilir. Bu yüzden doğrudan silmeden önce bağlamı anlamak gerekir.


## Kolay Seviye

Eksik değerleri tespit etmek için `np.isnan()` kullanılır.


```python
import numpy as np

notlar = np.array([70, 85, np.nan, 60, 95])

print(np.isnan(notlar))          # Hangi değerlerin eksik olduğunu True/False olarak gösterir
print(np.mean(notlar))           # nan içerdiği için sonuç nan olur
print(np.nanmean(notlar))        # nan değerleri yok sayarak ortalama hesaplar
```

```python
import numpy as np

gelirler = np.array([12000, 15000, np.nan, 21000, 24000])

ortalama_gelir = np.nanmean(gelirler)
doldurulmus_gelirler = np.where(np.isnan(gelirler), ortalama_gelir, gelirler)
# Eksikse ortalama ile doldur, değilse mevcut değeri koru

print(doldurulmus_gelirler)
```

## Orta Seviye

`np.where()` koşula göre değer üretir. Bu yapı veri temizleme ve yeni özellik üretme işlemlerinde sık kullanılır.

`np.clip()` ise değerleri belirli alt ve üst sınır arasında tutar. Aykırı değerleri sınırlamak için kullanılabilir.


```python
import numpy as np

satislar = np.array([500, 1200, 2500, 100000, 1800])

sinirlanmis_satislar = np.clip(satislar, 0, 5000)
# 5000 üzerindeki değerler 5000'e çekilir

print(satislar)
print(sinirlanmis_satislar)
```

```python
import numpy as np

skorlar = np.array([35, 55, 72, 88, 95])

segmentler = np.where(skorlar >= 70, "başarılı", "geliştirilmeli")

print(segmentler)
```

## İleri Seviye

Aykırı değerleri tespit etmek için çeyrekler arası açıklık yöntemi kullanılabilir.

IQR yöntemi:

* Q1: 25. yüzdelik
* Q3: 75. yüzdelik
* IQR = Q3 - Q1
* Alt sınır = Q1 - 1.5 * IQR
* Üst sınır = Q3 + 1.5 * IQR


```python
import numpy as np

satislar = np.array([900, 1100, 1200, 1300, 1500, 1600, 10000])

q1 = np.percentile(satislar, 25)
q3 = np.percentile(satislar, 75)
iqr = q3 - q1

alt_sinir = q1 - 1.5 * iqr
ust_sinir = q3 + 1.5 * iqr

aykiri_maske = (satislar < alt_sinir) | (satislar > ust_sinir)

print("Alt sınır:", alt_sinir)
print("Üst sınır:", ust_sinir)
print("Aykırı değerler:", satislar[aykiri_maske])
```

## Mini Veri Bilimi Uygulaması

Ham gelir verisini temizleyelim: eksikleri medyanla dolduralım, çok yüksek aykırı değerleri üst sınıra çekelim.


```python
import numpy as np

gelirler = np.array([12000, 15000, np.nan, 18000, 21000, 24000, 250000])

medyan = np.nanmedian(gelirler)
gelirler = np.where(np.isnan(gelirler), medyan, gelirler)

q1 = np.percentile(gelirler, 25)
q3 = np.percentile(gelirler, 75)
iqr = q3 - q1
ust_sinir = q3 + 1.5 * iqr

temiz_gelirler = np.clip(gelirler, 0, ust_sinir)

print("Doldurulmuş gelirler:", gelirler)
print("Temiz gelirler:", temiz_gelirler)
```

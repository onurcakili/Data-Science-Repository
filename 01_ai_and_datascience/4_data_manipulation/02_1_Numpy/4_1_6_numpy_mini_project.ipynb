Amaç:

## Konu Dokümantasyonu

Mini proje notebook'ları, önceki konuları tek bir senaryo içinde birleştirmek için hazırlanır. Amaç yalnızca fonksiyonları ayrı ayrı göstermek değil, veri bilimi akışının nasıl kurulduğunu görmektir.

Bu NumPy mini projesinde satış verisi matris olarak temsil edilir. Satırlar ürünleri, sütunlar günleri gösterir. Bu yapı sayesinde satır bazlı ürün performansı, sütun bazlı günlük performans ve hedefe göre etiketleme işlemleri yapılabilir.

Bu notebook'ta kullanılan temel NumPy becerileri:

* Array oluşturma
* `axis` mantığıyla satır/sütun bazlı özet çıkarma
* `argmax()` ile en yüksek değerin konumunu bulma
* `np.where()` ile koşullu etiket üretme
* Min-max normalizasyonu
* Basit müşteri segmentasyonu


## Senaryo

Elimizde 4 ürünün 5 günlük satış tutarları var. Satırlar ürünleri, sütunlar günleri temsil ediyor.

Veri bilimi açısından bu yapı küçük bir matris problemidir. Satır bazlı analiz ürün performansını, sütun bazlı analiz günlük performansı gösterir.


```python
import numpy as np

urunler = np.array(["Laptop", "Telefon", "Tablet", "Kulaklık"])
gunler = np.array(["Pzt", "Sal", "Çar", "Per", "Cum"])

satis = np.array([
    [32000, 28000, 35000, 30000, 42000],
    [22000, 25000, 24000, 27000, 31000],
    [9000, 12000, 11000, 10000, 14000],
    [3500, 4200, 3900, 4500, 5200],
])

print(satis)
```

## Kolay Seviye: Temel Özetler

Önce ürün bazlı ve gün bazlı toplamları hesaplayalım. `axis=1` satırları, `axis=0` sütunları özetler.


```python
import numpy as np

urun_toplamlari = satis.sum(axis=1)       # Her ürünün haftalık toplamı
gun_toplamlari = satis.sum(axis=0)        # Her günün toplam satışı

print("Ürün toplamları:")
for urun, toplam in zip(urunler, urun_toplamlari):
    print(urun, toplam)

print("Gün toplamları:")
for gun, toplam in zip(gunler, gun_toplamlari):
    print(gun, toplam)
```

## Orta Seviye: En İyi Ürün ve En İyi Gün

`argmax()` en büyük değerin indeksini verir. Bu indeksle ürün veya gün adını seçebiliriz.


```python
en_iyi_urun_indeksi = urun_toplamlari.argmax()
en_iyi_gun_indeksi = gun_toplamlari.argmax()

print("En iyi ürün:", urunler[en_iyi_urun_indeksi])
print("En iyi gün:", gunler[en_iyi_gun_indeksi])
```

```python
hedef = 100000

performans = np.where(urun_toplamlari >= hedef, "hedef üstü", "hedef altı")
# Ürün haftalık toplamı hedefi geçerse hedef üstü etiketi verilir

for urun, toplam, etiket in zip(urunler, urun_toplamlari, performans):
    print(urun, toplam, etiket)
```

## İleri Seviye: Müşteri Özelliklerini Normalize Etme

Şimdi müşteri yaşı, aylık gelir ve harcama skorunu normalize edelim. Bu işlem farklı ölçeklerdeki değişkenleri karşılaştırılabilir yapar.


```python
musteri_verisi = np.array([
    [22, 15000, 65],
    [35, 28000, 80],
    [46, 42000, 55],
    [28, 21000, 90],
    [52, 50000, 40],
], dtype=float)

min_degerler = musteri_verisi.min(axis=0)
max_degerler = musteri_verisi.max(axis=0)

normalize = (musteri_verisi - min_degerler) / (max_degerler - min_degerler)

print(normalize)
```

## Mini Sonuç

NumPy ile veri manipülasyonu yaparken temel fikir şudur:

* Veriyi array olarak düzenle
* Satır/sütun eksenini doğru anla
* Vektörize hesaplama kullan
* Koşullu seçimle etiket üret
* Gerekirse veriyi normalize et


```python
harcama_skoru = musteri_verisi[:, 2]
gelir = musteri_verisi[:, 1]

segment = np.where(
    (gelir >= 25000) & (harcama_skoru >= 70),
    "değerli müşteri",
    "standart müşteri",
)

print(segment)
```

# NumPy İstatistik ve Rastgele Veri Üretimi

Veri biliminde istatistiksel özetler ve simülasyonlar çok sık kullanılır. NumPy, ortalama, medyan, standart sapma, yüzdelik değerler ve rastgele veri üretimi için güçlü araçlar sunar.


## Konu Dokümantasyonu

İstatistiksel özetler, veri setinin genel davranışını anlamamızı sağlar.

Yaygın metrikler:

* Ortalama: Genel merkez eğilimi gösterir.
* Medyan: Sıralı verinin ortasındaki değerdir.
* Standart sapma: Değerlerin ne kadar dağıldığını gösterir.
* Yüzdelikler: Verinin belirli noktalarını anlamak için kullanılır.

Rastgele veri üretimi ise örnek veri oluşturmak, simülasyon yapmak ve makine öğrenmesi deneylerini tekrarlanabilir kılmak için kullanılır.


## Kolay Seviye

Temel istatistiksel fonksiyonlarla başlayalım.


```python
import numpy as np

satislar = np.array([1200, 950, 1800, 2100, 1750, 1300])

print("Ortalama:", np.mean(satislar))
print("Medyan:", np.median(satislar))
print("Standart sapma:", np.std(satislar))
print("Minimum:", np.min(satislar))
print("Maksimum:", np.max(satislar))
```

```python
import numpy as np

gelirler = np.array([12000, 15000, 18000, 21000, 24000, 50000])

print("25. yüzdelik:", np.percentile(gelirler, 25))
print("50. yüzdelik:", np.percentile(gelirler, 50))
print("75. yüzdelik:", np.percentile(gelirler, 75))
```

## Orta Seviye

Rastgele veri üretirken `np.random.default_rng()` kullanmak modern ve önerilen yöntemdir.

`seed` değeri verildiğinde aynı rastgele değerler tekrar üretilebilir. Bu, analiz sonuçlarının yeniden üretilebilir olması için önemlidir.


```python
import numpy as np

rng = np.random.default_rng(seed=42)

rastgele_sayilar = rng.integers(1, 100, size=10)
normal_dagilim = rng.normal(loc=50, scale=10, size=5)

print(rastgele_sayilar)
print(normal_dagilim)
```

```python
import numpy as np

rng = np.random.default_rng(seed=42)

gunluk_satislar = rng.normal(loc=1500, scale=300, size=30)
gunluk_satislar = np.round(gunluk_satislar, 2)

print("İlk 5 gün:", gunluk_satislar[:5])
print("Aylık ortalama:", gunluk_satislar.mean())
```

## İleri Seviye

Simülasyon, belirsiz durumları anlamak için tekrar tekrar rastgele senaryo üretmektir.

Örneğin bir kampanyanın dönüşüm oranını tahmin etmek için 1000 kullanıcı üzerinde satın alma olup olmadığını simüle edebiliriz.


```python
import numpy as np

rng = np.random.default_rng(seed=7)

kullanici_sayisi = 1000
donusum_orani = 0.08

satin_alma = rng.binomial(n=1, p=donusum_orani, size=kullanici_sayisi)
# Her kullanıcı için 1 satın aldı, 0 satın almadı anlamına gelir

toplam_satin_alma = satin_alma.sum()
gerceklesen_oran = satin_alma.mean()

print("Toplam satın alma:", toplam_satin_alma)
print("Gerçekleşen dönüşüm oranı:", gerceklesen_oran)
```

## Mini Veri Bilimi Uygulaması

A/B testi için iki farklı kampanya dönüşümünü simüle edelim. Bu örnek gerçek istatistiksel test yerine temel simülasyon mantığını göstermek içindir.


```python
import numpy as np

rng = np.random.default_rng(seed=10)

grup_a = rng.binomial(n=1, p=0.07, size=1000)
grup_b = rng.binomial(n=1, p=0.09, size=1000)

oran_a = grup_a.mean()
oran_b = grup_b.mean()
fark = oran_b - oran_a

print("A dönüşüm oranı:", oran_a)
print("B dönüşüm oranı:", oran_b)
print("Fark:", fark)
```

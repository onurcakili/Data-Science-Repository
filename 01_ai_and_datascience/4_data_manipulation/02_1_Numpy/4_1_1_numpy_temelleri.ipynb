## Konu Dokümantasyonu

NumPy'nin ana veri yapısı `ndarray` olarak adlandırılır. Bu yapı, aynı türden verileri çok boyutlu bir dizi içinde saklar.

Python listeleri esnektir; farklı türde verileri aynı listede tutabilir. NumPy array ise genellikle tek veri türüyle çalışır. Bu kısıtlama bir dezavantaj gibi görünse de sayısal hesaplamalarda büyük hız ve tutarlılık sağlar.

NumPy veri biliminde özellikle şu alanlarda kullanılır:

* Sayısal veri saklama
* Matris ve vektör işlemleri
* İstatistiksel hesaplamalar
* Makine öğrenmesi algoritmalarının temel hesaplamaları
* Pandas ve Scikit-learn gibi kütüphanelerin arka planındaki sayısal işlemler

NumPy'yi iyi anlamak, veri bilimi kütüphanelerinin nasıl çalıştığını daha net görmeyi sağlar.


## Kolay Seviye

NumPy genellikle `np` takma adıyla içe aktarılır. Array oluşturmak için `np.array()` kullanılır.

Array'in temel özellikleri:

* `shape`: Satır/sütun veya boyut bilgisini verir.
* `dtype`: İçindeki değerlerin veri türünü gösterir.
* `ndim`: Kaç boyutlu olduğunu gösterir.
* `size`: Toplam eleman sayısını verir.


```python
import numpy as np

notlar = np.array([70, 85, 90, 60, 95])

print(notlar)
print("Boyut:", notlar.ndim)      # 1 boyutlu array
print("Şekil:", notlar.shape)     # (5,) -> 5 elemanlı tek boyut
print("Veri türü:", notlar.dtype)
print("Eleman sayısı:", notlar.size)
```

```python
import numpy as np

satis_matrisi = np.array([
    [1200, 950, 1800],
    [2100, 1750, 1600],
    [900, 1100, 1300],
])

print(satis_matrisi)
print("Boyut:", satis_matrisi.ndim)      # 2 boyutlu: satır ve sütun
print("Şekil:", satis_matrisi.shape)     # 3 satır, 3 sütun
```

## Orta Seviye

NumPy'de array oluşturmak için sık kullanılan yardımcı fonksiyonlar vardır.

* `np.arange()`: Belirli aralıkta sayılar üretir.
* `np.linspace()`: Başlangıç ve bitiş arasında eşit aralıklı değerler üretir.
* `np.zeros()`: Sıfırlardan oluşan array üretir.
* `np.ones()`: Birlerden oluşan array üretir.
* `reshape()`: Array'in şeklini değiştirir.

`reshape()` yaparken toplam eleman sayısı değişemez. Örneğin 12 elemanlı bir array, `3x4` veya `2x6` hâline getirilebilir.


```python
import numpy as np

sayilar = np.arange(1, 11, 2)       # başlangıç=1, bitiş=11, artış=2
oranlar = np.linspace(0, 1, 5)      # 0 ile 1 arasında 5 eşit değer

print("Arange:", sayilar)
print("Linspace:", oranlar)
```

```python
import numpy as np

veri = np.arange(1, 13)
matris = veri.reshape(3, 4)         # 12 eleman -> 3 satır ve 4 sütun

print(veri)
print(matris)
```

## İleri Seviye

NumPy'nin en güçlü taraflarından biri vektörize işlemdir. Vektörize işlem, her eleman için tek tek döngü yazmadan tüm array üzerinde işlem yapmaktır.

Python listesinde her eleman için `for` döngüsü yazmak gerekirken NumPy array üzerinde doğrudan matematiksel işlem yapılabilir. Bu hem okunabilirliği artırır hem de büyük verilerde daha hızlı çalışır.


```python
import numpy as np

fiyatlar = np.array([100, 250, 400, 800])
kdv_orani = 0.20

kdv_dahil_fiyatlar = fiyatlar * (1 + kdv_orani)
# Her fiyat değeri 1.20 ile çarpılır; tek tek döngü yazmaya gerek yoktur

print(kdv_dahil_fiyatlar)
```

## Mini Veri Bilimi Uygulaması

Bir ürünün farklı aylardaki satışlarını array içinde tutup temel metrikleri hesaplayalım.

Bu örnek küçük görünse de veri bilimindeki birçok özetleme işleminin temelidir: toplam, ortalama, maksimum, minimum ve büyüme oranı.


```python
import numpy as np

aylik_satislar = np.array([12000, 15000, 14200, 18000, 21000, 24000])

toplam = aylik_satislar.sum()
ortalama = aylik_satislar.mean()
maksimum = aylik_satislar.max()
minimum = aylik_satislar.min()
buyume_orani = (aylik_satislar[-1] - aylik_satislar[0]) / aylik_satislar[0] * 100

print("Toplam satış:", toplam)
print("Ortalama satış:", ortalama)
print("En yüksek satış:", maksimum)
print("En düşük satış:", minimum)
print("İlk aydan son aya büyüme (%):", buyume_orani)
```

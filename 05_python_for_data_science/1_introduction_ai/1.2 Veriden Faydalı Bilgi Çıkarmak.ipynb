# 1.2) Veriden Faydalı Bilgi Çıkarmak

## Özet

Veri bilimi süreci; ham verinin problem tanımlama, veri toplama, veri temizleme, keşifsel analiz, modelleme, değerlendirme, yorumlama ve raporlama adımlarıyla anlamlı ve uygulanabilir bilgiye dönüştürülmesini ifade eden sistematik ve yinelemeli yaklaşımdır.  
Bu sürecin temel amacı, ham veriyi yalnızca düzenlemek veya görselleştirmek değil; bu veriden anlamlı, güvenilir ve karar süreçlerinde kullanılabilir bilgi üretmektir.  
Bilgi (Knowledge), enformasyonun analiz edilmesi ve yorumlanması sonucunda elde edilen içgörüdür. Veriden faydalı bilgi çıkarma süreci; istatistiksel analiz, keşifsel veri analizi, modelleme, hipotez testi, tahminleme ve karar destek yaklaşımlarının birlikte kullanılmasını gerektirir.

Veri setleri üzerinde yapılan analizler, bir problemin mevcut durumunu anlamayı, geçmiş olayların nedenlerini araştırmayı, geleceğe ilişkin olası sonuçları tahmin etmeyi ve en uygun aksiyonları belirlemeyi mümkün kılar.  

Veri bilimi; yapılandırılmış ve yapılandırılmamış verilerden anlamlı bilgi, öngörü ve karar destek çıktıları üretmek amacıyla istatistik, matematik, bilgisayar bilimleri ve alan bilgisini bir araya getiren disiplinler arası çalışma alanıdır.  

Bu nedenle veri bilimi, teknik bir modelleme faaliyeti olmanın yanında, karar alma süreçlerini destekleyen sistematik bir analitik yaklaşımdır.

## Veriden Bilgi Üretme Yaklaşımı

Veriden faydalı bilgi çıkarma süreci genellikle şu sorular etrafında şekillenir:

- Ne oldu?
- Neden oldu?
- Ne olabilir?
- Ne yapılmalıdır?

Bu sorular, veri analitiğinin dört temel düzeyiyle ilişkilidir:

1. Betimleyici Analitik (Descriptive Analytics)
2. Tanısal Analitik (Diagnostic Analytics)
3. Tahminsel Analitik (Predictive Analytics)
4. Yönergeli / Normatif Analitik (Prescriptive Analytics)

Bu düzeyler birbirinden tamamen bağımsız değildir. 

Gerçek veri bilimi projelerinde çoğu zaman önce mevcut durum betimlenir, ardından nedenler araştırılır, geleceğe yönelik tahminler yapılır ve son olarak karar vericilere uygulanabilir öneriler sunulur.

## 1. Betimleyici Analitik (Descriptive Analytics)

Betimleyici analitik, “Ne oldu?” sorusuna yanıt arar.

Bu yaklaşım, geçmişte veya mevcut durumda gerçekleşen olayları anlamak için verileri özetler, düzenler ve görselleştirir.
Temel amaç; veri setinin genel yapısını, eğilimlerini, dağılımlarını ve dikkat çeken örüntülerini ortaya koymaktır.

Betimleyici analitikte sık kullanılan yöntemler şunlardır:

- Frekans tabloları
- Ortalama, medyan, standart sapma gibi betimsel istatistikler
- Çubuk grafik, çizgi grafik, pasta grafik ve histogramlar
- Dashboard ve raporlama araçları

Örneğin bir uçuş rezervasyon şirketi, yıllara göre satılan bilet sayılarını analiz ederek hangi dönemlerde rezervasyonların arttığını veya azaldığını görebilir. Bu analiz, şirketin geçmiş performansını anlamasına ve dönemsel eğilimleri fark etmesine yardımcı olur.

## 2. Tanısal Analitik (Diagnostic Analytics)

Tanısal analitik, “Neden oldu?” sorusuna yanıt arar.

Betimleyici analitik bir olayın gerçekleştiğini gösterirken, tanısal analitik bu olayın arkasındaki olası nedenleri araştırır.
Bu aşamada veri daha ayrıntılı incelenir; değişkenler arasındaki ilişkiler, alt gruplar, sapmalar ve beklenmeyen durumlar analiz edilir.

Tanısal analitikte kullanılan yaklaşımlar arasında şunlar yer alır:

- Veri kırılımı ve detay analizi
- Korelasyon analizi
- Segmentasyon
- Karşılaştırmalı analizler
- Aykırı değer incelemeleri
- Hipotez testleri

Örneğin bir uçuş şirketi belirli bir ayda rezervasyonlarda olağan dışı bir artış gözlemlediğinde, bu artışın nedenini araştırabilir. 
Detaylı analiz sonucunda, ilgili şehirde düzenlenen büyük bir spor etkinliğinin talebi artırdığı görülebilir.

## 3. Tahminsel Analitik (Predictive Analytics)

Tahminsel analitik, “Ne olabilir?” veya “Ne olacak?” sorusuna yanıt arar.

Bu yaklaşım, geçmiş verilerden öğrenilen örüntüleri kullanarak gelecekte gerçekleşmesi muhtemel olayları tahmin etmeyi amaçlar. 
Tahminsel analitik, istatistiksel modeller ve makine öğrenmesi algoritmalarından yararlanır.

Tahminsel analitikte kullanılan yöntemlere örnek olarak şunlar verilebilir:

- Regresyon modelleri
- Sınıflandırma algoritmaları
- Zaman serisi analizi
- Makine öğrenmesi modelleri
- Talep tahmini
- Risk tahmini

Örneğin bir uçuş rezervasyon şirketi geçmiş yıllardaki rezervasyon verilerini kullanarak gelecek aylarda hangi destinasyonlarda talep artışı yaşanacağını tahmin edebilir. Bu tahminler, pazarlama kampanyalarının ve operasyonel planlamanın daha erken yapılmasını sağlar.

## 4. Yönergeli / Normatif Analitik (Prescriptive Analytics)

Yönergeli analitik, “Ne yapılmalıdır?” sorusuna yanıt arar.

Bu analitik düzey, yalnızca gelecekte ne olabileceğini tahmin etmekle kalmaz; aynı zamanda en uygun karar veya aksiyonun ne olabileceğini belirlemeye çalışır. Bu yönüyle tahminsel analitiğin bir adım ötesine geçer.

Yönergeli analitikte kullanılan yaklaşımlar şunları içerebilir:

- Optimizasyon modelleri
- Simülasyon
- Senaryo analizi
- Karar destek sistemleri
- Öneri sistemleri
- Makine öğrenmesi tabanlı aksiyon önerileri

Örneğin bir uçuş şirketi, belirli destinasyonlarda (varılmak istenen yer) talep artışı bekliyorsa, farklı pazarlama bütçesi senaryolarını karşılaştırabilir. 
Hangi kanalda ne kadar reklam harcaması yapılırsa en yüksek rezervasyon dönüşü alınacağı analiz edilerek karar vericilere uygulanabilir öneriler sunulabilir.

## Keşifsel Veri Analizi (Exploratory Data Analysis - EDA)

Keşifsel veri analizi; verinin genel yapısını, dağılımlarını, ilişkilerini, aykırı değerlerini ve potansiyel örüntülerini anlamak için yapılan ön analiz sürecidir. 
Modelleme öncesinde veri hakkında güçlü bir sezgi geliştirmek için kullanılan temel bir analiz aşamasıdır.

EDA, yalnızca grafik üretmekten ibaret değildir. Aynı zamanda verinin kalitesini, değişkenlerin dağılımını, eksik değerleri, aykırı gözlemleri ve değişkenler arasındaki ilişkileri sistematik olarak incelemeyi içerir.

Keşifsel veri analizinde sık kullanılan adımlar şunlardır:

- Veri setinin boyutunu (örn. 3, 4) ve değişken türlerini incelemek
- Eksik değerleri ve tekrar eden kayıtları belirlemek
- Sayısal değişkenlerin dağılımlarını analiz etmek
- Kategorik değişkenlerin frekanslarını (sıklığını) incelemek
- Aykırı değerleri tespit etmek
- Değişkenler arasındaki ilişkileri görselleştirmek
- İlk hipotezleri oluşturmak veya mevcut varsayımları kontrol etmek

EDA, makine öğrenmesi sürecinin başarısını doğrudan etkiler çünkü veri yeterince anlaşılmadan yapılan modelleme, yanıltıcı sonuçlara, yanlış değişken seçimlerine veya düşük genellenebilirliğe sahip modellere yol açabilir.

## Analitik Çıkarımlarda Dikkat Edilmesi Gerekenler

Veriden elde edilen her sonuç doğrudan doğruya karar almak için yeterli olmayabilir.  
Bilgi (Knowledge), enformasyonun analiz edilmesi ve yorumlanması sonucunda elde edilen içgörüdür.  
  
Analitik bulgular değerlendirilirken şu noktalara dikkat edilmelidir:

- Veri kalitesi sonuçların güvenilirliğini doğrudan etkiler.
- Korelasyon, nedensellik anlamına gelmez.
- Eksik, yanlı veya temsil gücü düşük veri hatalı çıkarımlara neden olabilir.
- Model sonuçları iş bağlamı içinde yorumlanmalıdır.
- Analiz çıktıları teknik doğruluğun yanında anlaşılabilir ve uygulanabilir olmalıdır.

## Sonuç

Veriden faydalı bilgi çıkarmak, veri bilimi sürecinin merkezinde yer alan sistematik bir yaklaşımdır. Betimleyici analitik mevcut durumu anlamayı, tanısal analitik nedenleri araştırmayı, tahminsel analitik geleceğe yönelik öngörüler üretmeyi, yönergeli analitik ise en uygun aksiyonları belirlemeyi sağlar.

Başarılı bir veri bilimi çalışması, yalnızca algoritma kullanımıyla değil; doğru problem tanımı, kaliteli veri, uygun analiz yöntemi, eleştirel yorumlama ve karar süreçlerine aktarılabilir çıktılarla mümkündür.

# 1.4) CRISP-DM Metodolojisi

Günümüzde teknolojik dönüşümün merkezinde yer alan en kritik unsurlardan biri veridir. 
Verinin hacim, hız ve çeşitlilik bakımından sürekli artması, geleneksel yöntemlerle analiz ve karar alma süreçlerini giderek daha karmaşık hâle getirmektedir. 
Bu durum, kurumların veri odaklı yaklaşımlara, sistematik analiz süreçlerine ve standartlaştırılmış proje metodolojilerine olan ihtiyacını artırmıştır.

Veri bilimi projeleri büyüdükçe, yalnızca teknik model geliştirme yeterli olmamakta; iş probleminin doğru anlaşılması, verinin uygun biçimde hazırlanması, model çıktılarının değerlendirilmesi ve sonuçların kurumsal süreçlere entegre edilmesi gerekmektedir. 
RISP-DM metodolojisi, bu ihtiyaca yanıt veren en yaygın veri madenciliği ve veri bilimi proje yönetim yaklaşımlarından biridir.

## CRISP-DM Nedir?

CRISP-DM, “Cross-Industry Standard Process for Data Mining” ifadesinin kısaltmasıdır. 
Farklı sektörlerde uygulanabilecek şekilde tasarlanmış, veri madenciliği ve veri bilimi projeleri için standartlaştırılmış bir süreç modelidir.

Bu metodoloji, iş problemlerinin veri temelli çözümlerle ele alınmasını ve proje sürecinin sistematik biçimde yönetilmesini amaçlar. 
Başka bir ifadeyle CRISP-DM, veri bilimi projelerinde izlenen genel yol haritasını tanımlar.

Tarihsel süreçte SEMMA ve KDD gibi yöntemlerle birlikte veri madenciliği projelerinin yapılandırılmasına katkı sağlamıştır. Daha sonraki yıllarda ise CRISP-DM yaklaşımı, kurumsal ölçekte veri bilimi ve analitik projelerde kullanılan farklı metodolojilere de temel oluşturmuştur.

CRISP-DM yüksek seviyeli ve esnek bir metodoloji olduğundan, modelde yer alan aşamalar her zaman doğrusal bir sırayla ilerlemek zorunda değildir. Projenin ihtiyaçlarına göre aşamalar arasında geri dönüşler yapılabilir ve süreç yinelemeli şekilde sürdürülebilir.

CRISP-DM metodolojisi toplam altı temel aşamadan oluşur:

1. İşi Anlama (Business Understanding)
2. Veriyi Anlama (Data Understanding)
3. Veriyi Hazırlama (Data Preparation)
4. Modelleme (Modeling)
5. Değerlendirme (Evaluation)
6. Uygulama / Canlıya Alma (Deployment)

Not: Bu aşamalar kesin ve değişmez bir sıra ifade etmez. Veri bilimi projelerinde çoğu zaman farklı aşamalar arasında tekrar tekrar geçiş yapılması gerekebilir.

## CRISP-DM Aşamaları

## 1. İşi Anlama (Business Understanding)

CRISP-DM sürecinin ilk aşaması, iş probleminin doğru biçimde anlaşılmasıdır. Bu aşamada projenin amacı, kapsamı, başarı kriterleri ve iş gereksinimleri belirlenir.

Doğru problem tanımı, projenin ilerleyen aşamalarında alınacak tüm teknik kararların temelini oluşturur. Bu nedenle yalnızca veri veya model odaklı düşünmek yerine, problemin kurumsal bağlamı ve iş hedefleri ayrıntılı biçimde değerlendirilmelidir.

Bu aşamada genellikle şu sorulara yanıt aranır:

- Çözülmek istenen iş problemi nedir?
- Projenin başarı ölçütleri nelerdir?
- Hangi karar süreçleri veriyle desteklenecektir?
- Kullanılabilecek veri kaynakları nelerdir?
- Projenin iş birimleri açısından beklenen çıktısı nedir?

Örneğin bir şirket, ürün reklamlarını müşterilere daha doğru şekilde göstermek isteyebilir. Bu durumda temel iş problemi, doğru müşteriye doğru ürün reklamını sunarak pazarlama etkinliğini artırmaktır.

## 2. Veriyi Anlama (Data Understanding)

Veriyi anlama aşamasında, iş problemine uygun veri kaynakları incelenir ve mevcut verinin genel yapısı değerlendirilir. Bu aşama, veri toplama sürecini, verinin ilk analizini ve veri kalitesine ilişkin temel incelemeleri kapsar.

Problem tanımının doğru yapılması bu aşamada kritik öneme sahiptir. Aksi durumda, eldeki tüm verinin gereksiz biçimde işlenmesi veya iş problemiyle doğrudan ilişkili olmayan değişkenlere odaklanılması gibi hatalar ortaya çıkabilir.

Bu aşamada veriyle ilgili aşağıdaki noktalar değerlendirilir:

- Veri kaynaklarının kapsamı
- Verinin yapısal, yarı yapısal veya yapısal olmayan niteliği
- Eksik değerler
- Gürültülü veya hatalı veriler
- Aykırı gözlemler
- Değişkenlerin anlamı ve iş problemiyle ilişkisi
- Ek veri ihtiyacı

Gerekli görüldüğünde ilave veri toplama kararı alınabilir. Ayrıca mevcut verinin hangi işlemlerle zenginleştirilebileceği de bu aşamada değerlendirilir.

## 3. Veriyi Hazırlama (Data Preparation)

Veriyi hazırlama aşaması, modelleme sürecinde kullanılacak nihai veri setinin oluşturulduğu aşamadır. Bu süreç, veri temizleme, dönüştürme, birleştirme, filtreleme ve yeni değişkenler üretme gibi işlemleri içerir.

Önceki aşamada tespit edilen eksik değerler, hatalı kayıtlar veya tutarsızlıklar bu aşamada ele alınır. Örneğin eksik değer içeren gözlemlerin veri setinden çıkarılmasına, eksik değerlerin belirli yöntemlerle tamamlanmasına veya ilgili değişkenlerin analiz dışında bırakılmasına karar verilebilir.

Eksik değerlerin tamamlanması süreci literatürde “imputation” olarak adlandırılır. Bu işlem, ortalama, medyan, regresyon temelli yöntemler veya daha gelişmiş makine öğrenmesi yaklaşımlarıyla gerçekleştirilebilir.

Veri hazırlama kapsamında yapılabilecek işlemlere şu örnekler verilebilir:

- Doğum tarihinden yaş değişkeninin türetilmesi
- Adres bilgisinden il ve ilçe bilgilerinin çıkarılması
- Kategorik değişkenlerin sayısal formata dönüştürülmesi
- Aykırı değerlerin incelenmesi
- Ölçeklendirme veya normalizasyon işlemleri
- Farklı veri kaynaklarının birleştirilmesi

Bu aşama, öznitelik mühendisliği (feature engineering) süreciyle yakından ilişkilidir.
Öznitelik mühendisliği, ham veriden model performansını artırabilecek anlamlı, açıklayıcı ve tahmin gücü yüksek değişkenler üretme sürecidir.

Başarılı bir öznitelik mühendisliği süreci, modelin doğruluğunu ve genellenebilirliğini önemli ölçüde artırabilir. Bu kapsamda genellikle üç temel işlem öne çıkar:

- Özellik seçimi
- Özellik dönüştürme
- Yeni özellik oluşturma

## 4. Modelleme (Modeling)

Modelleme aşamasında, tanımlanan iş problemi ve hazırlanan veri seti doğrultusunda uygun istatistiksel yöntemler veya makine öğrenmesi algoritmaları kullanılarak model geliştirilir.

Bu aşamada karar ağaçları, lojistik regresyon, doğrusal regresyon, destek vektör makineleri, kümeleme algoritmaları, sinir ağları veya daha gelişmiş makine öğrenmesi yöntemleri kullanılabilir. Kullanılacak yöntem, problemin türüne, veri yapısına ve başarı kriterlerine bağlı olarak belirlenir.

Modelleme sürecinde yalnızca algoritmanın seçilmesi yeterli değildir. Aynı zamanda modelin parametreleri ayarlanır, farklı algoritmalar karşılaştırılır ve model performansını artırmaya yönelik optimizasyon çalışmaları yapılır.

Bazı modellerin etkili çalışabilmesi için verinin belirli koşulları sağlaması gerekebilir. Örneğin bazı algoritmalar dengeli sınıf dağılımına ihtiyaç duyarken, bazıları aykırı değerlere karşı duyarlı olabilir. Benzer şekilde belirli modeller için ölçeklendirme, normalizasyon veya veri dönüşümü gerekebilir.

Bu nedenle modelleme aşaması, gerektiğinde veri hazırlama aşamasına geri dönülmesini gerektirebilir.

## 5. Değerlendirme (Evaluation)

Değerlendirme aşamasında, geliştirilen modelin iş hedeflerini ne ölçüde karşıladığı analiz edilir. Bu aşama yalnızca teknik performans metriklerinin incelenmesini değil, aynı zamanda model çıktılarının iş problemi açısından anlamlılığının değerlendirilmesini de kapsar.

İlk aşamada belirlenen başarı kriterleri bu noktada yeniden ele alınır. Modelin doğruluk, hata oranı, hassasiyet, duyarlılık, F1 skoru, RMSE veya benzeri metrikler açısından performansı ölçülebilir.

Örneğin fiyat tahmini yapılan bir problemde, gerçek fiyat ile tahmin edilen fiyat arasındaki fark RMSE gibi hata metrikleriyle değerlendirilebilir. Sınıflandırma problemlerinde ise doğruluk, kesinlik, duyarlılık ve F1 skoru gibi ölçütler kullanılabilir.

Bazı projelerde modelin başarısı saha testleriyle ölçülür. Örneğin bir e-ticaret sitesinde doğru reklam gösterimi için geliştirilen bir model, A/B testi aracılığıyla değerlendirilebilir. Bu testte model kullanılan ve kullanılmayan iki kullanıcı grubu karşılaştırılarak modelin iş sonuçlarına etkisi incelenir.

Değerlendirme sonucunda modelin yeterli performans göstermediği anlaşılırsa, önceki aşamalara geri dönülerek veri hazırlama, özellik mühendisliği veya modelleme süreçleri yeniden ele alınabilir.

## 6. Uygulama / Canlıya Alma (Deployment)

Uygulama aşaması, geliştirilen modelin gerçek iş süreçlerine entegre edildiği aşamadır. Bu aşamada model, kurumun teknik altyapısına, veri akışlarına ve operasyonel ihtiyaçlarına uygun şekilde kullanılabilir hâle getirilir.

Modelin yalnızca geliştirilmiş olması yeterli değildir; sürdürülebilir, izlenebilir ve güncellenebilir bir yapıda çalışması gerekir. Bu nedenle canlıya alma sürecinde yazılım geliştirme, veri mühendisliği, sistem entegrasyonu ve operasyonel izleme süreçleri birlikte ele alınmalıdır.

Örneğin bir e-ticaret sitesinde reklam gösterimlerini yönetecek bir veri bilimi modeli geliştirildiyse, bu modelin web sitesi veya reklam yönetim sistemiyle entegre edilmesi gerekir. Modelin ürettiği karar sinyallerine göre kullanıcılara gösterilecek reklamlar dinamik biçimde belirlenebilir.

Canlıya alınan modeller zaman içinde performans kaybına uğrayabilir. Bu nedenle model performansının düzenli olarak izlenmesi, veri dağılımındaki değişimlerin takip edilmesi ve gerektiğinde modelin yeniden eğitilmesi önemlidir.

Bu aşama, veri bilimi projesinin kurumsal faydaya dönüştüğü son aşama olarak değerlendirilebilir.

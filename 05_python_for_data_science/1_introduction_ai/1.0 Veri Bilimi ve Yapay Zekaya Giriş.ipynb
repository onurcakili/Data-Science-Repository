# Veri Bilimi ve Yapay Zekâya Giriş

### Giriş

Dijital dönüşümün hız kazandığı günümüzde veri, bireylerin, işletmelerin ve devletlerin karar alma süreçlerinde stratejik bir kaynak haline gelmiştir.
İnternet kullanımının yaygınlaşması, mobil cihazlar, sosyal medya platformları, sensör teknolojileri ve Nesnelerin İnterneti (IoT - Internet of Things) gibi teknolojilerin gelişmesi sonucunda her gün milyarlarca veri üretilmektedir.
Üretilen bu veriler; müşteri davranışlarının analiz edilmesinden hastalık teşhislerine, finansal tahminlerden otonom araç sistemlerine kadar çok geniş bir kullanım alanına sahiptir.
Ancak ham veriler tek başlarına anlam ifade etmezler. Verilerin işlenmesi, analiz edilmesi ve karar verme süreçlerinde kullanılabilecek bilgiye dönüştürülmesi gerekmektedir.
Bu noktada Veri Bilimi (Data Science) ve Yapay Zekâ (Artificial Intelligence - AI) kavramları ön plana çıkmaktadır.
Veri bilimi, verilerden anlamlı bilgiler ve öngörüler elde etmeyi amaçlayan disiplinler arası bir alan iken; yapay zekâ, insan zekâsını gerektiren görevleri gerçekleştirebilen sistemlerin geliştirilmesine odaklanmaktadır.

## 1. Veri Kavramı

### Veri Nedir?

Veri (Data), gözlem, ölçüm, deney, anket veya kayıt süreçleri sonucunda elde edilen ve henüz anlamlı bir bağlam içerisinde işlenmemiş ham gerçekler bütünüdür.
Günlük yaşamda karşılaşılan birçok unsur veri örneği olarak değerlendirilebilir:
- Yaş bilgisi
- Hava sıcaklığı
- Ürün fiyatları
- Müşteri isimleri
- Satış adetleri
- Sınav puanları
Veriler tek başlarına anlamlı değildir. Ancak işlendiğinde ve yorumlandığında karar verme süreçlerine katkı sağlayan bilgiye dönüşür.

### Veri, Enformasyon ve Bilgi

Veri biliminin temelini anlayabilmek için veri, enformasyon ve bilgi kavramlarının birbirinden ayrılması gerekir.
Örnek:

### Veri: "38"

### Enformasyon: "Hava sıcaklığı 38°C olarak ölçülmüştür."

### Bilgi: "Sıcaklık mevsim normallerinin üzerindedir."

### Karar: "Dışarı çıkarken gerekli önlemler alınmalıdır."

Bu dönüşüm süreci, ham verilerin karar destek mekanizmalarına dönüşmesini sağlar.

## 2. Veri Türleri

Veriler farklı özelliklerine göre sınıflandırılabilir.

### Niceliksel Veriler (Quantitative Data)

Sayısal olarak ifade edilebilen verilerdir.
Örnekler:
- Yaş
- Maaş
- Boy
- Kilo
- Satış miktarı

### Kesikli Veriler

Tam sayı değerler alır.
Örnek:
- Çalışan sayısı
- Sipariş sayısı

### Sürekli Veriler

Belirli bir aralıkta sonsuz değer alabilir.
Örnek:
- Boy
- Kilo
- Sıcaklık

### Niteliksel Veriler (Qualitative Data)

Sayısal olmayan kategorik verilerdir.
Örnekler:
- Cinsiyet
- Şehir
- Meslek
- Eğitim durumu

## 3. Veri Yapıları

### Yapılandırılmış Veri (Structured Data)

Belirli bir format içerisinde organize edilmiş verilerdir.
Örnekler:
- Excel tabloları
- SQL veritabanları
- ERP kayıtları
- CRM sistemleri
Örnek tablo:
- | Müşteri ID | İsim | Şehir |
| --- | --- | --- |
| 1 | Ahmet | İstanbul |
| 2 | Ayşe | Ankara |
Bu tür veriler kolayca sorgulanabilir ve analiz edilebilir.

### Yarı Yapılandırılmış Veri (Semi-Structured Data)

Belirli kurallara sahip olmakla birlikte tam anlamıyla tablo yapısında olmayan verilerdir.
Örnekler:
- JSON
- XML
- Log dosyaları

### Yapılandırılmamış Veri (Unstructured Data)

Belirli bir veri modeline sahip olmayan verilerdir.
Örnekler:
- Görseller
- Videolar
- Ses kayıtları
- E-postalar
- Sosyal medya içerikleri
Günümüzde üretilen verilerin büyük bölümü yapılandırılmamış veri kategorisine girmektedir.

## 4. Büyük Veri (Big Data)

Veri miktarının hızla artması sonucunda geleneksel veri işleme yöntemleri yetersiz kalmaya başlamıştır. Bu durum Büyük Veri (Big Data) kavramının ortaya çıkmasına neden olmuştur.
Büyük Veri genellikle 5V modeli ile açıklanır:

### Volume (Hacim)

Verinin miktarıdır.

### Velocity (Hız)

Verinin üretilme ve işlenme hızıdır.

### Variety (Çeşitlilik)

Farklı veri türlerinin bir arada bulunmasıdır.

### Veracity (Doğruluk)

Verinin güvenilirlik seviyesidir.

### Value (Değer)

Veriden elde edilen iş değeridir.

## 5. Veri Bilimi

### Veri Bilimi Nedir?

Veri bilimi, yapılandırılmış ve yapılandırılmamış verilerden anlamlı bilgi ve öngörüler elde etmek amacıyla istatistik, matematik, bilgisayar bilimleri ve alan bilgisini bir araya getiren disiplinler arası bir çalışma alanıdır.
Veri biliminin temel amacı:
- Veriyi anlamak
- Veriden bilgi üretmek
- Geleceği tahmin etmek
- Karar verme süreçlerini desteklemek

## 6. Veri Bilimi Süreci

Bir veri bilimi projesi genellikle aşağıdaki aşamalardan oluşur:
- Problemin Tanımlanması
- Veri Toplama
- Veri Temizleme
- Keşifsel Veri Analizi
- Modelleme
- Model Değerlendirme
- Canlı Ortama Alma (Deployment)
- İzleme ve İyileştirme
Veri bilimi projelerinin büyük bölümü veri toplama ve veri temizleme aşamalarında harcanmaktadır.

## 7. Veri Analizi ve Veri Bilimi Arasındaki Fark

Veri analizi ile veri bilimi kavramları sıklıkla karıştırılmaktadır.
Veri analizi daha çok geçmişte yaşanan olayları anlamaya odaklanırken, veri bilimi gelecekte gerçekleşebilecek olayları tahmin etmeyi amaçlar.
Örnek:
Veri Analizi:
- Geçen ay kaç ürün satıldı?
- Hangi mağaza daha yüksek performans gösterdi?
Veri Bilimi:
- Önümüzdeki ay ne kadar satış yapılacak?
- Hangi müşteriler alışveriş yapmayı bırakacak?

## 8. Yapay Zekâ

### Yapay Zekâ Nedir?

Yapay zekâ, çevresini algılayabilen, öğrenebilen, akıl yürütebilen ve belirli hedeflere ulaşmak amacıyla karar verebilen sistemlerin geliştirilmesini konu alan bilgisayar bilimi alanıdır.
Yapay zekânın amacı insanı tamamen değiştirmek değil, insanın karar verme ve problem çözme yeteneklerini desteklemektir.

## 9. Makine Öğrenmesi

Makine Öğrenmesi (Machine Learning), yapay zekânın alt dallarından biridir.
Makine öğrenmesinde sistemler açık şekilde programlanmak yerine geçmiş verilerden örüntüler öğrenir.
Örnekler:
- Ev fiyat tahmini
- Kredi risk analizi
- Kanser teşhisi
- Sahtekârlık tespiti

### Gözetimli Öğrenme

Etiketlenmiş verilerle çalışır.
Örnek:
- Ev fiyat tahmini
- Hastalık teşhisi

### Gözetimsiz Öğrenme

Etiketsiz verilerle çalışır.
Örnek:
- Müşteri segmentasyonu
- Ürün gruplandırması

### Pekiştirmeli Öğrenme

Deneme-yanılma yoluyla öğrenir.
Örnek:
- Oyun oynayan yapay zekâlar
- Robot sistemleri

## 10. Üretken Yapay Zekâ (Generative AI)

Üretken Yapay Zekâ, mevcut verilerden öğrenerek yeni içerikler oluşturabilen sistemleri ifade eder.
Üretebildiği içerikler:
- Metin
- Görsel
- Ses
- Video
- Yazılım kodu
Örnek sistemler:
- ChatGPT
- Claude
- Gemini
- Midjourney
- Sora
Bu sistemler son yıllarda yapay zekânın günlük yaşamda yaygınlaşmasını önemli ölçüde hızlandırmıştır.

## 12. Veri Biliminin İşletmelere Sağladığı Faydalar

Veri bilimi işletmelerin daha doğru ve hızlı kararlar almasına yardımcı olur.
Başlıca avantajları:
- Operasyonel verimlilik
- Maliyet azaltma
- Gelir artırma
- Risk yönetimi
- Müşteri memnuniyeti
- Rekabet avantajı
Veri odaklı karar alma yaklaşımı günümüzde kurumların sürdürülebilir büyümesinde kritik rol oynamaktadır.

## 13. Yapay Zekâ Çağında Gerekli Yetkinlikler

Gelecekte başarılı olabilmek için bireylerin aşağıdaki becerileri geliştirmesi önem taşımaktadır:
- Veri Okuryazarlığı
- Analitik Düşünme
- İstatistiksel Düşünme
- Problem Çözme
- Programlama
- Yapay Zekâ Araçlarını Kullanabilme
- Sürekli Öğrenme Yeteneği
Teknolojik gelişmelerin hızlandığı günümüzde öğrenmeyi öğrenmek, teknik becerilerin kendisi kadar önemli hale gelmiştir.

### Sonuç

Veri bilimi ve yapay zekâ, dijital çağın en önemli disiplinleri arasında yer almaktadır. Her geçen gün artan veri miktarı, kurumları ve bireyleri veri odaklı düşünmeye zorlamaktadır. Bu nedenle veri okuryazarlığı, analitik düşünme ve yapay zekâ teknolojilerini etkin kullanabilme becerileri geleceğin en değerli yetkinlikleri arasında değerlendirilmektedir.
Günümüzde veri yalnızca teknolojik bir unsur değil, aynı zamanda ekonomik ve stratejik bir değerdir. Bu nedenle veri bilimi ve yapay zekâ alanlarında temel bilgi sahibi olmak, farklı sektörlerde çalışan bireyler için önemli bir avantaj sağlamaktadır.

# 1.2) Veri Bilimi Süreci

## Özet

Veri bilimi süreci, ham verinin sistematik yöntemlerle işlenerek anlamlı bilgiye, öngörüye ve karar destek çıktısına dönüştürülmesini ifade eder.

Bu süreç yalnızca algoritma geliştirme faaliyeti değildir; problem tanımlama, veri toplama, veri kalitesi yönetimi, keşifsel analiz, modelleme, değerlendirme, yorumlama, raporlama ve gerektiğinde modelin üretim ortamına alınması gibi birbirini tamamlayan aşamalardan oluşur.

Modern veri bilimi uygulamalarında amaç, veriden yalnızca istatistiksel sonuçlar üretmek değil, bu sonuçları iş problemiyle ilişkilendirerek ölçülebilir fayda yaratmaktır. 
Bu nedenle veri bilimi süreci teknik, analitik ve iletişimsel becerilerin birlikte kullanılmasını gerektirir.

## Veri Bilimi Sürecinin Kapsamı

Veri bilimi süreci, belirli bir problem veya araştırma sorusu etrafında şekillenir.  
Sürecin başlangıç noktası genellikle "hangi problem çözülmek isteniyor?" sorusudur.  
Bu soru net biçimde tanımlanmadan veri toplama veya modelleme aşamasına geçmek; yanlış metriklerin izlenmesine, hatalı modellerin geliştirilmesine ve karar vericiler için düşük değerli çıktılar üretilmesine neden olabilir.

Bu bağlamda veri bilimi süreci üç temel eksende değerlendirilir:

- İş veya araştırma probleminin doğru tanımlanması
- Verinin güvenilir, tutarlı ve analiz edilebilir hale getirilmesi
- Analitik çıktının yorumlanabilir, ölçülebilir ve uygulanabilir olması

## Veri Bilimci Kimdir?

Veri bilimci, verilerden anlamlı bilgi çıkarabilen, analitik yöntemleri iş problemleriyle ilişkilendirebilen ve elde edilen sonuçları karar alma süreçlerine aktarabilen uzman kişidir.  
Veri bilimci yalnızca model kuran kişi değildir; aynı zamanda problem alanını anlayan, veri kaynaklarını değerlendiren, varsayımları test eden ve sonuçların güvenilirliğini sorgulayan analitik bir araştırmacıdır.

Bir veri bilimcinin temel sorumlulukları şunlardır:

- Problemi analitik olarak tanımlamak
- İlgili veri kaynaklarını belirlemek
- Veriyi toplamak, temizlemek ve dönüştürmek
- Keşifsel veri analizi yapmak
- Uygun istatistiksel veya makine öğrenmesi yöntemlerini seçmek
- Model performansını değerlendirmek
- Sonuçları teknik ve teknik olmayan paydaşlara açıklamak
- Analitik çıktının karar süreçlerinde kullanılabilirliğini sağlamak

## Veri Bilimcinin Yetkinlikleri

Veri bilimci profili disiplinler arası bir uzmanlık gerektirir.  
Bu yetkinlikler genel olarak teknik beceriler, analitik düşünme becerileri ve iletişim becerileri altında incelenebilir.

### Teknik Beceriler

Veri bilimcinin teknik becerileri; programlama, istatistik, veri tabanı yönetimi, makine öğrenmesi ve veri görselleştirme alanlarını kapsar. Python, R, SQL, veri işleme kütüphaneleri, makine öğrenmesi çerçeveleri ve bulut tabanlı veri platformları bu beceri setinin önemli bileşenleridir.

### Analitik Beceriler

Analitik beceriler, verideki örüntüleri fark etme, hipotez kurma, değişkenler arasındaki ilişkileri değerlendirme ve belirsizlik altında çıkarım yapma yeteneğini içerir. Bu beceriler, yalnızca araç kullanımından ibaret değildir; doğru soruyu sormayı ve elde edilen sonuçları eleştirel biçimde yorumlamayı gerektirir.

### İletişim ve Alan Bilgisi

Veri bilimi çıktıları, karar vericiler tarafından anlaşılabilir olmadığında uygulama değeri azalır.  
Bu nedenle veri bilimci, teknik sonuçları sade, tutarlı ve iş bağlamına uygun biçimde aktarabilmelidir.

## Veri Bilimi Süreci Nedir?

Veri bilimi süreci, bir problemin veri odaklı yöntemlerle çözülmesi için izlenen sistematik iş akışıdır.  
Literatürde bu süreç farklı çerçevelerle açıklanır. En yaygın yaklaşımlardan biri OSEMN modelidir.  
OSEMN; Obtain, Scrub, Explore, Model ve Interpret aşamalarından oluşur. Gerçek projelerde bu süreç çoğu zaman yinelemelidir; veri temizleme, modelleme ve yorumlama aşamaları birbirini geri besler.

## 1) Problemin Tanımlanması

Her veri bilimi çalışması açık biçimde tanımlanmış bir problemle başlamalıdır. Problem tanımı; hedef değişkeni, başarı ölçütlerini, veri kapsamını, kısıtları ve beklenen çıktıyı içermelidir.  
Örneğin "müşteri kaybını azaltmak" genel bir amaçtır; veri bilimi açısından bu amaç, "aboneliğini iptal etme olasılığı yüksek müşterileri tahmin eden bir sınıflandırma modeli geliştirmek" biçiminde teknikleştirilebilir.

## 2) Veri Toplama

Veri toplama aşaması, problem için gerekli verilerin belirlenmesi ve uygun kaynaklardan elde edilmesini kapsar. Veri kaynakları kurumsal veri tabanları, API servisleri, sensör verileri, web kaynakları, anketler, deneysel ölçümler veya açık veri platformları olabilir.  
Bu aşamada verinin güvenilirliği, güncelliği, kapsamı, temsil gücü ve yasal kullanım koşulları değerlendirilmelidir.

## 3) Veri Temizleme ve Ön İşleme

Veri temizleme, ham verinin analiz ve modelleme için uygun hale getirilmesi sürecidir. Eksik değerlerin ele alınması, yinelenen kayıtların tespit edilmesi, aykırı değerlerin incelenmesi, veri tiplerinin düzeltilmesi, tarih ve kategori formatlarının standartlaştırılması, gereksiz değişkenlerin çıkarılması, kategorik değişkenlerin kodlanması ve sayısal değişkenlerin ölçeklendirilmesi bu aşamanın temel bileşenleridir.

## 4) Keşifsel Veri Analizi

Keşifsel veri analizi, verinin genel yapısını, dağılımlarını, ilişkilerini ve potansiyel örüntülerini anlamak için yapılan ön analiz sürecidir. Betimsel istatistikler, görselleştirme teknikleri ve değişkenler arası ilişki analizleri bu aşamada kullanılır.

## 5) Özellik Mühendisliği

Özellik mühendisliği, ham değişkenlerden model performansını artırabilecek yeni değişkenler türetme sürecidir.  
Modelin başarısı çoğu zaman algoritma seçiminden çok, verinin nasıl temsil edildiğine bağlıdır.

## 6) Modelleme

Modelleme aşamasında problem türüne uygun istatistiksel veya makine öğrenmesi yöntemleri uygulanır.  
Problem bir sınıflandırma, regresyon, kümeleme, zaman serisi tahmini, öneri sistemi veya doğal dil işleme problemi olabilir.  
Model seçiminde performans, yorumlanabilirlik, hesaplama maliyeti ve üretim ortamına alınabilirlik birlikte değerlendirilmelidir.

## 7) Model Değerlendirme

Model değerlendirme, geliştirilen modelin başarısını ölçme ve güvenilirliğini analiz etme aşamasıdır.  
Sınıflandırma problemlerinde doğruluk, kesinlik, duyarlılık, F1 skoru ve ROC-AUC; regresyon problemlerinde MAE, MSE, RMSE ve R-kare gibi ölçütler kullanılabilir.  
Bunun yanında modelin yorumlanabilirliği, yanlılık içerip içermediği ve iş bağlamında uygulanabilirliği de incelenmelidir.

## 8) Yorumlama ve Raporlama

Veri bilimi sürecinin en kritik aşamalarından biri sonuçların yorumlanmasıdır.  
Raporlama aşamasında problem tanımı, kullanılan veri kaynakları, veri temizleme adımları, temel bulgular, modelleme yöntemi, performans metrikleri, modelin güçlü ve zayıf yönleri, varsayımlar, sınırlılıklar ve önerilen aksiyonlar açık biçimde sunulmalıdır.

## 9) Üretime Alma ve İzleme

Gerçek dünyadaki veri bilimi projelerinde modelin geliştirilmesi sürecin sonu değildir.  
Modelin bir iş sürecine, uygulamaya veya karar destek sistemine entegre edilmesi gerekir.  
Üretime alınan modeller zaman içinde performans kaybı yaşayabilir; bu nedenle model izleme, yeniden eğitim ve performans takibi veri bilimi yaşam döngüsünün parçalarıdır.

## Veri Biliminde Kullanılan Temel Teknikler

### Sınıflandırma

Sınıflandırma, dağınık verilerin önceden tanımlanmış sınıflara atanmasını amaçlayan denetimli öğrenme yaklaşımıdır.

### Regresyon

regresyon, birbiriyle ilişkili değişkenler arasındaki neden-sonuç ilişkisini modellemek ve tahmin yapmak için kullanılan bir yöntemdir.  
Satış miktarı, konut fiyatı, sıcaklık değeri veya talep miktarı gibi çıktılar regresyon problemlerine örnektir.

### Kümeleme

Kümeleme, etiketlenmemiş verilerde benzer gözlemleri gruplandırmak için kullanılan denetimsiz öğrenme yöntemidir.

### Korelasyon ve Nedensellik

Korelasyon, iki değişken arasındaki ilişkinin yönünü ve gücünü ifade eder.  
Ancak korelasyon nedensellik anlamına gelmez. Bu ayrım, veri bilimi çalışmalarında hatalı çıkarımlardan kaçınmak için kritik öneme sahiptir.

## Veri Bilimi Sürecinde Kullanılan Teknolojiler

Veri bilimi uygulamalarında Python, R, SQL, bulut bilişim platformları, büyük veri teknolojileri, makine öğrenmesi kütüphaneleri ve görselleştirme araçları yaygın olarak kullanılır. Yapay zekâ, IoT, bulut bilişim ve otomasyon sistemleri veri bilimi projelerinin kapsamını genişletmektedir.

## Sonuç

Veri bilimi süreci, ham veriden anlamlı ve uygulanabilir bilgi üretmeye yönelik sistematik, yinelemeli ve disiplinler arası bir yaklaşımdır. Başarılı bir veri bilimi projesi yalnızca güçlü algoritmalara değil; doğru problem tanımına, kaliteli veriye, uygun metodolojiye, yorumlanabilir sonuçlara ve etkili iletişime dayanır.

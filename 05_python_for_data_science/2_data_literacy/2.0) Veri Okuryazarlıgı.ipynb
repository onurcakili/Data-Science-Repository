# 2.0) Veri Okuryazarlığına Giriş
## Introduction to Data Literacy

Günümüzde veri, bireysel karar alma süreçlerinden kurumsal stratejilere kadar pek çok alanda merkezi bir rol  
oynamaktadır.  
Bu nedenle verinin yalnızca toplanması değil; doğru okunması, yorumlanması, analiz edilmesi ve etkili biçimde  
aktarılması da büyük önem taşımaktadır.  
Bu becerilerin bütünü **veri okuryazarlığı** kavramı altında ele alınır.

Özellikle veri bilimi alanında çalışan kişiler için veri okuryazarlığı temel bir yetkinliktir.  
Çünkü veri bilimciler yalnızca veriyi analiz etmekle kalmaz; aynı zamanda elde ettikleri bulguları farklı  
hedef kitlelere açık, doğru ve anlaşılır biçimde aktarmakla da sorumludur.

## Veri Okuryazarlığı Nedir?

Veri okuryazarlığı; verileri okuyabilme, verilerle çalışabilme, verileri analiz edebilme ve veriler üzerinden  
anlamlı tartışmalar yürütebilme becerisi olarak tanımlanabilir.  
Başka bir ifadeyle veri okuryazarlığı, veriyi yalnızca teknik bir çıktı olarak değil, karar alma süreçlerini  
destekleyen anlamlı bir bilgi kaynağı olarak değerlendirebilme yetkinliğidir.

Knight'a göre veri okuryazarlığı, bireylerin verilerle etkili biçimde etkileşim kurmasını ve veriden anlam  
üretmesini sağlayan temel bir beceri alanıdır.

> Kaynak: Knight, Michelle. “What Is Data Literacy?” 14 Ekim 2020.

## Veri, Enformasyon, Bilgi ve Bilgelik

Veri okuryazarlığını anlayabilmek için öncelikle **veri**, **enformasyon**, **bilgi** ve **bilgelik**  
kavramları arasındaki farkı netleştirmek gerekir.  
Bu kavramlar çoğu zaman birbirinin yerine kullanılsa da aslında farklı anlam düzeylerini ifade eder.

- **Veri (Data):** Ham, işlenmemiş ve tek başına anlamı sınırlı olan gözlem veya ölçüm değerleridir.
- **Enformasyon (Information):** Verilerin düzenlenmiş, sınıflandırılmış veya bağlam kazandırılmış halidir.
- **Bilgi (Knowledge):** Enformasyonun yorumlanması, ilişkilendirilmesi ve anlamlı çıkarımlara  
  dönüştürülmesiyle elde edilir.
- **Bilgelik (Wisdom):** Bilginin deneyim, muhakeme ve etik değerlendirme ile eyleme dönüştürülmüş halidir.

### Bilgi Piramidi

Bilgi piramidi veya bilgi hiyerarşisi, veriden bilgeliğe doğru ilerleyen anlam üretme sürecini açıklar:

- **Veri:** Toplanır ve organize edilir.
- **Enformasyon:** Analiz edilir ve özetlenir.
- **Bilgi:** Karar verme süreçlerinde kullanılır ve sentezlenir.
- **Bilgelik:** Bilginin doğru bağlamda uygulanmasını sağlar.

Örneğin:

- “50” ifadesi tek başına bir **veri**dir.
- “50 metre” ifadesi veriye bağlam kazandırdığı için **enformasyon** niteliği taşır.
- “50 metrenin belirli bir kullanım amacı için uzun veya kısa olup olmadığını değerlendirmek” **bilgi**  
  üretimidir.
- Bu değerlendirmeye dayanarak en uygun uzunluğu belirlemek ve uygulamak ise **bilgelik** düzeyine karşılık  
  gelir.

## Veri Okuryazarlığı Neden Önemlidir?

Veriler, günlük yaşamda ve iş hayatında giderek daha fazla kararın temelini oluşturmaktadır.  
Bireyler ve kurumlar, çoğu zaman farkında olmadan verilerle etkileşim kurar; bu verileri yorumlar,  
karşılaştırır ve sonuçlara ulaşır.

Veri okuryazarlığı şu tür durumlarda kritik önem taşır:

- Bir analistin gelecek ay için stok miktarını tahmin etmesinde,
- Bir doktorun tahlil sonuçlarını hastasına açık ve anlaşılır biçimde aktarmasında,
- Bir araştırmacının anket verilerini değerlendirerek sonuca ulaşmasında,
- Bir siyasi partinin sandık bazlı oy oranlarını yorumlamasında,
- TÜİK gibi ulusal veya uluslararası veri kaynaklarının doğru okunmasında.

Bu örnekler, veri okuryazarlığının yalnızca teknik meslekler için değil, farklı disiplinler ve gündelik yaşam  
için de temel bir beceri olduğunu göstermektedir.

## Veri Okuryazarlığı İçin Temel Gereksinimler

School of Data tarafından açıklanan veri okuryazarlığı yaklaşımına göre bireylerin veriyle etkili biçimde  
çalışabilmesi için bazı temel becerilere sahip olması gerekir:

- Bilgiye ve veriye farklı kaynaklardan ulaşabilmek,
- Veriye doğru sorular yöneltebilmek,
- Veriden anlamlı yanıtlar çıkarabilmek,
- Veride belirli örüntüler, hikayeler veya görselleştirme fırsatları bulabilmek,
- Veriyi kişisel, akademik veya mesleki hedefler doğrultusunda kullanabilmek,
- Veriyle çalışırken rahat ve eleştirel bir tutum geliştirebilmek,
- Temel istatistiksel analizleri uygulayabilmek.

Bu gereksinimler, veri okuryazarlığının yalnızca veri okuma becerisinden ibaret olmadığını; aynı zamanda  
sorgulama, analiz etme, yorumlama ve iletişim kurma becerilerini de içerdiğini göstermektedir.

## Veri Okuryazarlığı ve İstatistik Bilgisi

Veri okuryazarlığı ile istatistik bilgisi arasında güçlü bir ilişki vardır.  
İstatistiksel düşünme becerisi, verilerin anlamlandırılması ve güvenilir sonuçlara dönüştürülmesi açısından  
temel bir araçtır.

İstatistiksel bilgiler sayesinde günlük yaşamda, bilimsel çalışmalarda ve kurumsal karar alma süreçlerinde  
birçok durum daha açık, ölçülebilir ve karşılaştırılabilir hale gelir.  
Eğer istatistiksel yöntemler kullanılmasaydı, birçok veri kümesi yalnızca dağınık gözlemlerden ibaret kalır ve  
anlamlı biçimde yorumlanamazdı.

Veri hacminin artmasıyla birlikte verilerin özetlenmesi, sınıflandırılması, genellenmesi ve geleceğe yönelik  
çıkarımlar için kullanılması daha önemli hale gelmiştir.  
Bu noktada istatistik bilimi, verinin bilgiye dönüştürülmesinde temel bir rol üstlenir.

## İstatistik Bilimi Nedir?

İstatistik, verilerin doğru biçimde toplanması, düzenlenmesi, analiz edilmesi, yorumlanması ve bilgiye  
dönüştürülmesiyle ilgilenen bilim dalıdır.  
Başka bir ifadeyle istatistik, gözlemlerden öğrenmeyi ve bu gözlemleri karar alma süreçlerinde  
kullanılabilecek anlamlı bilgilere dönüştürmeyi sağlar.

## Örnek İstatistik Kullanımı

Veri görselleştirme, istatistiksel bilgilerin daha anlaşılır ve etkili biçimde sunulmasını sağlayan önemli bir  
iletişim aracıdır.  
Bu durumun erken örneklerinden biri, 1930 yılında bir gazetede yayımlanan görsel istatistik anlatımıdır.

Söz konusu gazete görselinde, tasarruf etmemenin ilerleyen yaşlarda ekonomik güvencesizlik ve yoksulluk riski  
doğurabileceği anlatılmaktadır.  
Grafik, yalnızca metinsel bir uyarı yapmak yerine istatistiksel ifadeleri görsel unsurlarla destekleyerek  
mesajın etkisini artırmaktadır.

Görselde öne çıkan temel ifadeler şunlardır:

- Çocuklukta ve gençlikte tasarruf etmeyen bireylerin ilerleyen yaşlarda yoksulluk riskiyle karşılaşabileceği  
  vurgulanmaktadır.
- “Uçurum tehlikesi” ifadesiyle ekonomik güvencesizliğin ciddi sonuçlarına dikkat çekilmektedir.
- “Bu yaşta tasarruftan istifade eden 5000'de 1 kişidir.” ifadesiyle istatistiksel bir oran üzerinden mesaj  
  güçlendirilmektedir.
- Görselin ana mesajı, geleceğini planlamayan bireylerin ilerleyen yaşlarda ekonomik zorluklarla  
  karşılaşabileceğidir.

Bu örnek, istatistiksel bilginin yalnızca sayısal ifadelerle değil, görsel anlatım teknikleriyle de güçlü  
biçimde aktarılabileceğini göstermektedir.  
Veri görselleştirme, karmaşık veya soyut bilgileri daha anlaşılır, akılda kalıcı ve ikna edici hale getiren  
önemli bir veri okuryazarlığı aracıdır.

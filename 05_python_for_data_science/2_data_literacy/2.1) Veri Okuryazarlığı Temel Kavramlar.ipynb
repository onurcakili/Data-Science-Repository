# 2.1) Veri Okuryazarlığı Temel Kavramları

Bu bölümde veri okuryazarlığı için gerekli temel istatistiksel kavramlar ele alınmaktadır.  
Popülasyon, örneklem, gözlem birimi, değişken türleri ve ölçek türleri; veriyi doğru tanımlama, analiz etme ve  
yorumlama süreçlerinin temel bileşenleridir.

Veri okuryazarlığı yalnızca veriyi okumakla sınırlı değildir.  
Aynı zamanda verinin hangi evrenden geldiğini, hangi örneklem üzerinden temsil edildiğini, hangi  
değişkenlerden oluştuğunu ve bu değişkenlerin hangi ölçüm düzeyinde değerlendirildiğini bilmeyi gerektirir.

## Popülasyon ve Örneklem

İstatistiksel bir çalışmada **popülasyon**, araştırmanın ilgilendiği tüm birimleri kapsayan ana kütleyi ifade eder.  
**Örneklem** ise bu ana kütleden seçilen ve popülasyonu temsil etmesi beklenen daha küçük alt kümedir.

### Popülasyon

Popülasyon, hakkında istatistiksel çıkarım yapılmak istenen tüm birimlerin oluşturduğu evrendir.  
Örneğin Türkiye'deki tüm üniversiteler bir araştırmanın popülasyonu olarak kabul edilebilir.

### Örneklem

Örneklem, popülasyon içinden seçilen ve popülasyonun temel özelliklerini yansıtması beklenen alt kümedir.  
Örneğin Türkiye'nin batısında bulunan üniversiteler, Türkiye'deki tüm üniversiteler popülasyonundan seçilmiş  
bir örneklem olarak değerlendirilebilir.

Bir örneklemin güvenilir sonuçlar üretmesi için popülasyonu yeterli düzeyde temsil etmesi ve mümkün olduğunca  
yansız biçimde seçilmesi gerekir.  
Temsil gücü düşük veya yanlı bir örneklem, istatistiksel analizlerin hatalı sonuçlar üretmesine neden  
olabilir.

### Seçim Anketi Örneği

Seçim dönemlerinde siyasi partilerin oy oranları çoğu zaman anketler aracılığıyla tahmin edilir.  
Bu anketler genellikle tüm seçmenlere değil, seçmen kitlesini temsil ettiği varsayılan yaklaşık 2000-2500  
kişilik örneklemlere uygulanır.

Bu nedenle anket sonuçlarında çoğunlukla güven düzeyi veya güven aralığı gibi ifadeler yer alır.  
Eğer örneklem belirli bir partinin güçlü olduğu bölgelerden seçilirse sonuçlar o parti lehine yanlı olabilir.  
Benzer biçimde, aynı partinin zayıf olduğu bölgelerden seçilen bir örneklem de sonuçları ters yönde  
saptırabilir.

Bu durum, veri bilimcinin üzerinde çalıştığı veri setini nesnel biçimde değerlendirmesi gerektiğini gösterir.  
Analizden önce örneklemin popülasyonu ne ölçüde temsil ettiği mutlaka sorgulanmalıdır.

## Gözlem Birimi

**Gözlem birimi**, araştırmada veri toplanan en küçük analiz birimidir.  
Başka bir ifadeyle örneklemi oluşturan her bir birey, kurum, nesne veya olay gözlem birimi olarak kabul  
edilir.

Örneğin Türkiye'deki üniversiteler popülasyon, Türkiye'nin batısındaki üniversiteler örneklem olarak kabul  
edilirse, Boğaziçi Üniversitesi bu örneklem içindeki bir gözlem birimini ifade eder.

## Değişken ve Değişken Türleri

Değişkenler, gözlem birimleri hakkında ölçülen veya sınıflandırılan özelliklerdir.  
Bir veri setinde yer alan “araç fiyatı”, “kilometre”, “vites türü”, “hasar durumu”, “marka” ve “model” gibi  
sütunlar değişkenlere örnektir.

Değişkenler genel olarak iki ana grupta incelenir:

- **Sayısal değişkenler:** Nicel özellikleri ifade eder. Araç fiyatı ve kilometre bu gruba örnektir.
- **Kategorik değişkenler:** Nitel sınıfları ifade eder. Vites türü, hasar durumu, marka ve model bu gruba  
  örnektir.

Örneğin “vites türü” kategorik bir değişkendir; “otomatik” ve “manuel” ise bu değişkenin sınıflarıdır.

## Ölçek Türleri

Ölçek türleri, değişkenlerin hangi düzeyde ölçüldüğünü ve bu değişkenler üzerinde hangi istatistiksel  
işlemlerin anlamlı biçimde yapılabileceğini belirler.

Veri türleri ve ölçekler genel olarak şu şekilde sınıflandırılabilir:

- **Kategorik veri**
  - Nominal ölçek
  - Ordinal ölçek
- **Sayısal veri**
  - Aralık ölçeği
  - Oran ölçeği

### Sayısal Değişkenlerde Ölçek Türleri

**Aralık ölçeği**, başlangıç noktası mutlak sıfır olmayan sayısal değişkenler için kullanılır.  
Sıcaklık bu ölçek türüne örnektir; çünkü sıcaklık hem negatif hem pozitif değerler alabilir.

**Oran ölçeği**, başlangıç noktası gerçek sıfır olan sayısal değişkenler için kullanılır.  
Araç hızı buna örnek verilebilir; çünkü hız negatif değer alamaz ve sıfır değeri gerçek yokluğu ifade eder.

### Kategorik Değişkenlerde Ölçek Türleri

**Nominal ölçek**, sınıflar arasında doğal bir sıralama bulunmayan kategorik değişkenleri ifade eder.  
Cinsiyet değişkeni bu ölçek türüne örnek verilebilir.  
“Kadın” ve “Erkek” sınıfları arasında sayısal veya sıralı bir üstünlük ilişkisi yoktur.

**Ordinal ölçek**, sınıflar arasında doğal bir sıralama bulunan kategorik değişkenleri ifade eder.  
Rütbe değişkeni bu ölçek türüne örnektir:

- Onbaşı
- Yüzbaşı
- Binbaşı
- Albay

Bu sınıflar arasında hiyerarşik bir sıralama bulunduğu için rütbe değişkeni ordinal ölçekte değerlendirilir.

# 2.2.1) Merkezi Eğilim Ölçüleri: Örnek Senaryo

Bu örnek senaryonun amacı, merkezi eğilim ölçülerinin günlük yaşamda ve veri bilimi süreçlerinde nasıl  
kullanılabileceğini göstermektir.  
Özellikle aritmetik ortalama ve medyanın temsil gücü açısından nasıl farklı sonuçlar verebildiği  
incelenmektedir.

Senaryoda A ve B olmak üzere iki emlak şirketi ele alınmaktadır.  
Her iki şirket de üst segment müşterilere hitap ettiğini iddia etmektedir.  
Bu iddianın veri okuryazarlığı yaklaşımıyla değerlendirilmesi için şirketlerin ilan fiyatları  
karşılaştırılacaktır.

## Veri Seti

| A - Emlak | B - Emlak |
|---:|---:|
| 900000 | 400000 |
| 670000 | 480000 |
| 800000 | 360000 |
| 700000 | 700000 |
| 500000 | 240000 |
| 700000 | 310000 |
| 890000 | 400000 |
| 550000 | 3500000 |
| 640000 | 200000 |

## Aritmetik Ortalama Karşılaştırması

İlk aşamada her iki şirketin ilan fiyatları için aritmetik ortalama hesaplanır:

| Şirket | Aritmetik Ortalama |
|---|---:|
| A - Emlak | 706000 |
| B - Emlak | 730000 |

Yalnızca aritmetik ortalamaya bakıldığında B - Emlak şirketinin portföyü daha yüksek fiyatlı görünmektedir.  
Ancak bu sonuç tek başına yeterli değildir; çünkü aritmetik ortalama aykırı değerlere duyarlıdır.

## Medyan ile Doğrulama

Aritmetik ortalamanın veri setini ne ölçüde temsil ettiğini değerlendirmek için medyan değerleri de hesaplanmalıdır.

| Ölçü | A - Emlak | B - Emlak |
|---|---:|---:|
| Ortalama | 706000 | 730000 |
| Medyan | 700000 | 400000 |

A - Emlak şirketinde ortalama ve medyan birbirine oldukça yakındır.  
Bu durum, portföy fiyatlarının daha homojen dağıldığını ve aritmetik ortalamanın veri setini görece iyi temsil  
ettiğini gösterir.

B - Emlak şirketinde ise ortalama `730000`, medyan ise `400000` değerindedir.  
İki ölçü arasındaki büyük fark, veri setinde aykırı bir değer bulunduğunu göstermektedir.  
Bu farkın temel nedeni B - Emlak portföyündeki `3500000` TL değerindeki ilandır.

## Yorum

Bu senaryoda iki şirket karşılaştırılırken yalnızca aritmetik ortalamaya dayanmak yanıltıcı olur.  
B - Emlak şirketindeki yüksek fiyatlı tek ilan, ortalamayı yukarı çekmekte ve portföyün genel yapısını  
olduğundan daha yüksek göstermektedir.

Bu nedenle, aykırı değer içeren veri setlerinde merkezi eğilimi temsil etmek için medyan daha güvenilir bir  
ölçüdür.  
A - Emlak için ortalama ve medyanın birbirine yakın olması, bu şirketin portföy yapısının daha tutarlı  
olduğunu göstermektedir.  
B - Emlak için ise medyan değeri, portföyün genel düzeyini ortalamadan daha doğru temsil etmektedir.

## Sonuç

Veri okuryazarlığı açısından bu örnek, tek bir istatistiksel ölçüye dayanarak karar vermenin riskli  
olabileceğini göstermektedir.  
Bir veri seti değerlendirilirken ortalama, medyan ve aykırı değerler birlikte ele alınmalıdır.  
Böylece veri setinin gerçek yapısı daha doğru biçimde yorumlanabilir.

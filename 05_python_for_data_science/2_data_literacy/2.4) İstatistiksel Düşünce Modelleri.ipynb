# 2.4) İstatistiksel Düşünce Modelleri

İstatistiksel düşünce modelleri, veri okuryazarlığından veri analitiğine uzanan süreçte bireylerin veriyi  
sistematik biçimde tanımlamasına, düzenlemesine, göstermesine, analiz etmesine ve yorumlamasına yardımcı olan  
kavramsal çerçevelerdir.

Bu modeller, veriye yalnızca teknik bir çıktı olarak değil, karar alma süreçlerinde anlam üreten bir kaynak  
olarak yaklaşmayı sağlar.

## İstatistiksel Düşünce Modellerine Genel Bakış

Literatürde istatistiksel düşünceyi açıklamak için farklı modeller geliştirilmiştir. Görselde yer alan başlıca çalışmalar şunlardır:

- Ben-Zvi ve Friedlander (1997)
- Jones ve diğerleri (2000)
- Wild ve Pfannkuch (1999)
- Hoerl ve Snee (2001)
- Mooney (2002)

Ben-Zvi ve Friedlander ile Jones ve diğerlerinin modelleri daha çok eğitim ortamlarında istatistiksel  
düşünmeyi geliştirmeye yöneliktir.  
İş dünyası ve analitik süreçler açısından ise Mooney modeli daha uygulanabilir ve sade bir yapı sunmaktadır.

## Mooney Veri Modeli

Mooney veri modeli, veriye dayalı düşünme sürecini aşamalı biçimde ele alan bir yaklaşımdır.  
Model, özellikle iş problemlerinin veriyle analiz edilmesi ve yorumlanması açısından kullanışlıdır.

Modelin temel aşamaları şunlardır:

1. Verinin tanımlanması
2. Verinin organize edilmesi ve indirgenmesi
3. Veri gösterimi
4. Verinin analiz edilmesi ve yorumlanması

### 1. Verinin Tanımlanması

Verinin tanımlanması, çalışmanın amacını, değişkenlerini, gözlem birimlerini ve ölçülmek istenen özellikleri  
açık biçimde belirleme sürecidir.  
Bu aşama, analiz sürecinin doğru kurulabilmesi için kritik öneme sahiptir.

### 2. Verinin Organize Edilmesi ve İndirgenmesi

Verinin organize edilmesi ve indirgenmesi, ham verinin daha kolay erişilebilir, sınıflandırılabilir ve analiz  
edilebilir hale getirilmesini ifade eder.  
Bu aşamada sınıflandırma, gruplama, özetleme ve veri temizleme işlemleri yapılabilir.

### 3. Veri Gösterimi

Veri gösterimi, verilerin tablo, grafik veya görsel araçlarla daha anlaşılır hale getirilmesidir.  
Etkili veri gösterimi, karmaşık bilgilerin daha hızlı kavranmasını ve içgörü üretilmesini sağlar.

### 4. Verinin Analiz Edilmesi ve Yorumlanması

Verinin analiz edilmesi ve yorumlanması, işlenmiş veriden anlamlı sonuçlar çıkarma sürecidir.  
Bu aşamada istatistiksel yöntemlerden yararlanılarak karar alma süreçlerini destekleyen bulgular üretilir.

## İstatistiksel Düşünce Düzeyleri

İstatistiksel düşünce farklı yeterlik düzeylerinde değerlendirilebilir:

- **Kişiye özgülük - Seviye 1:** Veriyi okuyamaz veya sınırlı biçimde algılar.
- **Geçici düzey - Seviye 2:** Veriyi kısmen okuyabilir ancak sistematik analiz yapamaz.
- **Nicel düzey - Seviye 3:** Veriyi okuyabilir ve temel nicel yorumlar yapabilir.
- **Analitik düzey - Seviye 4:** Veriyi okuyabilir, analiz edebilir ve sonuçları yorumlayabilir.

## Örnek Senaryo: İnternet Kullanımı Araştırması

Bir şirket internet kullanımıyla ilgili yaptığı araştırmada şu bilgileri paylaşmıştır:

Anketi yanıtlayan 2000 kişinin %43,4'ünü erkekler, %66,4'ünü kadınlar oluşturmaktadır.  
Anketi yanıtlayanların %80'i 15-27 yaş aralığındadır.  
Kadınların %72'si İngilizce bilmektedir.  
Erkekler günde ortalama 3 saat, kadınlar ise 4 saat internette zaman geçirmektedir.  
Erkeklerin %72'si, kadınların ise %75'i üniversite mezunudur.

### Değişkenlerin Belirlenmesi

Bu çalışmada ölçülmeye çalışılan temel değişkenler şunlardır:

1. Cinsiyet
2. Yaş
3. İngilizce bilme
4. İnternette geçirilen zaman
5. Eğitim durumu

### Değişken Türleri

Değişkenler türlerine göre şu şekilde sınıflandırılabilir:

1. Cinsiyet - Kategorik
2. Yaş - Sayısal
3. İngilizce bilme - Kategorik
4. İnternette geçirilen zaman - Sayısal
5. Eğitim durumu - Kategorik

### Ölçek Türleri

Değişkenlerin ölçek türleri şu şekilde değerlendirilebilir:

1. Cinsiyet - Kategorik - Nominal
2. Yaş - Sayısal - Oran
3. İngilizce bilme - Kategorik - Nominal
4. İnternette geçirilen zaman - Sayısal - Oran
5. Eğitim durumu - Kategorik - Ordinal

### Veri Kalitesi Açısından Değerlendirme

Metinde dikkat çeken bazı sorunlar bulunmaktadır.  
Erkek ve kadın oranlarının toplamı %100'ü aşmaktadır.  
Bu durum, verinin tanımlanması veya raporlanması aşamasında hata olabileceğini gösterir.

Ayrıca “Kadınların %72'si İngilizce bilmektedir.” ifadesi tek başına yeterli değildir; çünkü erkeklerin  
İngilizce bilme oranı verilmemiştir.  
Bu nedenle cinsiyete göre karşılaştırmalı bir yorum yapmak mümkün değildir.

Buna karşılık, yaş aralığı ve eğitim durumu gibi bilgiler hedef kitlenin profili hakkında anlamlı çıkarımlar  
yapılmasına imkan tanır.  
Örneğin anketi yanıtlayanların büyük bölümünün 15-27 yaş aralığında olması, araştırmanın genç bir kitle  
üzerinde yoğunlaştığını gösterir.

Bu nedenle verilerin belirli gruplar altında toplanması, sınıflandırılması ve özetlenmesi gerekir.

Aynı zamanda eksik, tutarsız veya gürültülü veriler analizden önce daha kolay tespit edilebilir.

## Veri Gösterimi

Veri gösterimi, analiz sonuçlarının doğru anlaşılması açısından kritik öneme sahiptir.  
Grafiklerde eksen seçimi, başlangıç noktası ve ölçek aralıkları, verinin algılanış biçimini doğrudan etkiler.

### Ortalama Ev Fiyatları Örneği

Bir medya şirketinde ortalama ev fiyatları için iki farklı grafik oluşturulmuştur.  
Grafik 1 “Ev fiyatlarında büyük artış!” yorumunu desteklerken, Grafik 2 “Ev fiyatları geçen yıla göre artış  
göstermiştir.” yorumunu desteklemektedir.

Grafik 1'de dikey eksen dar bir aralıkta gösterildiği için artış olduğundan daha büyük görünmektedir.  
Grafik 2'de ise eksen aralıkları daha geniş ve bağlama uygun olduğu için değişim daha dengeli biçimde  
sunulmaktadır.

Bu nedenle Grafik 2 daha doğru bir gösterimdir.  
Çünkü veri görselleştirmelerinde eksenlerin başlangıç noktası ve aralıkları, izleyiciyi yanıltmayacak biçimde  
düzenlenmelidir.

Bu örnek, veri gösteriminde alan bilgisi ve etik sorumluluğun önemini göstermektedir.  
Veri bilimci, yalnızca grafiği üretmekle değil, grafiğin doğru anlaşılmasını sağlamakla da sorumludur.

## Verinin Analiz Edilmesi ve Yorumlanması

Analiz sürecinde yalnızca ortalamaya bakmak çoğu zaman yeterli değildir.  
Dağılım ölçüleri, standart sapma ve veri setinin homojenliği de dikkate alınmalıdır.

### Survivor Dayanıklılık Örneği

Survivor yarışması için ünlüler ve gönüllüler olmak üzere iki grubun fiziksel ve ruhsal dayanıklılığı ölçülmüştür.

| Grup | Ortalama | Standart Sapma |
|---|---:|---:|
| Gönüllüler | 70 | 3 |
| Ünlüler | 74 | 8 |

Yalnızca ortalamaya bakıldığında ünlüler grubunun daha avantajlı olduğu düşünülebilir.  
Ancak standart sapma dikkate alındığında tablo farklı yorumlanır.

Gönüllüler grubunun ortalaması daha düşük olmasına rağmen standart sapması `3`tür.  
Bu durum, grup üyelerinin dayanıklılık düzeylerinin birbirine yakın olduğunu ve grubun daha homojen bir yapı  
sergilediğini gösterir.

Ünlüler grubunun ortalaması `74` ile daha yüksek olsa da standart sapması `8`dir.  
Bu durum, grup içinde dayanıklılık düzeylerinin daha fazla değişkenlik gösterdiğini ve bazı üyelerin  
ortalamadan belirgin biçimde uzaklaşabileceğini gösterir.

Bu nedenle uzun süreli, zorlu ve takım uyumunun önemli olduğu bir yarışma ortamında gönüllüler grubu daha  
güvenilir bir tercih olabilir.  
Bu örnek, karar alma süreçlerinde ortalama ile birlikte dağılım ölçülerinin de değerlendirilmesi gerektiğini  
göstermektedir.

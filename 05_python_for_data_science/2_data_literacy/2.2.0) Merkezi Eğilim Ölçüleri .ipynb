# 2.2) Merkezi Eğilim Ölçüleri

Merkezi eğilim ölçüleri, bir veri kümesinin merkezini, tipik değerini veya genel eğilimini tanımlamak için  
kullanılan temel istatistiksel ölçülerdir.  
Bu ölçüler, veri setindeki değerlerin hangi noktada yoğunlaştığını anlamaya yardımcı olur.

En sık kullanılan merkezi eğilim ölçüleri şunlardır:

1. Aritmetik ortalama
2. Medyan (ortanca)
3. Mod (tepe değer)
4. Kartiller (çeyreklikler)

## Aritmetik Ortalama

Aritmetik ortalama, bir değişkene ait tüm gözlem değerlerinin toplamının gözlem sayısına bölünmesiyle elde edilir.  
Simetrik dağılıma sahip veri setlerinde, veri kümesini temsil etmek için oldukça kullanışlıdır.

Ancak veri setinde aykırı değerler bulunuyorsa aritmetik ortalama yanıltıcı olabilir.  
Çünkü uç değerler ortalamayı kendi yönlerine doğru çekerek veri setinin tipik değerini olduğundan farklı gösterebilir.

### Aritmetik Ortalama Hesabı

Veri seti:

`13, 10, 15, 12, 17, 13`

Hesaplama:

`x̄ = (13 + 10 + 15 + 12 + 17 + 13) / 6 = 13,33`

### Örnek Senaryo

Bir web sitesinde kullanıcıların geçirdiği toplam süre ve kullanıcı sayısı biliniyorsa, kullanıcı başına  
ortalama zaman şu şekilde hesaplanabilir:

`Ortalama süre = Kullanıcıların toplam geçirdiği süre / Kullanıcı sayısı`

Bu bilgi, web sitesindeki içeriklerin ne ölçüde tüketildiğini değerlendirmek ve içerik optimizasyonu yapmak için kullanılabilir.

## Medyan (Ortanca)

Medyan, veri setindeki değerler küçükten büyüğe sıralandığında ortada kalan değerdir.  
Eğer gözlem sayısı çok fazlaysa veya veri setinde aykırı değerler bulunuyorsa, medyan merkezi eğilimi  
aritmetik ortalamadan daha sağlıklı temsil edebilir.

Medyan; aralık (negatif ve pozitif), oran (sadece pozitif) ve sıralama ölçeğinde ölçülen değişkenler için kullanılabilir.  
Ancak nominal değişkenler için kullanılamaz; çünkü bu tür değişkenlerde değerler doğal bir sıralamaya sahip değildir.

### Tek Sayıda Gözlem İçin Medyan

Gözlem sayısı (n) tek ise medyan ortadaki terimdir:

`Medyan = ((n + 1) / 2). terim`

Örnek:

Veri seti: `13, 10, 15, 12, 17`

Sıralanmış veri seti: `10, 12, 13, 15, 17`

`(n + 1) / 2 = (5 + 1) / 2 = 3. terim`

Bu durumda:

`Medyan = 13`

### Çift Sayıda Gözlem İçin Medyan

Gözlem sayısı (n) çift ise medyan, ortadaki iki değerin aritmetik ortalamasıdır:

`Medyan = [(n / 2). terim + ((n / 2) + 1). terim] / 2`

Örnek:

Veri seti: `13, 10, 15, 12, 17, 13`

Sıralanmış veri seti: `10, 12, 13, 13, 15, 17`

Bu durumda:

`Medyan = (13 + 13) / 2 = 13`

## Medyan ve Ortalama Karşılaştırması

Aritmetik ortalama, simetrik dağılımlarda veri setini iyi temsil eder.  
Ancak aykırı değerlerin bulunduğu veri setlerinde ortalama yanıltıcı sonuçlar verebilir.

Örnek veri seti:

`13, 10, 15, 12, 17, 12, 19, 18, 11, 12, 190`

Bu veri setinde `190` değeri aykırı değerdir ve aritmetik ortalamayı yukarı doğru çekmektedir.

- Medyan: `13`
- Aritmetik ortalama: `28,5`

Bu durumda veri setinin tipik değerini ifade etmek için medyan daha uygun bir ölçüdür.  
Analiz sürecinde bu durum şu şekilde açıklanabilir:

> Veri seti simetrik dağılmadığı ve aykırı değer içerdiği için merkezi eğilim ölçüsü olarak medyan tercih edilmiştir.

Bu nedenle bir veri seti incelenirken “Veri setinin dağılımı simetrik mi?” ve “Aykırı değer var mı?” soruları mutlaka sorulmalıdır.

## Mod (Tepe Değer)

Mod, bir veri setinde en sık tekrar eden değerdir. Bir veri setinin en yaygın gözlem değerini göstermek için kullanılır.

Örnek:

Veri seti: `13, 10, 13, 12, 17, 13, 14`

Bu veri setinde en çok tekrar eden değer `13` olduğu için:

`Mod = 13`

## Ortalama, Medyan ve Mod Arasındaki İlişki

Aritmetik ortalama, medyan ve mod arasındaki ilişki, dağılımın simetrik veya çarpık olup olmadığı hakkında bilgi verir.

- `AO < OR < TD`: Sola çarpık dağılım.
- `AO = OR = TD`: Simetrik dağılım.
- `TD < OR < AO`: Sağa çarpık dağılım.

Burada `AO` aritmetik ortalamayı, `OR` ortancayı, `TD` ise tepe değeri ifade eder.

## Kartiller (Çeyreklikler)

Kartiller, küçükten büyüğe sıralanmış bir veri setini dört eşit parçaya ayıran değerlerdir.  
Hem merkezi eğilim hem de dağılım yapısı hakkında bilgi verir.

Kartil gösterimi şu şekildedir:

- `Q1`: Birinci çeyrek, veri setinin %25'lik noktasını ifade eder.
- `Q2`: İkinci çeyrek, medyanı yani %50'lik noktayı ifade eder.
- `Q3`: Üçüncü çeyrek, veri setinin %75'lik noktasını ifade eder.

Formüller:

`Q1 = 1/4 (n + 1). terim`

`Q3 = 3/4 (n + 1). terim`

Örnek:

Veri seti:

`8, 10, 15, 12, 17, 20, 14`

Sıralanmış veri seti:

`8, 10, 12, 14, 15, 17, 20`

Bu durumda:

- `Q1 = 10`
- `Q2 = 14`
- `Q3 = 17`

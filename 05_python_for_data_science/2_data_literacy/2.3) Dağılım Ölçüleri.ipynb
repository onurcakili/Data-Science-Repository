# 2.3) Dağılım Ölçüleri

Dağılım ölçüleri, bir veri setindeki değerlerin birbirinden ve merkezi eğilim ölçülerinden ne kadar  
uzaklaştığını gösteren istatistiksel ölçülerdir.  
Merkezi eğilim ölçüleri veri setinin tipik değerini açıklarken, dağılım ölçüleri değerlerin bu tipik değer  
etrafında nasıl yayıldığını gösterir.

Bu bölümde değişim aralığı, standart sapma, varyans, çarpıklık ve basıklık kavramları ele alınmaktadır.

## Değişim Aralığı

Değişim aralığı, bir serideki maksimum değer ile minimum değer arasındaki farktır.

`Değişim Aralığı = Maksimum Değer - Minimum Değer`

Örnek veri seti:

`8, 10, 15, 12, 17, 20, 14`

Hesaplama:

`Değişim Aralığı = 20 - 8 = 12`

Değişim aralığı hesaplanırken yalnızca iki uç değer dikkate alınır.  
Bu nedenle veri setindeki diğer değerlerin dağılımı hakkında sınırlı bilgi verir.  
Özellikle aykırı değerlerin bulunduğu veri setlerinde tek başına kullanılması yanıltıcı olabilir.

### Örnek Senaryo

Bir iş pozisyonu için farklı şirketlerden maaş teklifleri alındığını düşünelim.  
Şirketlere “Maaşlardaki değişim aralığınız nedir?” sorusu yöneltildiğinde bazı şirketlerde bu aralık 1000 TL,  
bazılarında ise 10000 TL olabilir.

Maaş değişim aralığı düşük olan bir şirkette ücretlerin daha dengeli dağıldığı söylenebilir.  
Ancak daha sağlıklı bir değerlendirme için standart sapma ve varyans gibi ek dağılım ölçülerinin de  
incelenmesi gerekir.

## Standart Sapma

Standart sapma, gözlem değerlerinin aritmetik ortalamadan ortalama olarak ne kadar uzaklaştığını gösteren  
temel bir dağılım ölçüsüdür.

Formül:

`s = √[(1/n) Σ(xi - x̄)²]`

Standart sapma düşükse gözlemler ortalamaya yakın, standart sapma yüksekse gözlemler ortalamadan daha uzak dağılmıştır.

### Standart Sapma Hesabı

| Kazanç (xi) | xi - x̄ | (xi - x̄)² |
|---:|---:|---:|
| 12 | -12 | 144 |
| 15 | -9 | 81 |
| 20 | -4 | 16 |
| 30 | 6 | 36 |
| 45 | 21 | 441 |
| 22 | -2 | 4 |
| Toplam | 0 | 722 |

`x̄ = (12 + 15 + 20 + 30 + 45 + 22) / 6 = 24`

`s = √[(1/6) * 722]`

`s = 10,97`

Bu sonuç, gözlem değerlerinin ortalamadan yaklaşık `10,97` birim uzaklaştığını göstermektedir.

## Varyans

Varyans, ortalamadan sapmaların karelerinin ortalamasıdır. Standart sapmanın karesi olarak da ifade edilebilir.

Formül:

`s² = (1/n) Σ(xi - x̄)²`

Varyans özellikle birden fazla veri setinin yayılımını karşılaştırmak için kullanılabilir.  
Ancak kareli bir ölçü olduğu için yorumlanması standart sapmaya göre daha güç olabilir.

### Varyans Hesabı

| Kazanç (xi) | xi - x̄ | (xi - x̄)² |
|---:|---:|---:|
| 12 | -12 | 144 |
| 15 | -9 | 81 |
| 20 | -4 | 16 |
| 30 | 6 | 36 |
| 45 | 21 | 441 |
| 22 | -2 | 4 |
| Toplam | 0 | 722 |

`x̄ = 24`

`s² = (1/6) * 722`

`s² = 120,34`

Bu durumda standart sapma `10,97`, varyans ise `120,34` olarak hesaplanır.

## Çarpıklık

Çarpıklık, bir dağılımın simetrik olup olmadığını gösteren ölçüdür.  
Dağılımın sağa veya sola doğru uzaması, veri setinin merkezi eğilim ölçülerinin yorumunu etkileyebilir.

Pearson çarpıklık katsayısı şu şekilde hesaplanır:

`PÇK = 3(x̄ - medyan) / standart sapma`

Yorum:

- `PÇK < 0`: Negatif veya sola çarpık dağılım.
- `PÇK > 0`: Pozitif veya sağa çarpık dağılım.
- `PÇK = 0`: Simetrik dağılım.

### Pearson Çarpıklık Katsayısı Hesabı

| Kazanç (xi) | xi - x̄ | (xi - x̄)² |
|---:|---:|---:|
| 12 | -12 | 144 |
| 15 | -9 | 81 |
| 20 | -4 | 16 |
| 30 | 6 | 36 |
| 45 | 21 | 441 |
| 22 | -2 | 4 |
| Toplam | 0 | 722 |

`x̄ = 24`

`s = 10,97`

Sıralı değerler:

`12, 15, 20, 22, 30, 45`

`Medyan = (20 + 22) / 2 = 21`

`PÇK = 3(24 - 21) / 10,97`

`PÇK = 0,82`

Katsayının pozitif olması dağılımın sağa çarpık olduğunu gösterir.  
Değerin `1`e yakın olması, çarpıklığın dikkate alınması gerektiğini gösterir.

## Basıklık

Basıklık, dağılımın tepe noktasının sivri veya basık olma derecesini ifade eder.  
Bu ölçü, verilerin ortalama çevresinde ne kadar yoğunlaştığı hakkında bilgi verir.

Genel yorum şu şekildedir:

- `BK = 3`: Dağılım standart normal dağılıma yakındır.
- `BK > 3`: Dağılım daha sivridir; değerler merkeze daha yoğun biçimde toplanmıştır.
- `BK < 3`: Dağılım daha basıktır; değerler daha geniş bir alana yayılmıştır.

Basıklık katsayısı:

`Basıklık Katsayısı = m4 / s⁴`

Burada:

`m4 = Σ(xi - x̄)⁴ / n`

`s⁴ = standart sapmanın dördüncü kuvveti`

### Basıklık Katsayısı Hesabı

| Kazanç (xi) | xi - x̄ | (xi - x̄)⁴ |
|---:|---:|---:|
| 12 | -12 | 20736 |
| 15 | -9 | 6561 |
| 20 | -4 | 256 |
| 30 | 6 | 1296 |
| 45 | 21 | 194481 |
| 22 | -2 | 16 |
| Toplam | 0 | 223346 |

`m4 = 223346 / 6 = 37224,33`

`s⁴ = (10,97)⁴ = 14481,93`

`m4 / s⁴ = 37224,33 / 14481,93 = 2,57`

Bu durumda basıklık değeri `2,57`dir.  
Değer `3`ten küçük olduğu için dağılım basık olarak yorumlanır.  
Bu sonuç, değerlerin standart normal dağılıma göre daha geniş bir alana yayıldığını gösterir.

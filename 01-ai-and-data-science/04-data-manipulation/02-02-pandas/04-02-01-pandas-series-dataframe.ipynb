# Pandas Series ve DataFrame Temelleri

Pandas, veri bilimi çalışmalarında tablo verileriyle çalışmak için kullanılan en temel Python kütüphanelerinden biridir.

Bu notebook'ta `Series`, `DataFrame`, sütun seçme, temel inceleme ve özetleme konularını öğreneceğiz.


## Konu Dokümantasyonu

Pandas'ın iki ana veri yapısı vardır:

* `Series`: Tek boyutlu, indeksli veri yapısı
* `DataFrame`: Satır ve sütunlardan oluşan iki boyutlu tablo

Veri biliminde çoğu veri seti DataFrame olarak temsil edilir. CSV dosyaları, Excel tabloları, SQL sorguları ve API verileri genellikle DataFrame'e dönüştürülerek analiz edilir.

Pandas, NumPy üzerine kuruludur; bu yüzden sayısal işlemlerde NumPy mantığı arka planda çalışır.


## Kolay Seviye

Series tek bir sütun gibi düşünülebilir. DataFrame ise birden fazla sütundan oluşan tablo gibidir.


```python
import pandas as pd

satislar = pd.Series([1200, 950, 1800, 2100], name="satis")

print(satislar)
print("Ortalama:", satislar.mean())
```

```python
import pandas as pd

df = pd.DataFrame({
    "urun": ["Laptop", "Telefon", "Tablet"],
    "kategori": ["Elektronik", "Elektronik", "Elektronik"],
    "fiyat": [30000, 20000, 12000],
    "stok": [5, 8, 12],
})

print(df)
```

## Orta Seviye

DataFrame'i tanımak için `head()`, `shape`, `dtypes`, `info()` ve `describe()` gibi araçlar kullanılır.

Analize başlamadan önce veri setinin kaç satır/sütun olduğu, sütun tipleri ve temel istatistikleri mutlaka incelenmelidir.


```python
import pandas as pd

df = pd.DataFrame({
    "urun": ["Laptop", "Telefon", "Tablet", "Kulaklık"],
    "kategori": ["Elektronik", "Elektronik", "Elektronik", "Aksesuar"],
    "fiyat": [30000, 20000, 12000, 2500],
    "stok": [5, 8, 12, 30],
})

print(df.head())          # İlk 5 satırı gösterir
print(df.shape)           # Satır ve sütun sayısı
print(df.dtypes)          # Sütun veri türleri
print(df.describe())      # Sayısal sütunların istatistiksel özeti
```

```python
import pandas as pd

df = pd.DataFrame({
    "urun": ["Laptop", "Telefon", "Tablet", "Kulaklık"],
    "fiyat": [30000, 20000, 12000, 2500],
    "stok": [5, 8, 12, 30],
})

df["stok_degeri"] = df["fiyat"] * df["stok"]
# Yeni sütun, mevcut sütunlardan hesaplanarak oluşturulur

print(df)
```

## İleri Seviye

Pandas'ta sütunlar vektör gibi davranır. Bu sayede her satır için tek tek döngü yazmadan hesaplama yapılabilir.

`assign()` yöntemi, yeni sütun üretirken zincirleme ve okunabilir kullanım sağlar.


```python
import pandas as pd

df = pd.DataFrame({
    "urun": ["Laptop", "Telefon", "Tablet", "Kulaklık"],
    "fiyat": [30000, 20000, 12000, 2500],
    "stok": [5, 8, 12, 30],
})

sonuc = df.assign(
    stok_degeri=lambda veri: veri["fiyat"] * veri["stok"],
    kdv_dahil_fiyat=lambda veri: veri["fiyat"] * 1.20,
)

print(sonuc)
```

## Mini Veri Bilimi Uygulaması

Ürün tablosundan stok değeri hesaplayıp toplam stok değerini bulalım.


```python
import pandas as pd

df = pd.DataFrame({
    "urun": ["Laptop", "Telefon", "Tablet", "Kulaklık"],
    "kategori": ["Elektronik", "Elektronik", "Elektronik", "Aksesuar"],
    "fiyat": [30000, 20000, 12000, 2500],
    "stok": [5, 8, 12, 30],
})

df["stok_degeri"] = df["fiyat"] * df["stok"]

toplam_stok_degeri = df["stok_degeri"].sum()
en_degerli_urun = df.loc[df["stok_degeri"].idxmax(), "urun"]

print("Toplam stok değeri:", toplam_stok_degeri)
print("En değerli stok ürünü:", en_degerli_urun)
```

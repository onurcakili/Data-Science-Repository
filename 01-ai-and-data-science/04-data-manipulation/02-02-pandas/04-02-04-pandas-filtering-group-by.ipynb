# Pandas Filtreleme, Sıralama ve GroupBy

Veri analizinde çoğu soru belirli bir koşula göre veriyi seçmek ve gruplar bazında özetlemekle çözülür. Bu notebook'ta filtreleme, sıralama ve `groupby` işlemlerini öğreneceğiz.


## Konu Dokümantasyonu

Filtreleme, belirli koşulu sağlayan satırları seçmektir. GroupBy ise veriyi bir veya daha fazla sütuna göre gruplandırıp her grup için hesaplama yapmaktır.

Örnek sorular:

* Hangi kategorinin toplam satışı daha yüksek?
* Şehir bazında ortalama sipariş tutarı nedir?
* En çok satış yapılan ürün hangisi?
* Müşteri segmentlerine göre gelir nasıl değişiyor?

Bu tip sorular veri bilimi ve iş analitiğinde çok sık karşımıza çıkar.


## Kolay Seviye

Filtreleme için boolean koşullar kullanılır.


```python
import pandas as pd

df = pd.DataFrame({
    "urun": ["Laptop", "Telefon", "Tablet", "Kulaklık", "Masa"],
    "kategori": ["Elektronik", "Elektronik", "Elektronik", "Aksesuar", "Mobilya"],
    "satis": [30000, 20000, 12000, 2500, 7500],
})

elektronik = df[df["kategori"] == "Elektronik"]
pahali = df[df["satis"] >= 10000]

print(elektronik)
print(pahali)
```

```python
import pandas as pd

df = pd.DataFrame({
    "urun": ["Laptop", "Telefon", "Tablet", "Kulaklık", "Masa"],
    "satis": [30000, 20000, 12000, 2500, 7500],
})

sirali = df.sort_values("satis", ascending=False)
# Büyükten küçüğe sıralama

print(sirali)
```

## Orta Seviye

`groupby()` ile gruplama yapılır. Sonrasında `sum()`, `mean()`, `count()` gibi özet fonksiyonları uygulanır.


```python
import pandas as pd

df = pd.DataFrame({
    "kategori": ["Elektronik", "Elektronik", "Aksesuar", "Mobilya", "Mobilya"],
    "satis": [30000, 20000, 2500, 7500, 9000],
})

kategori_toplam = df.groupby("kategori")["satis"].sum()
print(kategori_toplam)
```

```python
import pandas as pd

df = pd.DataFrame({
    "sehir": ["İstanbul", "Ankara", "İstanbul", "İzmir", "Ankara"],
    "kategori": ["Elektronik", "Elektronik", "Mobilya", "Elektronik", "Mobilya"],
    "satis": [30000, 20000, 7500, 12000, 9000],
})

ozet = df.groupby(["sehir", "kategori"])["satis"].sum()
# Çok seviyeli groupby: şehir ve kategori birlikte kullanılır

print(ozet)
```

## İleri Seviye

`agg()` ile birden fazla özet metriği aynı anda hesaplanabilir. `transform()` ise grup bazlı sonucu orijinal satır sayısını koruyarak üretir.


```python
import pandas as pd

df = pd.DataFrame({
    "kategori": ["Elektronik", "Elektronik", "Aksesuar", "Mobilya", "Mobilya"],
    "satis": [30000, 20000, 2500, 7500, 9000],
})

ozet = df.groupby("kategori").agg(
    toplam_satis=("satis", "sum"),
    ortalama_satis=("satis", "mean"),
    islem_sayisi=("satis", "count"),
)

print(ozet)
```

```python
import pandas as pd

df = pd.DataFrame({
    "kategori": ["Elektronik", "Elektronik", "Aksesuar", "Mobilya", "Mobilya"],
    "satis": [30000, 20000, 2500, 7500, 9000],
})

df["kategori_toplami"] = df.groupby("kategori")["satis"].transform("sum")
df["kategori_ici_pay"] = df["satis"] / df["kategori_toplami"]
# Her satırın kendi kategorisi içindeki payı hesaplanır

print(df)
```

## Mini Veri Bilimi Uygulaması

Şehir ve kategori bazında toplam satışları hesaplayıp en yüksek performanslı kırılımları bulalım.


```python
import pandas as pd

df = pd.DataFrame({
    "sehir": ["İstanbul", "Ankara", "İstanbul", "İzmir", "Ankara", "İstanbul"],
    "kategori": ["Elektronik", "Elektronik", "Mobilya", "Elektronik", "Mobilya", "Aksesuar"],
    "satis": [30000, 20000, 7500, 12000, 9000, 2500],
})

ozet = (
    df.groupby(["sehir", "kategori"])
    .agg(toplam_satis=("satis", "sum"))
    .sort_values("toplam_satis", ascending=False)
)

print(ozet)
```

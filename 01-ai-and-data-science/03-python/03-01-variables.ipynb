## Python Değişkenler (Variables)

Değişken, bellekte tutulan her değere verilen isimdir.  
Program boyunca bir sayıyı, metni, listeyi, sonucu veya ara hesabı tekrar kullanmak istediğimizde değişken tanımlarız.  

Değişkenleri sadece teknik bir saklama alanı gibi düşünmemek gerekir.  
İyi seçilmiş bir değişken adı, kodun niyetini anlatır.  

Örneğin `x = 125000` çalışır ama `toplam_satis = 125000` ne saklandığını doğrudan gösterir.  

Python'da değişken tanımlarken veri türünü ayrıca yazmayız. Python değere bakarak türü kendisi belirler.  
Bu esneklik hızlı kod yazmayı sağlar; fakat yanlış türde veriyle işlem yaparsak beklenmeyen sonuçlar oluşabilir.

Veri bilimi tarafında değişkenler çoğunlukla şu amaçlarla kullanılır:

* Ham veriyi temsil etmek
* Ara hesapları saklamak
* Temizlenmiş veriyi ayırmak
* Model veya analiz sonuçlarını tutmak
* Okunabilir rapor değişkenleri oluşturmak

Bu nedenle değişken isimleri kısa ama anlamlı olmalıdır.  

Genel tavsiye `snake_case` kullanmaktır: `ortalama_satis`, `aktif_musteri_sayisi`, `temiz_veri` gibi.


## Kolay Seviye

Değişken oluşturma üç parçadan oluşur:

* Değişken adı
* Atama operatörü olan `=`
* Saklanacak değer

Örnek:

```python
yas = 26
```

Burada `yas` değişken adıdır, `=` atama yapar, `26` ise değişkende saklanan değerdir.  
`print()` fonksiyonu değişkenin içindeki değeri ekrana yazdırır.  
`type()` fonksiyonu ise değişkenin veri türünü gösterir.


````python
isim = "Onur"          # str: metin türünde değer saklar
yas = 26               # int: tam sayı türünde değer saklar
ortalama = 84.5        # float: ondalıklı sayı türünde değer saklar
aktif_kullanıcı = True        # bool: True veya False değerlerinden birini saklar

print(isim)
print(yas)
print(ortalama)
print(aktif_kullanıcı)
````

````python
puan = 70              # İlk değer atandı
print("İlk puan:", puan)

puan = 85              # Aynı değişkene yeni değer atanırsa eski değer değişir
print("Güncel puan:", puan)

print(type(puan))      # type() değişkenin veri türünü gösterir
````

## Orta Seviye

Gerçek programlarda değişkenler genellikle tek başına durmaz; birbirleriyle işlem yaparak yeni sonuçlar üretir.   
Örneğin ürün fiyatı, adet ve vergi oranı ayrı değişkenlerde saklanır; toplam tutar bu değişkenlerden hesaplanır.

Bu yaklaşımın faydası şudur: Bir değer değiştiğinde sadece ilgili değişken güncellenir, hesaplama mantığı aynı kalır.  
Ayrıca f-string kullanarak sonucu kullanıcıya daha anlaşılır biçimde gösterebiliriz.


````python
birim_fiyat = 250          # Bir ürünün fiyatı
adet = 4                   # Kaç adet ürün alındı?
kdv_orani = 0.20           # Sabit gibi kullanılan oranlar açıklayıcı isimle yazılır

ara_toplam = birim_fiyat * adet
kdv_tutari = ara_toplam * kdv_orani
genel_toplam = ara_toplam + kdv_tutari

print(f"Ara toplam: {ara_toplam} TL")
print(f"KDV tutarı: {kdv_tutari} TL")
print(f"Genel toplam: {genel_toplam} TL")
````

````python
# Python'da birden fazla değişkene aynı anda değer atamak mümkündür. 
# Örneğin, bir kişinin adı, soyadı ve yaşı gibi bilgileri tek satırda atayabiliriz:

ad, soyad, yas = "Zeynep", "Yılmaz", 22  # Çoklu atama: soldaki isimler sağdaki değerleri sırayla alır

kullanici_ozeti = f"{ad} {soyad} - {yas} yaşında"

print(kullanici_ozeti)
````

## İleri Seviye - Referanslar

Basit sayılar ve metinlerde genellikle doğrudan değerle çalışıyormuş gibi düşünmek yeterlidir.  
Fakat liste (list) ve sözlük (dictionary) gibi değiştirilebilir yapılarda değişkenler aynı nesneyi işaret edebilir.  
Bu durumda bir değişken üzerinden yapılan değişiklik diğer değişkende de görünür.

Bu konu veri biliminde önemlidi çünkü ham veri, temiz veri ve dönüştürülmüş veri aynı nesneyi paylaşıyorsa, farkında olmadan orijinal veri de değişebilir.  
Bu yüzden gerektiğinde `.copy()` ile kopya oluşturmak gerekir.

````python
orijinal_notlar = [70, 80, 90]
sahte_kopya = orijinal_notlar      # Bu satır listeyi kopyalamaz, aynı listeye ikinci isim verir

sahte_kopya.append(100)            # Aynı liste değiştiği için orijinal_notlar da değişmiş görünür

print("Orijinal:", orijinal_notlar)
print("Sahte kopya:", sahte_kopya)
````

````python
orijinal_notlar = [70, 80, 90]
guvenli_kopya = orijinal_notlar.copy()    # copy() yeni bir liste oluşturur

guvenli_kopya.append(100)                 # Sadece kopya liste değişir

print("Orijinal:", orijinal_notlar)
print("Güvenli kopya:", guvenli_kopya)
````

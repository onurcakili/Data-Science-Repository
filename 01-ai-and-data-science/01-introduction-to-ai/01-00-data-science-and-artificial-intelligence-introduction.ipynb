# Veri Bilimi ve Yapay Zekâya Giriş

## Giriş

Dijital dönüşümün hız kazandığı günümüzde **veri**, bireylerin, işletmelerin ve devletlerin karar alma süreçlerinde stratejik bir kaynak hâline gelmiştir.

İnternet kullanımının yaygınlaşması, mobil cihazların gelişmesi, sosyal medya platformlarının kullanımının artması, sensör teknolojileri ve Nesnelerin İnterneti (IoT - Internet of Things) gibi teknolojilerin gelişmesi sonucunda her gün milyarlarca veri üretilmektedir.

Üretilen bu veriler;

* Müşteri davranışlarının analiz edilmesi,
* Hastalıkların teşhis edilmesi,
* Finansal tahminlerin yapılması,
* Otonom araç sistemlerinin geliştirilmesi

gibi çok geniş bir kullanım alanına sahiptir.

Ancak **ham veriler tek başlarına anlamlı bir sonuç ortaya koymaz.**

Verilerin işlenmesi, analiz edilmesi ve karar verme süreçlerinde kullanılabilecek **bilgiye** dönüştürülmesi gerekir.

Bu noktada **Veri Bilimi (Data Science)** ve **Yapay Zekâ (Artificial Intelligence - AI)** kavramları ön plana çıkmaktadır.

> **Veri Bilimi**, yapılandırılmış ve yapılandırılmamış verilerden anlamlı bilgi, öngörü ve karar destek çıktıları üretmek amacıyla istatistik, matematik, bilgisayar bilimleri ve alan bilgisini bir araya getiren disiplinler arası bir çalışma alanıdır.

> **Yapay Zekâ**, insan zekâsını gerektiren görevleri gerçekleştirebilen sistemlerin geliştirilmesine odaklanan bilgisayar bilimi alanıdır.

Veri bilimi ve yapay zekânın anlaşılabilmesi için öncelikle bu alanların temelini oluşturan **veri kavramının** ve veriyle ilişkili diğer kavramların açıklanması gerekmektedir.

---

# 1. Veri Kavramı

## 1.1. Veri Nedir?

**Veri (Data)**; gözlem, ölçüm, deney, kayıt, anket ve benzeri yöntemlerle elde edilen ham ve işlenmemiş gerçeklerdir.

Veri, tek başına ele alındığında çoğu zaman yeterli bir anlam ifade etmez. Anlam kazanabilmesi için belirli bir bağlam içerisinde değerlendirilmesi ve işlenmesi gerekir.

Günlük yaşamda karşılaşılan birçok unsur veri olarak değerlendirilebilir:

* Yaş bilgisi
* Hava sıcaklığı
* Ürün fiyatları
* Müşteri isimleri
* Satış adetleri
* Sınav puanları

Örneğin yalnızca **"38"** değeri tek başına değerlendirildiğinde anlamı belirsizdir.

Bu değer;

* Bir kişinin yaşı,
* Bir ürünün fiyatı,
* Bir sınav puanı,
* Hava sıcaklığı

olabilir.

Dolayısıyla verinin anlam kazanabilmesi için **bağlama** ihtiyaç vardır.

Bu noktada veri ile birlikte **enformasyon, bilgi ve karar** kavramlarının da birbirinden ayrılması önem taşımaktadır.

---

## 1.2. Veri, Enformasyon, Bilgi ve Karar

Veri biliminin temel mantığını anlayabilmek için **veri, enformasyon, bilgi ve karar** arasındaki ilişkiyi anlamak gerekir.

Bkz. Bilgi Piramidi

### Veri (Data)

Ham, işlenmemiş ve tek başına yeterli anlam ifade etmeyen gerçeklerdir.

### Enformasyon (Information)

Verilerin düzenlenmesi, kategorize edilmesi ve belirli bir bağlam içerisinde anlam kazanması sonucunda ortaya çıkan çıktıdır.

### Bilgi (Knowledge)

Enformasyonun analiz edilmesi ve yorumlanması sonucunda elde edilen anlamlı içgörü ve çıkarımlardır.

### Karar (Decision)

Elde edilen bilgiye dayanarak gerçekleştirilen eylem veya yapılan seçimdir.

### Örnek

Bu ilişki aşağıdaki örnek üzerinden açıklanabilir:

| Aşama           | Açıklama                                      |
| --------------- | --------------------------------------------- |
| **Veri**        | 38                                            |
| **Enformasyon** | Hava sıcaklığı 38°C olarak ölçülmüştür.       |
| **Bilgi**       | Sıcaklık mevsim normallerinin üzerindedir.    |
| **Karar**       | Dışarı çıkarken gerekli önlemler alınmalıdır. |

Bu dönüşüm süreci, **ham verilerin işlenerek karar destek mekanizmalarında kullanılabilecek anlamlı çıktılara dönüştürülmesini** sağlar.

---

# 2. Veri Türleri

Veriler farklı özelliklerine göre çeşitli şekillerde sınıflandırılabilir.

Temel sınıflandırmalardan biri, verilerin **niceliksel** ve **niteliksel** olarak ayrılmasıdır.

---

## 2.1. Niceliksel Veriler (Quantitative Data)

Niceliksel veriler, **sayısal olarak ifade edilebilen ve miktar belirten verilerdir.**

Örnekler:

* Yaş
* Maaş
* Boy
* Kilo
* Satış miktarı

Niceliksel veriler kendi içerisinde:

1. **Kesikli Veriler**
2. **Sürekli Veriler**

olmak üzere iki gruba ayrılabilir.

---

## 2.2. Kesikli Veriler

**Kesikli veriler**, belirli ve sayılabilir değerler alan verilerdir.

Genellikle tam sayı değerleriyle ifade edilirler.

Örnekler:

* Çalışan sayısı
* Sipariş sayısı

Örneğin bir işletmede çalışan sayısı **125** olabilir.

Ancak **125,5 çalışan** şeklinde bir değer söz konusu değildir.

---

## 2.3. Sürekli Veriler

**Sürekli veriler**, belirli bir aralık içerisinde teorik olarak sonsuz sayıda değer alabilen verilerdir.

Örnekler:

* Boy
* Kilo
* Sıcaklık

Örneğin bir kişinin boyu;

* 175 cm,
* 175,2 cm,
* 175,25 cm

olarak ölçülebilir.

---

## 2.4. Niteliksel Veriler (Qualitative Data)

**Niteliksel veriler**, sayısal bir ölçümden ziyade bir özelliği veya kategoriyi ifade eden verilerdir.

Örnekler:

* Cinsiyet
* Şehir
* Meslek
* Eğitim durumu

Örneğin bir müşterinin bulunduğu şehir **İstanbul**, mesleği ise **Mühendis** olabilir.

Bu değerler sayısal bir büyüklükten ziyade **kategorik özellikleri** ifade eder.

---

# 3. Veri Yapıları

Veriler yalnızca içeriklerine göre değil, aynı zamanda **nasıl yapılandırıldıklarına** göre de sınıflandırılabilir.

Bu açıdan veriler genel olarak üç grupta incelenmektedir:

1. **Yapılandırılmış Veri**
2. **Yarı Yapılandırılmış Veri**
3. **Yapılandırılmamış Veri**

---

## 3.1. Yapılandırılmış Veri (Structured Data)

**Yapılandırılmış veri**, belirli bir veri modeli ve format içerisinde organize edilmiş verilerdir.

Genellikle satır ve sütunlardan oluşan tablo yapısında bulunurlar.

Örnekler:

* Excel tabloları
* SQL veritabanları
* ERP kayıtları
* CRM sistemleri

### Örnek

| Müşteri ID | İsim  | Şehir    |
| ---------: | ----- | -------- |
|          1 | Ahmet | İstanbul |
|          2 | Ayşe  | Ankara   |

Yapılandırılmış veriler belirli bir yapıya sahip oldukları için **kolayca sorgulanabilir, işlenebilir ve analiz edilebilir.**

---

## 3.2. Yarı Yapılandırılmış Veri (Semi-Structured Data)

**Yarı yapılandırılmış veriler**, belirli kurallara ve organizasyon biçimine sahip olmakla birlikte tamamen tablo yapısında olmayan verilerdir.

Örnekler:

* JSON
* XML
* Log dosyaları

Bu veriler belirli etiketler, anahtarlar veya yapısal kurallar içerebilir.

Ancak klasik ilişkisel veritabanlarındaki gibi sabit satır ve sütun yapısına sahip olmak zorunda değildir.

---

## 3.3. Yapılandırılmamış Veri (Unstructured Data)

**Yapılandırılmamış veri**, önceden tanımlanmış belirli bir veri modeline sahip olmayan verilerdir.

Örnekler:

* Görseller
* Videolar
* Ses kayıtları
* E-postalar
* Sosyal medya içerikleri

Günümüzde dijital ortamda üretilen verilerin önemli bir bölümü yapılandırılmamış veri niteliğindedir.

Bu nedenle bu tür verilerin işlenmesi ve analiz edilmesi, **veri bilimi ve yapay zekâ çalışmalarında önemli bir yere sahiptir.**


# 5. Veri Bilimi

## 5.1. Veri Bilimi Nedir?

**Veri Bilimi (Data Science)**; yapılandırılmış ve yapılandırılmamış verilerden anlamlı bilgi, öngörü ve karar destek çıktıları üretmek amacıyla;

* İstatistik
* Matematik
* Bilgisayar bilimleri
* Alan bilgisi

gibi farklı disiplinleri bir araya getiren **disiplinler arası bir çalışma alanıdır.**

Veri biliminin temel amaçları şunlardır:

* Veriyi anlamak
* Veriden bilgi üretmek
* Geleceğe yönelik tahminlerde bulunmak
* Karar verme süreçlerini desteklemek

Bu amaçlara ulaşabilmek için veri biliminde veri toplama, veri temizleme, analiz, modelleme ve değerlendirme gibi çeşitli süreçler birlikte yürütülür.

---

# 7. Veri Analizi ve Veri Bilimi Arasındaki Fark

Veri analizi ve veri bilimi birbirleriyle yakından ilişkili olmakla birlikte aynı kavramlar değildir.

## Veri Analizi

**Veri analizi**, çoğunlukla mevcut verileri inceleyerek geçmişte gerçekleşen olayları ve mevcut durumu anlamaya odaklanır.

Örnekler:

* Geçen ay kaç ürün satıldı?
* Hangi mağaza daha yüksek performans gösterdi?

## Veri Bilimi

**Veri bilimi**, veri analizi süreçlerinin yanı sıra istatistiksel ve makine öğrenmesi yöntemlerinden yararlanarak geleceğe yönelik tahminler ve öngörüler üretmeyi de kapsayabilir.

Örnekler:

* Önümüzdeki ay ne kadar satış yapılacak?
* Hangi müşteriler alışveriş yapmayı bırakacak?

### Temel Fark

| Veri Analizi                                 | Veri Bilimi                                                         |
| -------------------------------------------- | ------------------------------------------------------------------- |
| Geçmişi ve mevcut durumu anlamaya odaklanır. | Geleceğe yönelik tahminler ve öngörüler de üretebilir.              |
| Mevcut verileri inceler.                     | Analiz, istatistik ve modelleme yöntemlerini birlikte kullanabilir. |
| "Ne oldu?" sorusuna odaklanabilir.           | "Ne olacak?" gibi sorulara da cevap arayabilir.                     |

Bu nedenle **veri analizi, veri biliminin içerisinde yer alabilen önemli çalışma alanlarından biridir; ancak veri bilimi yalnızca veri analizinden ibaret değildir.**

---

# 8. Yapay Zekâ

## 8.1. Yapay Zekâ Nedir?

**Yapay Zekâ (Artificial Intelligence - AI)**; çevresini algılayabilen, verilerden veya deneyimlerden öğrenebilen, akıl yürütebilen ve belirli hedeflere ulaşmak amacıyla karar verebilen sistemlerin geliştirilmesini konu alan bilgisayar bilimi alanıdır.

Yapay zekâ sistemlerinin temel amacı, insan zekâsını gerektiren görevlerin bilgisayar sistemleri tarafından gerçekleştirilebilmesini sağlamaktır.

Bu kapsamda yapay zekâ;

* Öğrenme
* Problem çözme
* Karar verme
* Algılama

gibi farklı yetenekleri içerebilir.

---

# 9. Makine Öğrenmesi

**Makine Öğrenmesi (Machine Learning - ML)**, yapay zekânın önemli alt alanlarından biridir.

Makine öğrenmesinde sistemler, her bir durum için açıkça programlanmak yerine **geçmiş verilerden örüntüler ve ilişkiler öğrenerek** belirli görevleri gerçekleştirebilir.

Makine öğrenmesinin kullanım alanlarına örnek olarak şunlar verilebilir:

* Ev fiyat tahmini
* Kredi risk analizi
* Hastalık teşhisi
* Sahtekârlık tespiti

Makine öğrenmesi temel olarak öğrenme biçimine göre farklı yaklaşımlarla ele alınabilir.

---

## 9.1. Gözetimli Öğrenme (Supervised Learning)

**Gözetimli öğrenme**, etiketlenmiş veriler kullanılarak gerçekleştirilen öğrenme yaklaşımıdır.

Model, girdiler ile bilinen sonuçlar arasındaki ilişkiyi öğrenmeye çalışır.

Örnekler:

* Ev fiyat tahmini
* Hastalık teşhisi

---

## 9.2. Gözetimsiz Öğrenme (Unsupervised Learning)

**Gözetimsiz öğrenme**, önceden belirlenmiş etiketlerin bulunmadığı veriler üzerinde çalışır.

Amaç, veriler içerisindeki doğal örüntüleri veya grupları ortaya çıkarmaktır.

Örnekler:

* Müşteri segmentasyonu
* Ürün gruplandırması

---

## 9.3. Pekiştirmeli Öğrenme (Reinforcement Learning)

**Pekiştirmeli öğrenme**, bir sistemin çevresiyle etkileşime girerek deneme-yanılma yoluyla öğrenmesini ifade eder.

Sistem, gerçekleştirdiği eylemlerin sonuçlarına göre ödül veya ceza niteliğinde geri bildirimler alır.

Örnekler:

* Oyun oynayan yapay zekâ sistemleri
* Robot sistemleri

---

# 10. Üretken Yapay Zekâ (Generative AI)

**Üretken Yapay Zekâ (Generative AI)**; mevcut verilerden öğrenerek metin, görsel, ses, video, yazılım kodu veya benzeri yeni içerikler oluşturabilen yapay zekâ sistemlerini ifade eder.

Üretken yapay zekâ sistemleri farklı türlerde içerikler oluşturabilir:

* Metin
* Görsel
* Ses
* Video
* Yazılım kodu

Bu alandaki sistemlere örnek olarak:

* ChatGPT
* Claude
* Gemini
* Midjourney
* Sora

verilebilir.

Üretken yapay zekâ teknolojilerindeki gelişmeler, yapay zekânın günlük yaşamda ve farklı sektörlerde kullanımının son yıllarda önemli ölçüde yaygınlaşmasına katkıda bulunmuştur.

# 12. Veri Biliminin İşletmelere Sağladığı Faydalar

Veri bilimi, işletmelerin veriye dayalı kararlar almasına ve mevcut süreçlerini daha etkin şekilde yönetmesine yardımcı olur.

Başlıca faydaları şunlardır:

* Operasyonel verimlilik
* Maliyet azaltma
* Gelir artırma
* Risk yönetimi
* Müşteri memnuniyeti
* Rekabet avantajı

Veri odaklı karar alma yaklaşımı, kurumların **daha doğru ve hızlı kararlar almasına** yardımcı olmakta ve sürdürülebilir büyümede önemli bir rol oynamaktadır.

---

# 13. Yapay Zekâ Çağında Gerekli Yetkinlikler

Veri ve yapay zekâ teknolojilerinin hızla geliştiği günümüzde bireylerin yalnızca teknik araçları kullanabilmesi değil, aynı zamanda **veriyi doğru şekilde yorumlayabilmesi ve problem çözme becerilerini geliştirmesi** önem taşımaktadır.

Bu kapsamda geliştirilmesi gereken temel yetkinlikler şunlardır:

* Veri okuryazarlığı
* Analitik düşünme
* İstatistiksel düşünme
* Problem çözme
* Programlama
* Yapay zekâ araçlarını kullanabilme
* Sürekli öğrenme yeteneği

Teknolojik gelişmelerin hızlandığı günümüzde **öğrenmeyi öğrenmek**, teknik becerilerin kendisi kadar önemli hâle gelmiştir.

# Sonuç

Veri bilimi ve yapay zekâ, dijital çağın önemli disiplinleri arasında yer almaktadır.

Teknolojik gelişmelerin hızlanması ve her geçen gün daha fazla veri üretilmesi, bireyleri ve kurumları veriyi yalnızca depolayan değil, aynı zamanda **anlamlandıran ve karar süreçlerinde etkin biçimde kullanan** yapılara dönüşmeye yöneltmektedir.

Bu süreçte veri;

**Ham Veri → Enformasyon → Bilgi → Karar**

şeklinde ilerleyen bir dönüşüm içerisinde değer kazanmaktadır.

Veri bilimi, bu dönüşümün gerçekleştirilmesinde **istatistik, matematik, bilgisayar bilimleri ve alan bilgisini** bir araya getirirken; yapay zekâ ve özellikle makine öğrenmesi, verilerden öğrenebilen ve belirli görevleri gerçekleştirebilen sistemlerin geliştirilmesini mümkün kılmaktadır.

Günümüzde veri yalnızca teknolojik bir unsur değil, aynı zamanda **ekonomik ve stratejik bir değerdir.**

Bu nedenle;

* Veri okuryazarlığı,
* Analitik düşünme,
* Problem çözme,
* Programlama,
* Yapay zekâ teknolojilerini etkin kullanabilme

gibi beceriler giderek daha önemli hâle gelmektedir.

Sonuç olarak veri bilimi ve yapay zekâ alanlarında temel bilgi sahibi olmak, yalnızca bu alanlarda uzmanlaşmak isteyen bireyler için değil, farklı sektörlerde çalışan ve geleceğin çalışma dünyasına uyum sağlamak isteyen bireyler için de önemli bir avantaj sağlamaktadır.

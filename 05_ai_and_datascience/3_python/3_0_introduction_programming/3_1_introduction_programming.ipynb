# Programlamaya Giriş

## 1. Programlama Nedir?

Programlama, bir donanıma (bilgisayar, telefon vb.) belirli bir görevi yerine getirmesi için adım adım talimatlar verme sürecidir.  

Bilgisayarlar insanlar gibi düşünmezler. Kendilerine verilen komutları belirli bir sıra ve mantık içerisinde uygular.  
Bu nedenle bir programcı, çözmek istediği problemi bilgisayarın anlayabileceği açık ve uygulanabilir adımlara dönüştürür.  

Örneğin bir öğrencinin not ortalamasını hesaplayan program şu işlemleri gerçekleştirir:

1. Öğrencinin notlarını alır.
2. Notları toplar.
3. Toplamı not sayısına böler.
4. Ortalamayı kullanıcıya gösterir.

Bu işlemlerin bir programlama dili kullanılarak bilgisayara aktarılmasına **programlama** adı verilir.

---

## 2. Program Nedir?

Program, bilgisayarın belirli bir görevi yerine getirmesi için hazırlanmış komutlar bütünüdür.

Bir hesap makinesi uygulaması, internet tarayıcısı, mobil uygulama, oyun veya veri analiz sistemi birer bilgisayar programıdır.

Bir program genel olarak üç temel aşamadan oluşur:

* Girdi alma
* Veriyi işleme
* Çıktı üretme

Örneğin:

```text
Girdi: Kullanıcının girdiği iki sayı
İşlem: Sayıların toplanması
Çıktı: Toplama sonucu
```

Python ile basit bir toplama programı:

```python
sayi1 = 10
sayi2 = 20

sonuc = sayi1 + sayi2

print(sonuc)
```

Programın çıktısı:

```text
30
```

## 3. Programlama Dili Nedir?

Programlama dili, bilgisayara komut vermek için kullanılan kurallar ve semboller bütünüdür.

Bilgisayarlar doğrudan makine dili adı verilen `0` ve `1` değerlerinden oluşan komutları çalıştırır.  
Ancak makine dili insanlar için okunması ve yazılması zor bir yapıya sahiptir.

Programlama dilleri, insanların bilgisayarlara daha anlaşılır biçimde komut vermesini sağlar.

Yaygın programlama dillerinden bazıları şunlardır:

* Python
* Java
* C
* C++
* C#
* JavaScript
* R
* Go

Her programlama dilinin kullanım alanı ve çalışma yapısı farklı olabilir.

Python; sade yazım kuralları, geniş kütüphane desteği ve veri bilimi alanındaki yaygın kullanımı nedeniyle programlamaya başlamak için uygun dillerden biridir.

## 5. Algoritma Nedir?

Algoritma, bir sorunu çözmek veya belirli bir hedefe ulaşmak için tasarlanan mantıksal, adım adım işlem basamakları bütünüdür.

Bir algoritmanın bilgisayar kodu içermesi zorunlu değildir. Günlük hayatta uyguladığımız birçok işlem de algoritmaya örnek olarak gösterilebilir.

Örneğin çay hazırlama algoritması:

1. Su ısıtılır.
2. Çay demliğe eklenir.
3. Sıcak su demliğe dökülür.
4. Çayın demlenmesi beklenir.
5. Çay bardağa dökülür.

Bir sayının çift veya tek olduğunu belirleyen algoritma:

1. Kullanıcıdan bir sayı al.
2. Sayının 2 ile bölümünden kalanını hesapla.
3. Kalan 0 ise sayının çift olduğunu belirt.
4. Kalan 0 değilse sayının tek olduğunu belirt.

Python ile uygulanışı:

```python
sayi = int(input("Bir sayı giriniz: "))

if sayi % 2 == 0:
    print("Sayı çifttir.")
else:
    print("Sayı tektir.")
```

## 6. Algoritmanın Temel Özellikleri

Bir algoritmanın aşağıdaki temel özellikleri taşıması beklenir.

### 6.1. Açıklık

Algoritmadaki her adım açık ve anlaşılır olmalıdır.

Yanlış örnek:

```text
Sayılarla gerekli işlemleri yap.
```

Doğru örnek:

```text
İki sayıyı topla ve sonucu ekrana yazdır.
```

### 6.2. Sonluluk

Algoritma belirli sayıda adımdan sonra sona ermelidir.

### 6.3. Sıralılık

Adımlar mantıksal bir sıra içerisinde uygulanmalıdır.

### 6.4. Girdi

Algoritma dışarıdan bir veya daha fazla veri alabilir.

### 6.5. Çıktı

Algoritma işlemler sonucunda en az bir sonuç üretmelidir.

### 6.6. Uygulanabilirlik

Algoritmadaki işlemler bilgisayar veya uygulayıcı tarafından gerçekleştirilebilir olmalıdır.

## 7. Problem Çözme Süreci

Programlama yalnızca kod yazmaktan ibaret değildir. Kod yazmadan önce problemin anlaşılması ve çözümün planlanması gerekir.

Temel problem çözme süreci şu aşamalardan oluşur:

1. Problemi tanımlama
2. Girdileri belirleme
3. Beklenen çıktıyı belirleme
4. Çözüm adımlarını oluşturma
5. Algoritmayı yazma
6. Algoritmayı programlama diline aktarma
7. Programı test etme
8. Hataları düzeltme
9. Programı geliştirme

Örneğin bir dikdörtgenin alanını hesaplayan program için:

**Problem:**

Dikdörtgenin alanını hesaplamak.

**Girdiler:**

* Kısa kenar
* Uzun kenar

**İşlem:**

```text
Alan = Kısa Kenar × Uzun Kenar
```

**Çıktı:**

Dikdörtgenin alanı

Python kodu:

```python
kisa_kenar = 5
uzun_kenar = 10

alan = kisa_kenar * uzun_kenar

print(alan)
```

## 8. Sözde Kod Nedir?

Sözde kod, bir algoritmanın herhangi bir programlama dilinin yazım kurallarına bağlı kalmadan, günlük dile yakın biçimde ifade edilmesidir.

Sözde kod doğrudan bilgisayar tarafından çalıştırılamaz. Amacı, problemin çözüm mantığını kod yazmadan önce planlamaktır.

Örneğin bir öğrencinin sınavdan geçip geçmediğini belirleyen sözde kod:

```text
Başla

Notu kullanıcıdan al

Eğer not 50 veya daha büyükse
    "Geçti" yazdır
Değilse
    "Kaldı" yazdır

Bitir
```

Python karşılığı:

```python
not_degeri = int(input("Notunuzu giriniz: "))

if not_degeri >= 50:
    print("Geçti")
else:
    print("Kaldı")
```

## 9. Akış Şeması Nedir?

Akış şeması, bir algoritmanın semboller kullanılarak görsel biçimde gösterilmesidir.

Akış şemaları, programın hangi adımlardan oluştuğunu ve karar noktalarının nasıl ilerlediğini anlamayı kolaylaştırır.

Yaygın akış şeması sembolleri:

| Sembol          | Anlamı               |
| --------------- | -------------------- |
| Oval            | Başlangıç veya bitiş |
| Dikdörtgen      | İşlem                |
| Paralelkenar    | Girdi veya çıktı     |
| Eşkenar dörtgen | Karar                |
| Ok              | İşlem yönü           |

Örneğin bir sayının pozitif veya negatif olduğunu belirleyen akış:

```text
Başla
  |
Sayıyı al
  |
Sayı > 0 mı?
 /        \
Evet      Hayır
 |          |
Pozitif    Pozitif değil
  \        /
     Bitir
```

## 14. Değişken Nedir?

Değişken, program içerisinde bir değeri saklamak için kullanılan isimlendirilmiş bellek alanıdır.

Python örneği:

```python
yas = 26
isim = "Onur"
maas = 45000
```

Bu örnekte:

* `yas` değişkeni `26` değerini,
* `isim` değişkeni `"Onur"` değerini,
* `maas` değişkeni `45000` değerini saklar.

Bir değişkenin değeri program içerisinde değiştirilebilir:

```python
puan = 70
puan = 85

print(puan)
```

Çıktı:

```text
85
```

Veri türü, bir değişkenin hangi türde değer sakladığını belirtir.

Python'da temel veri türleri şunlardır:

| Veri türü  | Açıklama          | Örnek           |
| ---------- | ----------------- | --------------- |

yas = 26

yas = 26

## 17. Operatör Nedir?

Operatörler, değerler ve değişkenler üzerinde işlem yapmak için kullanılan sembollerdir.

### 17.1. Aritmetik Operatörler

| Operatör | İşlem          |
| -------- | -------------- |
| `+`      | Toplama        |
| `-`      | Çıkarma        |
| `*`      | Çarpma         |
| `/`      | Bölme          |
| `//`     | Tam bölme      |
| `%`      | Bölümden kalan |
| `**`     | Üs alma        |

Örnek:

```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```

### 17.2. Karşılaştırma Operatörleri

| Operatör | Anlamı             |
| -------- | ------------------ |
| `==`     | Eşittir            |
| `!=`     | Eşit değildir      |
| `>`      | Büyüktür           |
| `<`      | Küçüktür           |
| `>=`     | Büyük veya eşittir |
| `<=`     | Küçük veya eşittir |

Örnek:

```python
yas = 26

print(yas >= 18)
```

Çıktı:

```text
True
```

## 18. Girdi ve Çıktı İşlemleri

Programların kullanıcıdan veri almasına girdi, kullanıcıya sonuç göstermesine çıktı adı verilir.

Python'da çıktı vermek için `print()` fonksiyonu kullanılır:

```python
print("Python programlamaya giriş")
```

Kullanıcıdan veri almak için `input()` fonksiyonu kullanılır:

```python
isim = input("Adınızı giriniz: ")

print("Merhaba", isim)
```

`input()` fonksiyonundan alınan değer varsayılan olarak metin türündedir.

Sayısal işlem yapılacaksa veri türü dönüştürülmelidir:

```python
yas = int(input("Yaşınızı giriniz: "))

print(yas + 1)
```

## 19. Koşullu İfadeler

Koşullu ifadeler, programın belirli bir koşula göre farklı işlemler gerçekleştirmesini sağlar.

Python'da temel koşul yapıları şunlardır:

* `if`
* `elif`
* `else`

Örnek:

```python
not_degeri = 75

if not_degeri >= 50:
    print("Başarılı")
else:
    print("Başarısız")
```

Birden fazla koşul kullanılabilir:

```python
not_degeri = 75

if not_degeri >= 85:
    print("Pekiyi")
elif not_degeri >= 70:
    print("İyi")
elif not_degeri >= 50:
    print("Orta")
else:
    print("Başarısız")
```

## 20. Döngüler

Döngüler, belirli işlemlerin birden fazla kez gerçekleştirilmesini sağlar.

Python'da iki temel döngü yapısı bulunur:

* `for`
* `while`

### 20.1. For Döngüsü

Belirli bir veri grubu üzerinde sırasıyla işlem yapmak için kullanılır.

```python
for sayi in range(1, 6):
    print(sayi)
```

Çıktı:

```text
1
2
3
4
5
```

### 20.2. While Döngüsü

Belirli bir koşul doğru olduğu sürece çalışır.

```python
sayi = 1

while sayi <= 5:
    print(sayi)
    sayi = sayi + 1
```

## 21. Fonksiyon Nedir?

Fonksiyon, belirli bir görevi gerçekleştiren ve ihtiyaç duyulduğunda tekrar kullanılabilen kod bloğudur.

Fonksiyonlar kod tekrarını azaltır ve programın daha düzenli olmasını sağlar.

Python'da fonksiyon tanımlamak için `def` anahtar kelimesi kullanılır.

```python
def selamla():
    print("Merhaba")
```

Fonksiyonu çalıştırmak için adı yazılır:

```python
selamla()
```

Parametre alan bir fonksiyon:

```python
def selamla(isim):
    print("Merhaba", isim)

selamla("Onur")
```

Değer döndüren bir fonksiyon:

```python
def topla(sayi1, sayi2):
    return sayi1 + sayi2

sonuc = topla(10, 20)

print(sonuc)
```

## 22. Hata Nedir?

Programın beklenen biçimde çalışmasını engelleyen durumlara hata adı verilir.

Programlama hataları genel olarak üç grupta incelenebilir.

### 22.1. Söz Dizimi Hatası

Programlama dilinin yazım kurallarına uyulmadığında oluşur.

Hatalı örnek:

```python
if yas >= 18
    print("Yetişkin")
```

Doğru kullanım:

```python
if yas >= 18:
    print("Yetişkin")
```

### 22.2. Çalışma Zamanı Hatası

Program çalışırken meydana gelir.

```python
sonuc = 10 / 0
```

Bir sayı sıfıra bölünemeyeceği için hata oluşur.

### 22.3. Mantık Hatası

Program çalışır ancak yanlış sonuç üretir.

```python
kisa_kenar = 5
uzun_kenar = 10

alan = kisa_kenar + uzun_kenar
```

Kod çalışır ancak dikdörtgen alanı toplama ile değil, çarpma ile hesaplanmalıdır.

Doğru kullanım:

```python
alan = kisa_kenar * uzun_kenar
```

## 23. Hata Ayıklama Nedir?

Hata ayıklama veya debugging, programdaki hataların bulunması ve düzeltilmesi sürecidir.

Hata ayıklama sırasında şu adımlar izlenebilir:

1. Hata mesajı dikkatlice okunur.
2. Hatanın bulunduğu satır belirlenir.
3. Kullanılan değişkenler kontrol edilir.
4. Veri türleri incelenir.
5. Program küçük parçalar hâlinde test edilir.
6. Ara sonuçlar `print()` ile görüntülenir.
7. Beklenen ve elde edilen sonuç karşılaştırılır.

Örnek:

```python
sayi1 = 10
sayi2 = "20"

print(type(sayi1))
print(type(sayi2))
```

Burada `sayi1` tam sayı, `sayi2` ise metindir. Toplama yapılmadan önce veri türleri uyumlu hâle getirilmelidir.

```python
sayi2 = int(sayi2)

sonuc = sayi1 + sayi2

print(sonuc)
```

## 24. Programlama ve Veri Bilimi Arasındaki İlişki

Veri bilimi; verilerin toplanması, düzenlenmesi, analiz edilmesi, yorumlanması ve sonuçların karar verme süreçlerinde kullanılmasıyla ilgilenen disiplinler arası bir alandır.

Programlama, veri biliminin temel araçlarından biridir.

Bir veri bilimci programlama kullanarak:

* Verileri dosyalardan veya veritabanlarından alabilir.
* Verileri temizleyebilir.
* Eksik ve hatalı değerleri düzeltebilir.
* Hesaplamalar yapabilir.
* Verileri görselleştirebilir.
* İstatistiksel analizler gerçekleştirebilir.
* Makine öğrenmesi modelleri geliştirebilir.
* Tekrarlanan işlemleri otomatikleştirebilir.

Python, veri bilimi alanında en yaygın kullanılan programlama dillerinden biridir.

Ancak veri bilimine başlamadan önce değişkenler, veri türleri, koşullar, döngüler, fonksiyonlar ve hata yönetimi gibi temel programlama konularının öğrenilmesi gerekir.

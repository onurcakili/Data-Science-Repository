Python'da birden fazla değeri saklamak için koleksiyon yapıları kullanılır. Veri bilimine geçmeden önce `list`, `tuple`, `set` ve `dict` mantığını iyi anlamak gerekir.

## Konu Dokümantasyonu

Python'da birden fazla değeri tek bir yapı altında saklamak için koleksiyonlar kullanılır. Günlük konuşmada bunlara çoğu zaman dizi denir; Python'da en sık kullanılan karşılığı `list` yapısıdır.

Temel koleksiyon türleri:

* `list`: Sıralı ve değiştirilebilir yapı
* `tuple`: Sıralı ama değiştirilemez yapı
* `set`: Benzersiz değerlerden oluşan yapı
* `dict`: Anahtar-değer çiftleriyle çalışan yapı

Liste, aynı türden değerleri sırayla saklamak için idealdir. Örneğin satış tutarları, öğrenci notları veya ürün fiyatları liste olabilir. Dictionary ise bir kaydı anlamlı alan adlarıyla temsil etmek için kullanılır.

Veri bilimine geçmeden önce koleksiyon mantığını öğrenmek çok önemlidir. Çünkü Pandas DataFrame, NumPy array ve JSON verileri bu temel fikirlerin üzerine kurulur.


## Kolay Seviye

Bu seviyede liste oluşturmayı, listedeki elemanlara erişmeyi ve listeyi güncellemeyi öğreneceğiz.

Listelerde indeksleme `0` ile başlar. Yani ilk elemanın indeksi `0`, ikinci elemanın indeksi `1` olur. Negatif indeksler ise sondan erişim sağlar.

Liste değiştirilebilir bir yapıdır. Bu yüzden eleman eklenebilir, silinebilir veya güncellenebilir.


````python
notlar = [70, 85, 90, 60, 95]

print(notlar[0])          # İlk eleman: indeks 0
print(notlar[2])          # Üçüncü eleman: indeks 2
print(notlar[-1])         # Son eleman: negatif indeks sondan başlar

notlar.append(100)        # Listenin sonuna yeni değer ekler
print(notlar)
````

````python
urunler = ["laptop", "telefon", "tablet"]

urunler[1] = "akıllı telefon"       # İndeks 1'deki değer güncellenir
urunler.remove("tablet")            # Değere göre silme yapar

print(urunler)
````

## Orta Seviye

Bu seviyede slicing, list comprehension, `set` ve `dict` yapılarını kullanacağız.

Slicing, listenin belirli bir bölümünü almak için kullanılır. Genel yapı şudur:

```python
liste[başlangıç:bitiş:adım]
```

Bitiş indeksi dahil edilmez. List comprehension ise döngüyle yeni liste üretmenin kısa ve okunabilir yoludur. Dictionary yapısı, alan adıyla değer saklamamızı sağlar; bu nedenle tablo satırı veya kayıt mantığında çok kullanılır.


````python
sayilar = [10, 20, 30, 40, 50, 60]

print(sayilar[1:4])       # [başlangıç:bitiş] -> 1 dahil, 4 dahil değil
print(sayilar[:3])        # Baştan indeks 3'e kadar alır
print(sayilar[::2])       # [başlangıç:bitiş:adım] -> ikişer ikişer ilerler
print(sayilar[::-1])      # Listeyi tersten okur
````

````python
satislar = [1200, 950, 1800, 2100, 1750]

yuksek_satislar = [satis for satis in satislar if satis >= 1500]
# Üstteki yapı: her satis değerini gez, koşulu sağlayanı yeni listeye ekle

kdv_dahil = [satis * 1.20 for satis in satislar]
# Her satış tutarının KDV dahil değerini üretir

print(yuksek_satislar)
print(kdv_dahil)
````

````python
kategoriler = ["A", "B", "A", "C", "B", "D"]
benzersiz_kategoriler = set(kategoriler)        # set tekrar eden değerleri tekilleştirir

ogrenci = {
    "ad": "Ayşe",
    "yas": 22,
    "not": 88
}

print(benzersiz_kategoriler)
print(ogrenci["ad"])                            # Dictionary değerlerine anahtar ile erişilir
````

## İleri Seviye

Bu seviyede iç içe veri yapılarıyla çalışacağız.

Gerçek hayatta veriler çoğu zaman tek bir düz liste olarak gelmez. Liste içinde dictionary, dictionary içinde liste veya iç içe JSON benzeri yapılar sık görülür. Bu yapılarda filtreleme, toplama, sayma ve gruplama işlemleri veri biliminin temel pratiklerindendir.

`collections.Counter` gibi yardımcı sınıflar, tekrar eden değerleri saymak için kullanışlıdır. Bu tarz araçlar basit ama gerçek analizlerde çok işe yarar.


````python
satis_kayitlari = [
    {"urun": "Laptop", "kategori": "Elektronik", "tutar": 30000},
    {"urun": "Telefon", "kategori": "Elektronik", "tutar": 20000},
    {"urun": "Masa", "kategori": "Mobilya", "tutar": 7500},
    {"urun": "Sandalye", "kategori": "Mobilya", "tutar": 2500},
]

elektronik_satislar = [
    kayit for kayit in satis_kayitlari
    if kayit["kategori"] == "Elektronik"        # Sadece elektronik kategorisini seçer
]

toplam_elektronik = sum(kayit["tutar"] for kayit in elektronik_satislar)

print(elektronik_satislar)
print("Elektronik toplam:", toplam_elektronik)
````

````python
from collections import Counter

sehirler = ["İstanbul", "Ankara", "İstanbul", "İzmir", "Ankara", "İstanbul"]
sehir_sayilari = Counter(sehirler)               # Her değerin kaç kez geçtiğini sayar

print(sehir_sayilari)
print(sehir_sayilari.most_common(1))             # En sık geçen 1 değeri gösterir
````

## Mini Alıştırma ve Özet

Bu bölümde koleksiyonların veriyi düzenli tutmak için kullanıldığını gördük.

Hatırlanması gerekenler:

* Liste sıralı ve değiştirilebilir yapıdadır.
* İndeksler `0` ile başlar.
* Slicing ile listenin parçası alınır.
* `set` benzersiz değerleri tutar.
* `dict` anahtar-değer mantığıyla çalışır.
* İç içe koleksiyonlar gerçek veri yapılarını temsil eder.

Şimdi müşteri kayıtları üzerinde filtreleme ve ortalama hesaplama yapalım.


````python
musteriler = [
    {"ad": "Ali", "yas": 17},
    {"ad": "Ayşe", "yas": 24},
    {"ad": "Mehmet", "yas": 31},
    {"ad": "Zeynep", "yas": 16},
]

yetiskinler = [musteri for musteri in musteriler if musteri["yas"] >= 18]
ortalama_yas = sum(musteri["yas"] for musteri in yetiskinler) / len(yetiskinler)

print(yetiskinler)
print("Ortalama yaş:", ortalama_yas)
````

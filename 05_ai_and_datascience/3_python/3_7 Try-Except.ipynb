# Python'da Try-Except ile Hata Yönetimi

Hata yönetimi, programın beklenmeyen durumlarda tamamen durmasını engeller. Veri biliminde bozuk veri, yanlış tür, eksik dosya ve sıfıra bölme gibi durumlar sık görülür.

Öğrenme akışı:

* **Kolay seviye:** Basit `try-except`
* **Orta seviye:** Belirli hata türleri, `else`, `finally`
* **İleri seviye:** Güvenli dönüştürme fonksiyonları, `raise`, kontrollü veri temizleme


## Konu Dokümantasyonu

Hata yönetimi, programın beklenmeyen durumlarda kontrollü davranmasını sağlar. Hata oluştuğunda programın tamamen durması yerine kullanıcıya anlaşılır mesaj göstermek, varsayılan değer üretmek veya hatalı veriyi raporlamak mümkündür.

Python'da hata yönetimi için temel yapı `try-except` bloğudur.

Yaygın hata türleri:

* `ValueError`: Değer uygun formatta değildir.
* `TypeError`: Veri türü işlem için uygun değildir.
* `ZeroDivisionError`: Sıfıra bölme yapılmıştır.
* `FileNotFoundError`: Dosya bulunamamıştır.
* `KeyError`: Dictionary içinde olmayan anahtar istenmiştir.

Veri biliminde hata yönetimi özellikle ham veriyle çalışırken önemlidir. Bozuk satırlar, eksik değerler, yanlış türler veya eksik dosyalar analiz sürecini durdurmamalı; kontrollü şekilde ele alınmalıdır.


## Kolay Seviye

Bu seviyede temel `try-except` yapısını öğreneceğiz.

`try` bloğuna hata çıkarma ihtimali olan kod yazılır. Hata oluşursa Python uygun `except` bloğuna geçer. Böylece program kontrollü şekilde devam eder.


```python
try:
    sonuc = 10 / 0
except ZeroDivisionError:
    print("Bir sayı sıfıra bölünemez.")  # Program durmak yerine kontrollü mesaj verir
```

```python
deger = "abc"

try:
    sayi = int(deger)                    # "abc" sayıya çevrilemez
except ValueError:
    print("Değer sayıya dönüştürülemedi.")
```

## Orta Seviye

Bu seviyede belirli hata türlerini yakalamayı, `else` ve `finally` bloklarını kullanmayı göreceğiz.

Belirli hata türünü yakalamak, genel `except` yazmaktan daha doğrudur. Çünkü her hata için farklı çözüm gerekebilir. `else` hata oluşmadığında, `finally` ise hata olsa da olmasa da çalışır.


```python
deger = "0"

try:
    sayi = int(deger)
    sonuc = 100 / sayi
except ValueError:
    print("Sayıya çevrilemeyen değer var.")
except ZeroDivisionError:
    print("Sıfıra bölme hatası var.")
else:
    print("Sonuç:", sonuc)               # Hata yoksa çalışır
finally:
    print("İşlem denemesi tamamlandı.")  # Hata olsa da olmasa da çalışır
```

```python
fiyatlar = ["100", "250", "hatalı", "400"]
temiz_fiyatlar = []

for fiyat in fiyatlar:
    try:
        temiz_fiyatlar.append(float(fiyat))
    except ValueError:
        print("Geçersiz fiyat atlandı:", fiyat)

print(temiz_fiyatlar)
```

## İleri Seviye

Bu seviyede hata yönetimini veri temizleme fonksiyonlarına taşıyacağız.

Amaç hataları saklamak değil, anlamlı şekilde yönetmektir. Örneğin sayıya çevrilemeyen değerler analiz dışı bırakılabilir, raporlanabilir veya varsayılan değerle doldurulabilir.

`raise`, bilinçli olarak hata üretmek için kullanılır. Geçersiz veri programın sessizce yanlış sonuç üretmesine neden olacaksa hata yükseltmek daha sağlıklıdır.


```python
def guvenli_float(deger, varsayilan=None):
    try:
        return float(deger)
    except (TypeError, ValueError):
        return varsayilan              # Hatalı değer gelirse None veya verilen varsayılan döner


ham_degerler = ["10.5", "20", None, "hatalı", "35"]
temiz_degerler = [guvenli_float(deger) for deger in ham_degerler]

print(temiz_degerler)
```

```python
def yas_kontrol_et(yas):
    if yas < 0:
        raise ValueError("Yaş negatif olamaz.")  # Geçersiz veri bilinçli olarak reddedilir
    return yas


try:
    print(yas_kontrol_et(25))
    print(yas_kontrol_et(-3))
except ValueError as hata:
    print("Hata:", hata)
```

## Mini Alıştırma ve Özet

Bu bölümde hataları kontrollü şekilde yönetmeyi öğrendik.

Hatırlanması gerekenler:

* `try` hata riski olan kodu içerir.
* `except` hata oluşursa çalışır.
* Belirli hata türlerini yakalamak daha sağlıklıdır.
* `else` başarılı işlemden sonra çalışır.
* `finally` her durumda çalışır.
* Hatalı veri analiz sürecinde raporlanmalıdır.

Şimdi karışık satış verilerini temizleyip geçersiz değerleri ayrı raporlayalım.


```python
ham_satislar = ["1200", "950", "boş", "1800", "", "2100"]
temiz_satislar = []
hatali_degerler = []

for deger in ham_satislar:
    try:
        temiz_satislar.append(float(deger))
    except ValueError:
        hatali_degerler.append(deger)          # Hatalı değerleri tamamen kaybetmeden saklıyoruz

print("Temiz satışlar:", temiz_satislar)
print("Hatalı değerler:", hatali_degerler)
print("Toplam:", sum(temiz_satislar))
```

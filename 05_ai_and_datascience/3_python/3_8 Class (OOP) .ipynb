# Python'da Sınıflar ve Nesne Yönelimli Programlama

Sınıflar, veri ve davranışı tek yapı altında toplar. Küçük scriptlerde fonksiyonlar yeterli olabilir; ancak veri işleme adımları büyüdükçe sınıflar kodu düzenlemeye yardımcı olur.

Öğrenme akışı:

* **Kolay seviye:** `class`, nesne, `__init__`, `self`
* **Orta seviye:** Metot, `__repr__`, class attribute
* **İleri seviye:** `dataclass`, özellik doğrulama, küçük analiz sınıfı


## Konu Dokümantasyonu

Nesne yönelimli programlama, veri ve o veriyle çalışan davranışları aynı yapı altında toplamaya yarar. Bu yapıya sınıf denir. Sınıftan oluşturulan örneğe nesne denir.

Temel kavramlar:

* `class`: Nesne üretmek için şablon
* Nesne: Sınıftan oluşturulan örnek
* Özellik: Nesnenin taşıdığı veri
* Metot: Nesnenin yapabildiği işlem
* `self`: Mevcut nesneyi temsil eden referans
* `__init__`: Nesne oluşurken çalışan başlangıç metodu

Veri bilimi tarafında sınıflar; analiz akışını, veri dönüştürme adımlarını, model ayarlarını veya tekrar kullanılan yardımcı araçları düzenlemek için kullanılır. Her küçük script için sınıf şart değildir; fakat proje büyüdükçe sınıflar kodu toparlar.


## Kolay Seviye

Bu seviyede sınıf ve nesne mantığını öğreneceğiz.

`class` bir şablondur. Bu şablondan nesne oluşturduğumuzda her nesne kendi değerlerini taşıyabilir. `__init__` metodu nesne oluşturulurken başlangıç değerlerini alır ve `self` üzerinden nesneye kaydeder.


```python
class Ogrenci:
    pass                         # Şimdilik boş sınıf


ogrenci = Ogrenci()              # Ogrenci sınıfından nesne oluşturulur
print(type(ogrenci))
```

```python
class Ogrenci:
    def __init__(self, ad, not_ortalamasi):
        self.ad = ad
        self.not_ortalamasi = not_ortalamasi
        # self, oluşturulan mevcut nesneyi temsil eder


ogrenci = Ogrenci("Ayşe", 88)
print(ogrenci.ad)
print(ogrenci.not_ortalamasi)
```

## Orta Seviye

Bu seviyede sınıflara metot eklemeyi göreceğiz.

Metotlar, sınıf içinde tanımlanan fonksiyonlardır. Nesnenin kendi özelliklerini kullanarak işlem yaparlar. Örneğin bir ürün nesnesi kendi fiyatını bilir ve indirimli fiyatını hesaplayabilir.

`__repr__` nesneyi ekrana daha okunabilir yazdırmak için kullanılır. Class attribute ise tüm nesneler için ortak olan bilgiyi temsil eder.


```python
class Urun:
    para_birimi = "TL"           # class attribute: tüm ürünler için ortak bilgi

    def __init__(self, ad, fiyat):
        self.ad = ad
        self.fiyat = fiyat

    def indirimli_fiyat(self, indirim_orani):
        return self.fiyat - (self.fiyat * indirim_orani)

    def __repr__(self):
        return f"Urun(ad={self.ad}, fiyat={self.fiyat} {self.para_birimi})"


urun = Urun("Laptop", 30000)
print(urun)
print(urun.indirimli_fiyat(0.10))
```

```python
class SatisAnalizi:
    def __init__(self, satislar):
        self.satislar = satislar

    def toplam_satis(self):
        return sum(self.satislar)

    def ortalama_satis(self):
        return sum(self.satislar) / len(self.satislar)

    def yuksek_satislari_getir(self, esik_deger):
        return [satis for satis in self.satislar if satis >= esik_deger]


analiz = SatisAnalizi([1200, 950, 1800, 2100])
print("Toplam:", analiz.toplam_satis())
print("Ortalama:", analiz.ortalama_satis())
print("Yüksek satışlar:", analiz.yuksek_satislari_getir(1500))
```

## İleri Seviye

Bu seviyede veri odaklı sınıflarda kullanılan pratik yapılara bakacağız.

`dataclass`, çoğunlukla veri taşıyan sınıfları kısa ve okunabilir yazmayı sağlar. Model ayarları, konfigürasyonlar veya basit kayıt yapıları için kullanışlıdır.

Ayrıca veri dönüştürme mantığını sınıfa almak, `fit` ve `transform` gibi makine öğrenmesi kütüphanelerinde sık görülen bir yaklaşımdır. Bu yapı, öğrenilen bilgiyi nesnede saklayıp sonra yeni veri üzerinde kullanmayı sağlar.


```python
from dataclasses import dataclass


@dataclass
class ModelAyari:
    model_adi: str
    test_orani: float = 0.20
    random_state: int = 42


ayar = ModelAyari(model_adi="Karar Ağacı")
print(ayar)
```

```python
class OrtalamaDoldurucu:
    def __init__(self):
        self.ortalama = None

    def fit(self, degerler):
        temiz_degerler = [deger for deger in degerler if deger is not None]
        self.ortalama = sum(temiz_degerler) / len(temiz_degerler)
        return self                         # return self zincirleme kullanım için yaygındır

    def transform(self, degerler):
        return [
            self.ortalama if deger is None else deger
            for deger in degerler
        ]


doldurucu = OrtalamaDoldurucu()
veri = [10, None, 20, 30, None]

doldurucu.fit(veri)
print(doldurucu.transform(veri))
```

## Mini Alıştırma ve Özet

Bu bölümde sınıfların veri ve davranışı birlikte tuttuğunu öğrendik.

Hatırlanması gerekenler:

* `class` nesne şablonudur.
* `__init__` başlangıç değerlerini alır.
* `self` mevcut nesneyi temsil eder.
* Metotlar nesnenin davranışlarını tanımlar.
* `dataclass` veri taşıyan sınıfları sadeleştirir.
* Sınıflar büyük analiz projelerinde kod organizasyonu sağlar.

Şimdi müşteri harcamalarını sınıf içinde modelleyelim.


```python
class Musteri:
    def __init__(self, ad, harcamalar):
        self.ad = ad
        self.harcamalar = harcamalar

    def toplam_harcama(self):
        return sum(self.harcamalar)

    def segment(self):
        toplam = self.toplam_harcama()

        if toplam >= 50_000:
            return "A"
        if toplam >= 10_000:
            return "B"
        return "C"


musteri = Musteri("Onur", [12000, 8500, 42000])
print(musteri.toplam_harcama())
print(musteri.segment())
```

# Python'da Koşullu İfadeler

Koşullu ifadeler, programın karar verebilmesini sağlar. Veri analizinde koşullar genellikle değerleri sınıflandırmak, hatalı kayıtları ayıklamak veya farklı hesaplama yolları seçmek için kullanılır.

Öğrenme akışı:

* **Kolay seviye:** `if`, `else`, karşılaştırma operatörleri
* **Orta seviye:** `elif`, `and`, `or`, üyelik kontrolü
* **İleri seviye:** Veri doğrulama, guard clause, sözlük ile karar haritası


## Konu Dokümantasyonu

Koşullu ifadeler, programın karar vermesini sağlar. Bir koşul doğruysa bir işlem, yanlışsa başka bir işlem yapılabilir.

Python'da temel koşul yapıları:

* `if`: Koşul doğruysa çalışır.
* `elif`: Önceki koşul yanlışsa yeni koşul dener.
* `else`: Hiçbir koşul doğru değilse çalışır.

Koşullar genellikle karşılaştırma operatörleriyle kurulur: `==`, `!=`, `>`, `<`, `>=`, `<=`. Birden fazla koşul varsa `and`, `or`, `not` kullanılır.

Veri bilimi tarafında koşullar; veri temizleme, kategori üretme, aykırı değer işaretleme, müşteri segmentasyonu ve iş kurallarını uygulama gibi alanlarda kullanılır.


## Kolay Seviye

Bu seviyede `if` ve `else` mantığını öğreneceğiz.

Koşul sonucu `True` ise `if` bloğu çalışır. Koşul sonucu `False` ise varsa `else` bloğu çalışır. Python'da bloklar girinti ile belirlenir, bu yüzden girinti kullanımı önemlidir.


```python
yas = 20

if yas >= 18:                    # Koşul True olduğu için bu blok çalışır
    print("Yetişkin")
else:
    print("Çocuk")
```

```python
not_degeri = 45

if not_degeri >= 50:
    print("Geçti")
else:
    print("Kaldı")               # 45 >= 50 olmadığı için else çalışır
```

## Orta Seviye

Bu seviyede birden fazla koşulu yöneteceğiz.

`elif`, çok seçenekli karar yapılarında kullanılır. Python koşulları yukarıdan aşağıya kontrol eder ve ilk doğru koşulu bulduğunda ilgili bloğu çalıştırır.

`and` tüm koşulların doğru olmasını ister. `or` ise koşullardan en az biri doğruysa yeterlidir. `in` operatörü, bir değerin liste veya benzeri koleksiyon içinde bulunup bulunmadığını kontrol eder.


```python
not_degeri = 78

if not_degeri >= 85:
    harf_notu = "AA"
elif not_degeri >= 70:           # İlk doğru koşul burada bulunduğu için aşağısı çalışmaz
    harf_notu = "BB"
elif not_degeri >= 50:
    harf_notu = "CC"
else:
    harf_notu = "FF"

print(harf_notu)
```

```python
not_ortalamasi = 64
devamsizlik = 8

if not_ortalamasi >= 50 and devamsizlik <= 10:
    # and kullanıldığı için iki koşul da doğru olmalıdır
    print("Dersi geçti")
else:
    print("Dersi geçemedi")
```

```python
kategori = "Elektronik"
kampanyali_kategoriler = ["Elektronik", "Giyim", "Kitap"]

if kategori in kampanyali_kategoriler:      # in, değerin koleksiyon içinde olup olmadığını kontrol eder
    print("Kampanya uygulanabilir.")
else:
    print("Kampanya yok.")
```

## İleri Seviye

Bu seviyede koşulları daha okunabilir ve sürdürülebilir yazmaya odaklanacağız.

Çok karmaşık `if-elif-else` zincirleri okunabilirliği azaltabilir. Fonksiyon içinde önce geçersiz durumları ayıklamak, sonra ana mantığa geçmek daha temiz bir yaklaşımdır. Buna guard clause denir.

Bazı kararlar sözlük ile de temsil edilebilir. Örneğin müşteri segmentine göre indirim oranı seçmek için uzun koşullar yerine dictionary kullanmak daha anlaşılır olabilir.


```python
def satis_kategorisi_bul(satis_tutari):
    if satis_tutari < 0:
        return "geçersiz"            # Geçersiz veri erken döndürülür

    if satis_tutari >= 50_000:
        return "çok yüksek"
    if satis_tutari >= 10_000:
        return "yüksek"
    if satis_tutari >= 1_000:
        return "orta"
    return "düşük"


print(satis_kategorisi_bul(12500))
print(satis_kategorisi_bul(-1))
```

```python
segment = "A"

indirim_oranlari = {
    "A": 0.20,
    "B": 0.10,
    "C": 0.05,
}

indirim = indirim_oranlari.get(segment, 0)   # Anahtar yoksa 0 döner, KeyError oluşmaz

print("İndirim oranı:", indirim)
```

## Mini Alıştırma ve Özet

Bu bölümde programın karar vermesini sağlayan koşullu ifadeleri öğrendik.

Hatırlanması gerekenler:

* `if` doğru koşulda çalışır.
* `else` alternatif durumdur.
* `elif` çoklu koşullar için kullanılır.
* `and`, `or`, `not` koşulları birleştirir.
* Koşullar veri sınıflandırma ve iş kuralları için çok kullanılır.

Şimdi bir kredi uygunluğu senaryosunu koşullarla değerlendirelim.


```python
gelir = 28_000
kredi_skoru = 735
aktif_borc = 42_000

gelir_uygun = gelir >= 20_000
skor_uygun = kredi_skoru >= 700
borc_uygun = aktif_borc < 50_000

if gelir_uygun and skor_uygun and borc_uygun:
    print("Kredi uygun")
else:
    print("Kredi uygun değil")
```

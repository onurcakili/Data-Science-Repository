# Python'da Veri Türleri

Veri türü, bir değerin hangi işlemlerde kullanılabileceğini belirler. Veri bilimi açısından bu konu çok kritiktir; çünkü sayısal görünen bir değer metin olarak saklanıyorsa matematiksel analiz yapılamaz.

Öğrenme akışı:

* **Kolay seviye:** `int`, `float`, `str`, `bool`, `None`
* **Orta seviye:** Tür dönüşümü, formatlama, doğrulama
* **İleri seviye:** Doğruluk değeri, `None` kontrolü, para hesabında `Decimal`


## Konu Dokümantasyonu

Veri türü, bir değerin program tarafından nasıl yorumlanacağını belirler. Python için `"10"` ile `10` aynı şey değildir. İlki metindir, ikincisi sayıdır.

Veri türlerini anlamak, hatasız program yazmanın temelidir. Çünkü her veri türünün desteklediği işlemler farklıdır. Sayılarla matematik yapılabilir, metinlerle birleştirme veya temizleme yapılabilir, mantıksal değerlerle karar mekanizmaları kurulabilir.

Temel veri türleri:

* `int`: Tam sayılar
* `float`: Ondalıklı sayılar
* `str`: Metinler
* `bool`: `True` veya `False`
* `NoneType`: Henüz değer olmadığını anlatan `None`

Veri biliminde veri türleri daha da kritik hâle gelir. Yaş, fiyat, satış adedi gibi alanlar metin olarak gelirse hesaplama yapılamaz. Tarih alanı yanlış türdeyse zaman analizi bozulur. Bu yüzden analizden önce veri türleri kontrol edilir ve gerekirse dönüştürülür.


## Kolay Seviye

Bu seviyede temel veri türlerini tanıyacağız.

Python'da `type()` fonksiyonu bir değişkenin hangi türde olduğunu gösterir. Bu kontrol özellikle yeni veriyle çalışırken önemlidir.

Örneğin `yas = 26` tam sayı, `boy = 1.78` ondalıklı sayı, `isim = "Onur"` metin, `mezun_mu = True` mantıksal değerdir. `None` ise değer yokluğunu temsil eder; bu, `0` veya boş metinle aynı şey değildir.


```python
yas = 26                         # int: tam sayı
boy = 1.78                       # float: ondalıklı sayı
isim = "Onur"                    # str: metin
mezun_mu = True                  # bool: mantıksal değer
ikinci_telefon = None            # None: şu an değer yok

print(type(yas))
print(type(boy))
print(type(isim))
print(type(mezun_mu))
print(type(ikinci_telefon))
```

```python
sayi1 = 10
sayi2 = 3

print(sayi1 + sayi2)             # Toplama
print(sayi1 / sayi2)             # Normal bölme sonucu float döner
print(sayi1 // sayi2)            # Tam bölme, küsuratı atar
print(sayi1 % sayi2)             # Kalanı verir
print(sayi1 ** sayi2)            # Üs alma: 10 üzeri 3
```

## Orta Seviye

Bu seviyede tür dönüşümü ve formatlama göreceğiz.

Dışarıdan gelen veriler çoğu zaman metin olarak gelir. Kullanıcıdan alınan input, CSV dosyasından okunan değer veya API'den gelen bazı alanlar sayı gibi görünse bile `str` olabilir. Matematiksel işlem yapmadan önce `int()` veya `float()` ile dönüştürme gerekebilir.

Ayrıca sonuçları okunabilir göstermek için f-string formatlama kullanılır. Para, oran ve rapor çıktılarında formatlama hem okunabilirliği hem profesyonelliği artırır.


```python
yas_metin = "26"
gelir_metin = "18500.75"

yas = int(yas_metin)                 # "26" metnini tam sayıya çevirir
gelir = float(gelir_metin)           # "18500.75" metnini ondalıklı sayıya çevirir

print(yas + 1)
print(gelir * 12)
```

```python
urun_adi = "Laptop"
fiyat = 32_500.5
stok_adedi = 7

mesaj = f"{urun_adi} fiyatı {fiyat:,.2f} TL, stok: {stok_adedi}"  # ,.2f sayıyı 2 basamaklı biçimler

print(mesaj)
```

```python
deger = "125"

if deger.isdigit():                 # isdigit() metnin sadece rakamlardan oluşup oluşmadığını kontrol eder
    sayi = int(deger)
    print(sayi * 2)
else:
    print("Sayısal olmayan değer var.")
```

## İleri Seviye

Bu seviyede Python'ın doğruluk değeri mantığını ve hassas sayı kullanımını ele alacağız.

Python'da bazı değerler koşul içinde otomatik olarak `False` kabul edilir. Boş liste, boş metin, `0`, `None` gibi değerler buna örnektir. Bu davranış pratik olsa da neyi kontrol ettiğimizi bilmek gerekir.

Para gibi hassas hesaplarda `float` küçük yuvarlama farkları oluşturabilir. Bu yüzden finansal hesaplarda `Decimal` kullanmak daha doğru olabilir.


```python
kullanici_adi = ""

if kullanici_adi:                  # Boş metin False kabul edilir
    print("Kullanıcı adı var.")
else:
    print("Kullanıcı adı boş.")

sepet = []

if not sepet:                      # Boş liste False kabul edildiği için not ile kontrol edilir
    print("Sepet boş.")
```

```python
from decimal import Decimal

fiyat = Decimal("19.99")           # Para hesaplarında metinden Decimal oluşturmak daha güvenlidir
adet = Decimal("3")
kdv_orani = Decimal("0.20")

ara_toplam = fiyat * adet
genel_toplam = ara_toplam + (ara_toplam * kdv_orani)

print(genel_toplam)
```

## Mini Alıştırma ve Özet

Bu bölümde veri türlerinin program davranışını doğrudan etkilediğini gördük.

Hatırlanması gerekenler:

* Sayı gibi görünen her değer sayı değildir.
* `type()` kontrol için kullanılır.
* `int()`, `float()`, `str()` tür dönüşümü yapar.
* `None`, değer yokluğunu ifade eder.
* Veri bilimi çalışmalarında analizden önce tür kontrolü yapılmalıdır.

Şimdi ham değerleri doğru türlere dönüştüren küçük bir örnek çözelim.


```python
yas_raw = "34"
skor_raw = "72.5"
aktif_raw = "True"

yas = int(yas_raw)
skor = float(skor_raw)
aktif_mi = aktif_raw == "True"      # Metin tam olarak "True" ise bool değer True olur

print(yas, type(yas))
print(skor, type(skor))
print(aktif_mi, type(aktif_mi))
```

# Pandas ile Veri Temizleme

Gerçek veri setlerinde eksik değerler, tekrar eden satırlar, hatalı türler ve yazım farklılıkları çok sık görülür. Bu notebook'ta veri temizleme sürecinin temel adımlarını öğreneceğiz.


## Konu Dokümantasyonu

Veri temizleme, analizden önce veriyi güvenilir hâle getirme sürecidir.

Sık yapılan işlemler:

* Eksik değer kontrolü
* Eksik değer doldurma veya silme
* Tekrarlı kayıtları kaldırma
* Veri türü dönüştürme
* Metin temizleme
* Hatalı değerleri işaretleme

Veri bilimi projelerinde çoğu zaman model kurmadan önce en fazla vakit veri temizlemeye harcanır.


## Kolay Seviye

Eksik değerleri görmek için `isna()` kullanılır. Sütun bazında eksik sayısını bulmak için `sum()` ile birlikte kullanılır.


```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "musteri": ["Ali", "Ayşe", "Mehmet", "Zeynep"],
    "yas": [25, np.nan, 34, 29],
    "gelir": [15000, 22000, np.nan, 18000],
})

print(df)
print(df.isna())
print(df.isna().sum())
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "musteri": ["Ali", "Ayşe", "Mehmet", "Zeynep"],
    "yas": [25, np.nan, 34, 29],
    "gelir": [15000, 22000, np.nan, 18000],
})

df["yas"] = df["yas"].fillna(df["yas"].median())
df["gelir"] = df["gelir"].fillna(df["gelir"].mean())
# Yaş medyanla, gelir ortalamayla dolduruldu

print(df)
```

## Orta Seviye

Tekrarlı kayıtlar analiz sonuçlarını bozabilir. `duplicated()` tekrar eden satırları gösterir, `drop_duplicates()` tekrarları kaldırır.

Veri türü dönüşümü için `astype()` ve `pd.to_numeric()` kullanılır.


```python
import pandas as pd

df = pd.DataFrame({
    "siparis_id": [1, 2, 2, 3],
    "musteri": ["Ali", "Ayşe", "Ayşe", "Mehmet"],
    "tutar": [1200, 950, 950, 1800],
})

print(df.duplicated())

temiz_df = df.drop_duplicates()
print(temiz_df)
```

```python
import pandas as pd

df = pd.DataFrame({
    "musteri": ["Ali", "Ayşe", "Mehmet"],
    "tutar": ["1200", "950", "hatalı"],
})

df["tutar_sayisal"] = pd.to_numeric(df["tutar"], errors="coerce")
# Sayıya çevrilemeyen değerler NaN olur

print(df)
```

## İleri Seviye

Metin sütunlarında boşluk, büyük/küçük harf farkı ve yazım tutarsızlıkları olabilir. `str` erişicisi metin temizleme için kullanılır.

Kategorik değişkenleri standartlaştırmak, groupby ve modelleme adımlarında daha doğru sonuç verir.


```python
import pandas as pd

df = pd.DataFrame({
    "sehir": [" İstanbul ", "istanbul", "ANKARA", "İzmir ", "ankara"],
    "satis": [1200, 950, 1800, 2100, 1750],
})

df["sehir_temiz"] = (
    df["sehir"]
    .str.strip()       # Baştaki ve sondaki boşlukları siler
    .str.lower()       # Küçük harfe çevirir
)

print(df)
```

## Mini Veri Bilimi Uygulaması

Küçük bir müşteri veri setini temizleyelim: tekrarları kaldıralım, tutarı sayıya çevirelim, eksikleri dolduralım ve şehir adlarını standartlaştıralım.


```python
import pandas as pd

df = pd.DataFrame({
    "musteri_id": [1, 2, 2, 3, 4],
    "sehir": [" İstanbul ", "ANKARA", "ANKARA", "İzmir", "istanbul"],
    "tutar": ["1200", "950", "950", "hatalı", "1800"],
})

df = df.drop_duplicates()
df["sehir"] = df["sehir"].str.strip().str.lower()
df["tutar"] = pd.to_numeric(df["tutar"], errors="coerce")
df["tutar"] = df["tutar"].fillna(df["tutar"].median())

print(df)
```

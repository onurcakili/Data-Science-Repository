# NumPy Operasyonları ve Broadcasting

Bu notebook'ta array'ler üzerinde matematiksel işlemleri, eksen mantığını ve broadcasting kavramını öğreneceğiz.

Broadcasting, farklı şekillerdeki array'lerin uyumlu şekilde işlem yapabilmesini sağlar. Veri biliminde normalize etme, standartlaştırma ve matris hesaplamalarında sık karşımıza çıkar.


## Konu Dokümantasyonu

NumPy array'leri üzerinde aritmetik işlemler eleman bazlı yapılır. Yani iki array toplanırsa aynı konumdaki elemanlar toplanır.

`axis` kavramı çok önemlidir:

* `axis=0`: Sütunlar boyunca işlem yapar.
* `axis=1`: Satırlar boyunca işlem yapar.

Broadcasting ise daha küçük bir array'in daha büyük array'e uygun şekilde genişletilmiş gibi davranmasıdır. Bu işlem gerçek kopyalama yapmadan gerçekleşir.


## Kolay Seviye

Array üzerinde dört işlem doğrudan yapılabilir.


```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a + b)       # Aynı konumdaki elemanlar toplanır
print(a - b)
print(a * b)
print(a / b)
```

```python
import numpy as np

fiyatlar = np.array([100, 250, 400])

print(fiyatlar + 10)      # Her elemana 10 eklenir
print(fiyatlar * 1.20)    # Her eleman 1.20 ile çarpılır
```

## Orta Seviye

İki boyutlu array'lerde satır ve sütun bazlı hesaplama için `axis` kullanılır.

Bu mantık veri biliminde satır bazlı müşteri toplamı veya sütun bazlı ürün toplamı hesaplamak için kullanılır.


```python
import numpy as np

satis_matrisi = np.array([
    [1200, 950, 1800],
    [2100, 1750, 1600],
    [900, 1100, 1300],
])

urun_toplamlari = satis_matrisi.sum(axis=1)     # Her satırın toplamı
gun_toplamlari = satis_matrisi.sum(axis=0)      # Her sütunun toplamı

print("Ürün toplamları:", urun_toplamlari)
print("Gün toplamları:", gun_toplamlari)
```

```python
import numpy as np

veri = np.array([
    [10, 200],
    [20, 300],
    [30, 400],
])

sutun_ortalamalari = veri.mean(axis=0)
merkezlenmis_veri = veri - sutun_ortalamalari
# Broadcasting: sutun_ortalamalari her satırdan çıkarılır

print(merkezlenmis_veri)
```

## İleri Seviye

Standartlaştırma, veriyi ortalaması 0 ve standart sapması 1 olacak şekilde dönüştürmektir. Makine öğrenmesi modellerinde sık kullanılan bir ön işleme adımıdır.

NumPy ile bu işlem broadcasting sayesinde birkaç satırda yapılabilir.


```python
import numpy as np

veri = np.array([
    [18, 5000],
    [25, 8500],
    [40, 15000],
    [32, 12000],
], dtype=float)

ortalama = veri.mean(axis=0)
standart_sapma = veri.std(axis=0)

standart_veri = (veri - ortalama) / standart_sapma
# Her sütun kendi ortalaması ve standart sapması ile dönüştürülür

print(standart_veri)
print("Yeni ortalama:", standart_veri.mean(axis=0))
print("Yeni std:", standart_veri.std(axis=0))
```

## Mini Veri Bilimi Uygulaması

Müşteri özelliklerini normalize edelim. Bu işlem farklı ölçeklerdeki değişkenleri karşılaştırılabilir hâle getirir.


```python
import numpy as np

musteri_verisi = np.array([
    [22, 1200],
    [35, 4500],
    [28, 3200],
    [46, 9000],
], dtype=float)

minimum = musteri_verisi.min(axis=0)
maksimum = musteri_verisi.max(axis=0)

normalize_veri = (musteri_verisi - minimum) / (maksimum - minimum)
# Min-max normalizasyonu: her sütunu 0 ile 1 arasına taşır

print(normalize_veri)
```

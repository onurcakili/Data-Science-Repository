# NumPy Indexing, Slicing ve Filtreleme

NumPy array içindeki verilere erişmek, veri analizinin en temel adımlarından biridir. Bu notebook'ta indeksleme, slicing, boolean filtreleme ve koşullu seçim konularını öğreneceğiz.

Öğrenme akışı:

* **Kolay Seviye:** Tek boyutlu array indeksleme
* **Orta Seviye:** İki boyutlu array ve slicing
* **İleri Seviye:** Boolean mask, `np.where`, fancy indexing


## Konu Dokümantasyonu

Indexing, array içindeki belirli bir elemana erişmektir. Slicing ise array'in belirli bir bölümünü almaktır.

Python'da olduğu gibi NumPy'de de indeksler `0` ile başlar. Bitiş indeksi slicing işlemlerinde dahil edilmez.

Veri biliminde indexing ve filtering şu işler için kullanılır:

* Belirli satırları veya sütunları seçmek
* Belirli koşulu sağlayan değerleri almak
* Aykırı değerleri tespit etmek
* Analiz için alt veri kümesi oluşturmak


## Kolay Seviye

Tek boyutlu array'lerde indeksleme liste mantığına benzer.


````python
import numpy as np

notlar = np.array([70, 85, 90, 60, 95])

print(notlar[0])       # İlk eleman
print(notlar[2])       # Üçüncü eleman
print(notlar[-1])      # Son eleman
print(notlar[1:4])     # 1 dahil, 4 dahil değil
````

## Orta Seviye

İki boyutlu array'lerde satır ve sütun birlikte seçilir.

Genel kullanım:

```python
array[satır, sütun]
```

Slicing yaparken satır ve sütun aralıkları ayrı ayrı yazılabilir.


````python
import numpy as np

satislar = np.array([
    [1200, 950, 1800],
    [2100, 1750, 1600],
    [900, 1100, 1300],
])

print(satislar[0, 0])       # 1. satır, 1. sütun
print(satislar[1])          # 2. satırın tamamı
print(satislar[:, 0])       # Tüm satırlar, 1. sütun
print(satislar[:2, 1:])     # İlk 2 satır, 2. sütundan sonrası
````

````python
import numpy as np

satislar = np.array([
    [1200, 950, 1800],
    [2100, 1750, 1600],
    [900, 1100, 1300],
])

ilk_iki_gun = satislar[:, :2]        # Tüm ürünler için ilk iki gün
son_gun = satislar[:, -1]            # Tüm ürünler için son gün

print("İlk iki gün:")
print(ilk_iki_gun)
print("Son gün:", son_gun)
````

## İleri Seviye

Boolean mask, koşulu sağlayan elemanları seçmek için kullanılır. Bu yapı veri biliminde çok sık kullanılır.

`np.where()` ise koşula göre farklı değer üretmek için kullanılır. Örneğin satış hedefi geçen değerleri `"başarılı"`, geçemeyenleri `"düşük"` olarak etiketleyebiliriz.


````python
import numpy as np

satislar = np.array([1200, 950, 1800, 2100, 1750])

maske = satislar >= 1500             # Her eleman için True/False üretir
yuksek_satislar = satislar[maske]    # Sadece True olan değerler seçilir

print(maske)
print(yuksek_satislar)
````

````python
import numpy as np

satislar = np.array([1200, 950, 1800, 2100, 1750])

etiketler = np.where(satislar >= 1500, "yüksek", "normal")
# Koşul True ise "yüksek", False ise "normal" yazılır

print(etiketler)
````

## Mini Veri Bilimi Uygulaması

Bir satış array'inde hedef altındaki değerleri bulalım ve hedefe göre farklarını hesaplayalım.


````python
import numpy as np

satislar = np.array([1200, 950, 1800, 2100, 1750])
hedef = 1500

hedef_alti = satislar[satislar < hedef]
farklar = hedef - hedef_alti

print("Hedef altı satışlar:", hedef_alti)
print("Hedefe ulaşmak için gereken farklar:", farklar)
````
